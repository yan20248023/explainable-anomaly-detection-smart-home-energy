```python
# Cell 9 — Audit Canonical NAB and BLUED Artifacts:
# Verify the reusable saved inputs, manifests, split counts, tensor shapes,
# and original model metrics before starting any new baseline or multi-seed experiments.

import json
import numpy as np
import pandas as pd
from pathlib import Path

# ---------------------------------------------------------------------
# Canonical experiment roots
# ---------------------------------------------------------------------

NAB_ROOT = CODE_ROOT / "experiments" / "EXP_NAB_OCSVM_LSTM_Autoencoder"
BLUED_ROOT = CODE_ROOT / "experiments" / "EXP_BLUED_FIVE_MODEL_LABEL_BLIND"

# ---------------------------------------------------------------------
# Key artifact paths
# ---------------------------------------------------------------------

nab_paths = {
    "sequence_tensors": NAB_ROOT / "common" / "nab_lstm_sequence_tensors.npz",
    "split_manifest": NAB_ROOT / "common" / "nab_split_manifest.csv",
    "dense_metrics": NAB_ROOT / "Controlled_Five_Model_Comparison" / "dense_autoencoder_metrics.csv",
    "lstm_metrics": NAB_ROOT / "LSTM_Autoencoder" / "metrics.csv",
    "lstm_config": NAB_ROOT / "LSTM_Autoencoder" / "config.json",
}

blued_paths = {
    "normalized_inputs": BLUED_ROOT / "common" / "blued_normalized_model_inputs.npz",
    "model_manifest": BLUED_ROOT / "common" / "blued_1min_stride30s_model_manifest.csv",
    "split_summary": BLUED_ROOT / "common" / "blued_1min_split_summary.csv",
    "dense_metrics": BLUED_ROOT / "dense_autoencoder" / "metrics.csv",
    "lstm_metrics": BLUED_ROOT / "lstm_autoencoder" / "metrics.csv",
}

# ---------------------------------------------------------------------
# Helper functions
# ---------------------------------------------------------------------

def print_section(title):
    print("\n" + "=" * 120)
    print(title)
    print("=" * 120)


def inspect_npz(path):
    print(f"\nNPZ: {path}")

    if not path.exists():
        print("❌ Missing")
        return

    with np.load(path, allow_pickle=False) as data:
        print("Keys:")
        for key in data.files:
            arr = data[key]
            print(
                f"  {key:<30} "
                f"shape={str(arr.shape):<22} "
                f"dtype={arr.dtype}"
            )


def inspect_csv(path, name, show_columns=True):
    print(f"\n{name}: {path}")

    if not path.exists():
        print("❌ Missing")
        return None

    df = pd.read_csv(path)

    print(f"Rows:    {len(df):,}")
    print(f"Columns: {len(df.columns):,}")

    if show_columns:
        print("Column names:")
        print(list(df.columns))

    return df


def inspect_metrics(path, name):
    print(f"\n{name}: {path}")

    if not path.exists():
        print("❌ Missing")
        return

    df = pd.read_csv(path)
    print(df.to_string(index=False))


# =====================================================================
# 1. Artifact existence audit
# =====================================================================

print_section("1. ARTIFACT EXISTENCE AUDIT")

for dataset_name, path_dict in [
    ("NAB", nab_paths),
    ("BLUED", blued_paths),
]:
    print(f"\n{dataset_name}")

    for label, path in path_dict.items():
        status = "✅" if path.exists() else "❌"
        print(f"{status} {label:<20} -> {path.name}")


# =====================================================================
# 2. NAB tensor and manifest audit
# =====================================================================

print_section("2. NAB CANONICAL INPUT AUDIT")

inspect_npz(nab_paths["sequence_tensors"])

nab_manifest = inspect_csv(
    nab_paths["split_manifest"],
    "NAB split manifest"
)

if nab_manifest is not None:
    # Automatically inspect likely split columns.
    candidate_split_cols = [
        c for c in nab_manifest.columns
        if any(word in c.lower() for word in ["split", "partition", "set"])
    ]

    if candidate_split_cols:
        print("\nPossible split distributions:")

        for col in candidate_split_cols:
            print(f"\nColumn: {col}")
            print(
                nab_manifest[col]
                .value_counts(dropna=False)
                .to_string()
            )


# =====================================================================
# 3. NAB original stochastic-model metrics
# =====================================================================

print_section("3. NAB ORIGINAL STOCHASTIC-MODEL METRICS")

inspect_metrics(
    nab_paths["dense_metrics"],
    "NAB Dense Autoencoder metrics"
)

inspect_metrics(
    nab_paths["lstm_metrics"],
    "NAB LSTM Autoencoder metrics"
)

if nab_paths["lstm_config"].exists():
    print("\nNAB LSTM configuration:")

    with open(nab_paths["lstm_config"], "r") as f:
        config = json.load(f)

    print(json.dumps(config, indent=2))


# =====================================================================
# 4. BLUED tensor and manifest audit
# =====================================================================

print_section("4. BLUED CANONICAL INPUT AUDIT")

inspect_npz(blued_paths["normalized_inputs"])

blued_manifest = inspect_csv(
    blued_paths["model_manifest"],
    "BLUED model manifest"
)

blued_split_summary = inspect_csv(
    blued_paths["split_summary"],
    "BLUED split summary"
)

if blued_manifest is not None:
    candidate_split_cols = [
        c for c in blued_manifest.columns
        if any(word in c.lower() for word in ["split", "partition", "set"])
    ]

    if candidate_split_cols:
        print("\nPossible split distributions:")

        for col in candidate_split_cols:
            print(f"\nColumn: {col}")
            print(
                blued_manifest[col]
                .value_counts(dropna=False)
                .to_string()
            )


# =====================================================================
# 5. BLUED original stochastic-model metrics
# =====================================================================

print_section("5. BLUED ORIGINAL STOCHASTIC-MODEL METRICS")

inspect_metrics(
    blued_paths["dense_metrics"],
    "BLUED Dense Autoencoder metrics"
)

inspect_metrics(
    blued_paths["lstm_metrics"],
    "BLUED LSTM Autoencoder metrics"
)


# =====================================================================
# Final audit statement
# =====================================================================

print_section("AUDIT COMPLETE")

print(
    "No retraining has been performed. "
    "This cell only inspected the saved canonical artifacts."
)
```

```python
# Cell 11 — Verify Final Corrected NAB Neural Artifacts:
# Load the authoritative post-purge Dense AE and LSTM-AE inputs, metadata, and metrics,
# and verify that they represent the final leakage-controlled NAB experiment.

import json
import numpy as np
import pandas as pd
from pathlib import Path

NAB_CORRECTION_ROOT = (
    CODE_ROOT
    / "experiments"
    / "EXP_SPLIT_BOUNDARY_PURGE_CORRECTION"
    / "NAB"
    / "FIVE_MODEL_CORRECTION"
)

corrected_nab_paths = {
    "dense_inputs": NAB_CORRECTION_ROOT / "dense_autoencoder_corrected_inputs.npz",
    "dense_metrics": NAB_CORRECTION_ROOT / "dense_autoencoder_metrics_corrected.csv",
    "dense_metadata": NAB_CORRECTION_ROOT / "dense_autoencoder_correction_metadata.json",

    "lstm_inputs": NAB_CORRECTION_ROOT / "lstm_autoencoder_corrected_inputs.npz",
    "lstm_metrics": NAB_CORRECTION_ROOT / "lstm_autoencoder_metrics_corrected.csv",
    "lstm_metadata": NAB_CORRECTION_ROOT / "lstm_autoencoder_correction_metadata.json",

    "publication_table": NAB_CORRECTION_ROOT / "FINAL_NAB_FIVE_MODEL_PUBLICATION_TABLE_CORRECTED.csv",
    "provenance": NAB_CORRECTION_ROOT / "FINAL_NAB_FIVE_MODEL_CORRECTION_PROVENANCE.csv",
}

def audit_npz(path, label):
    print("\n" + "=" * 120)
    print(label)
    print("=" * 120)
    print(f"Path: {path}")

    if not path.exists():
        print("❌ Missing")
        return

    with np.load(path, allow_pickle=False) as data:
        print("\nKeys and shapes:")

        for key in data.files:
            arr = data[key]

            print(
                f"{key:<35}"
                f"shape={str(arr.shape):<24}"
                f"dtype={arr.dtype}"
            )

        # Print basic class distributions automatically if label-like arrays exist.
        label_keys = [
            key for key in data.files
            if key.lower().startswith("y_")
            or "label" in key.lower()
        ]

        if label_keys:
            print("\nLabel distributions:")

            for key in label_keys:
                arr = data[key]

                if arr.ndim == 1:
                    values, counts = np.unique(arr, return_counts=True)

                    print(f"\n{key}:")
                    for value, count in zip(values, counts):
                        print(f"  {value}: {count:,}")


def audit_csv(path, label):
    print("\n" + "=" * 120)
    print(label)
    print("=" * 120)
    print(f"Path: {path}")

    if not path.exists():
        print("❌ Missing")
        return

    df = pd.read_csv(path)

    print(f"Rows: {len(df):,}")
    print(f"Columns: {len(df.columns):,}")
    print("\nColumns:")
    print(list(df.columns))

    print("\nContents:")
    print(df.to_string(index=False))


def audit_json(path, label):
    print("\n" + "=" * 120)
    print(label)
    print("=" * 120)
    print(f"Path: {path}")

    if not path.exists():
        print("❌ Missing")
        return

    with open(path, "r") as f:
        data = json.load(f)

    print(json.dumps(data, indent=2))


# ---------------------------------------------------------------------
# 1. Corrected Dense Autoencoder inputs
# ---------------------------------------------------------------------

audit_npz(
    corrected_nab_paths["dense_inputs"],
    "1. CORRECTED NAB DENSE AUTOENCODER INPUTS"
)


# ---------------------------------------------------------------------
# 2. Corrected LSTM Autoencoder inputs
# ---------------------------------------------------------------------

audit_npz(
    corrected_nab_paths["lstm_inputs"],
    "2. CORRECTED NAB LSTM AUTOENCODER INPUTS"
)


# ---------------------------------------------------------------------
# 3. Corrected metrics
# ---------------------------------------------------------------------

audit_csv(
    corrected_nab_paths["dense_metrics"],
    "3. CORRECTED NAB DENSE AUTOENCODER METRICS"
)

audit_csv(
    corrected_nab_paths["lstm_metrics"],
    "4. CORRECTED NAB LSTM AUTOENCODER METRICS"
)


# ---------------------------------------------------------------------
# 4. Correction metadata
# ---------------------------------------------------------------------

audit_json(
    corrected_nab_paths["dense_metadata"],
    "5. DENSE AUTOENCODER CORRECTION METADATA"
)

audit_json(
    corrected_nab_paths["lstm_metadata"],
    "6. LSTM AUTOENCODER CORRECTION METADATA"
)


# ---------------------------------------------------------------------
# 5. Final five-model publication table and provenance
# ---------------------------------------------------------------------

audit_csv(
    corrected_nab_paths["publication_table"],
    "7. FINAL CORRECTED NAB FIVE-MODEL PUBLICATION TABLE"
)

audit_csv(
    corrected_nab_paths["provenance"],
    "8. FINAL NAB CORRECTION PROVENANCE"
)


# ---------------------------------------------------------------------
# Final verification statement
# ---------------------------------------------------------------------

print("\n" + "=" * 120)
print("FINAL CHECK")
print("=" * 120)

print(
    "Expected final leakage-controlled NAB population:\n"
    "  Train      = 62,096\n"
    "  Validation = 20,648\n"
    "  Test       = 20,668\n"
    "\nNo models were retrained in this cell."
)
```

```python
# Cell 15 — Safe BLUED Training-Code Provenance Search:
# Search only thesis experiment/code/log and Colab-notebook locations for the original
# BLUED Dense/LSTM Autoencoder implementation while safely skipping datasets and broken Drive entries.

import os
from pathlib import Path


# ---------------------------------------------------------------------
# 1. Search only locations where implementation code could realistically live
# ---------------------------------------------------------------------

SAFE_SEARCH_ROOTS = [
    CODE_ROOT / "experiments",
    CODE_ROOT / "experiment_logs",
    CODE_ROOT / "exploratory",
    CODE_ROOT / "documentation",
    CODE_ROOT / "src",
    DRIVE_ROOT / "Colab Notebooks",
]

SEARCH_TERMS = [
    "BLUED_Dense_Autoencoder",
    "BLUED_LSTM_Autoencoder",
    "ReduceLROnPlateau",
    "EarlyStopping",
    "encoder_48",
    "encoder_lstm_32",
    "latent_lstm_16",
    "dense_autoencoder",
    "lstm_autoencoder",
]

SEARCH_EXTENSIONS = {
    ".py",
    ".ipynb",
    ".txt",
    ".md",
    ".json",
    ".log",
}

MAX_FILE_SIZE_MB = 25

matches = []
skipped_errors = []


# ---------------------------------------------------------------------
# 2. Safe recursive search using os.walk
# ---------------------------------------------------------------------

print("=" * 130)
print("SAFE BLUED TRAINING-CODE PROVENANCE SEARCH")
print("=" * 130)

for root in SAFE_SEARCH_ROOTS:

    print(f"\nSearching: {root}")

    if not root.exists():
        print("  ⚠️ Root does not exist.")
        continue

    def handle_walk_error(error):
        skipped_errors.append(str(error))

    try:
        walker = os.walk(
            root,
            topdown=True,
            onerror=handle_walk_error
        )

        for current_dir, dirnames, filenames in walker:

            # ---------------------------------------------------------
            # Prevent accidental traversal into irrelevant/heavy folders
            # ---------------------------------------------------------

            dirnames[:] = [
                dirname
                for dirname in dirnames
                if dirname not in {
                    "Datasets",
                    "datasets",
                    "processed_data",
                    "__pycache__",
                    ".git",
                    ".ipynb_checkpoints",
                }
            ]

            for filename in filenames:

                path = Path(current_dir) / filename

                if path.suffix.lower() not in SEARCH_EXTENSIONS:
                    continue

                try:
                    file_size_mb = path.stat().st_size / (1024 ** 2)

                except OSError as exc:
                    skipped_errors.append(
                        f"{path}: {exc}"
                    )
                    continue

                if file_size_mb > MAX_FILE_SIZE_MB:
                    continue

                try:
                    text_content = path.read_text(
                        encoding="utf-8",
                        errors="ignore"
                    )

                except (OSError, UnicodeError) as exc:
                    skipped_errors.append(
                        f"{path}: {exc}"
                    )
                    continue

                matched_terms = [
                    term
                    for term in SEARCH_TERMS
                    if term.lower() in text_content.lower()
                ]

                if matched_terms:

                    matches.append({
                        "path": path,
                        "matched_terms": matched_terms,
                        "size_kb": path.stat().st_size / 1024,
                        "content": text_content,
                    })

    except OSError as exc:
        skipped_errors.append(
            f"{root}: {exc}"
        )


# ---------------------------------------------------------------------
# 3. Print matching files
# ---------------------------------------------------------------------

print("\n" + "=" * 130)
print("MATCHING FILES")
print("=" * 130)

if not matches:
    print("No matching implementation files were found.")

else:

    for i, match in enumerate(matches, start=1):

        print(f"\n[{i}] {match['path']}")
        print(f"    Size: {match['size_kb']:.2f} KB")
        print(f"    Matched terms: {match['matched_terms']}")


# ---------------------------------------------------------------------
# 4. Extract focused snippets around callback/model definitions
# ---------------------------------------------------------------------

print("\n" + "=" * 130)
print("RELEVANT SOURCE SNIPPETS")
print("=" * 130)

SNIPPET_PATTERNS = [
    "ReduceLROnPlateau",
    "EarlyStopping",
    "BLUED_Dense_Autoencoder",
    "BLUED_LSTM_Autoencoder",
    "encoder_48",
    "encoder_lstm_32",
]

for match in matches:

    path = match["path"]
    lines = match["content"].splitlines()

    relevant_indices = [
        idx
        for idx, line in enumerate(lines)
        if any(
            pattern.lower() in line.lower()
            for pattern in SNIPPET_PATTERNS
        )
    ]

    if not relevant_indices:
        continue

    print("\n" + "-" * 130)
    print(path)
    print("-" * 130)

    printed_ranges = []

    for idx in relevant_indices:

        start = max(0, idx - 10)
        end = min(len(lines), idx + 16)

        overlaps_existing = any(
            start <= previous_end and end >= previous_start
            for previous_start, previous_end in printed_ranges
        )

        if overlaps_existing:
            continue

        printed_ranges.append((start, end))

        print(f"\nLines {start + 1}–{end}:")

        for line_no in range(start, end):

            line = lines[line_no]

            # Keep notebook JSON lines readable.
            if len(line) > 600:
                line = line[:600] + " ..."

            print(
                f"{line_no + 1:>6}: {line}"
            )


# ---------------------------------------------------------------------
# 5. Report safely skipped Drive errors
# ---------------------------------------------------------------------

print("\n" + "=" * 130)
print("SEARCH SUMMARY")
print("=" * 130)

print(f"Matching files found: {len(matches)}")
print(f"Drive/file errors safely skipped: {len(skipped_errors)}")

if skipped_errors:

    print("\nFirst 10 safely skipped errors:")

    for error in skipped_errors[:10]:
        print(f"  - {error}")

print(
    "\nNo model training was performed in Cell 15. "
    "Dataset directories were intentionally excluded from this search."
)
```

```python
# Cell 16 — Extract Original BLUED Neural Training Cells:
# Parse the historical thesis notebooks as JSON and print the BLUED Dense/LSTM
# Autoencoder model-building, callback, compile, and fit cells with nearby context.

import json
from pathlib import Path


# ---------------------------------------------------------------------
# 1. Historical notebooks most likely to contain the authoritative code
# ---------------------------------------------------------------------

NOTEBOOK_PATHS = [
    DRIVE_ROOT / "Colab Notebooks" / "Thesis.ipynb",
    DRIVE_ROOT / "Colab Notebooks" / "Thesis_Split_Boundary_Correction.ipynb",
]


# ---------------------------------------------------------------------
# 2. Strong anchors for the BLUED neural experiments
# ---------------------------------------------------------------------

PRIMARY_ANCHORS = [
    "BLUED_Dense_Autoencoder",
    "BLUED_LSTM_Autoencoder",
    "encoder_48",
    "encoder_lstm_32",
    "latent_lstm_16",
]

SUPPORTING_TERMS = [
    "ReduceLROnPlateau",
    "EarlyStopping",
    "Adam",
    "learning_rate",
    "batch_size",
    "epochs",
    ".fit(",
    "model.fit(",
    "dense_autoencoder",
    "lstm_autoencoder",
]


# ---------------------------------------------------------------------
# 3. Helper to normalize notebook cell source
# ---------------------------------------------------------------------

def get_cell_source(cell):
    source = cell.get("source", [])

    if isinstance(source, list):
        return "".join(source)

    return str(source)


# ---------------------------------------------------------------------
# 4. Parse notebooks and identify matching BLUED cells
# ---------------------------------------------------------------------

for notebook_path in NOTEBOOK_PATHS:

    print("\n" + "=" * 140)
    print(f"NOTEBOOK: {notebook_path}")
    print("=" * 140)

    if not notebook_path.exists():
        print("❌ Notebook not found.")
        continue

    try:
        with open(
            notebook_path,
            "r",
            encoding="utf-8",
            errors="ignore"
        ) as f:
            notebook = json.load(f)

    except Exception as exc:
        print(f"❌ Could not parse notebook: {exc}")
        continue

    cells = notebook.get("cells", [])

    code_cell_indices = [
        i
        for i, cell in enumerate(cells)
        if cell.get("cell_type") == "code"
    ]

    primary_matches = []

    for i in code_cell_indices:

        source = get_cell_source(cells[i])
        source_lower = source.lower()

        matched_anchors = [
            anchor
            for anchor in PRIMARY_ANCHORS
            if anchor.lower() in source_lower
        ]

        if matched_anchors:
            primary_matches.append({
                "index": i,
                "matched_anchors": matched_anchors,
            })

    print(f"\nPrimary BLUED neural cells found: {len(primary_matches)}")

    if not primary_matches:
        print("No primary BLUED neural architecture cells found.")
        continue


    # -----------------------------------------------------------------
    # 5. Include nearby CODE cells for callbacks / fit logic
    # -----------------------------------------------------------------

    cells_to_print = set()

    for match in primary_matches:

        matched_index = match["index"]

        code_position = code_cell_indices.index(matched_index)

        # Include up to 3 code cells before and after the matching cell.
        start_pos = max(0, code_position - 3)
        end_pos = min(len(code_cell_indices), code_position + 4)

        for pos in range(start_pos, end_pos):
            cells_to_print.add(code_cell_indices[pos])


    # -----------------------------------------------------------------
    # 6. Also include any BLUED-related callback / fit cells
    # -----------------------------------------------------------------

    for i in code_cell_indices:

        source = get_cell_source(cells[i])
        source_lower = source.lower()

        has_blued_context = (
            "blued" in source_lower
            or "dense_autoencoder" in source_lower
            or "lstm_autoencoder" in source_lower
        )

        has_training_term = any(
            term.lower() in source_lower
            for term in SUPPORTING_TERMS
        )

        if has_blued_context and has_training_term:
            cells_to_print.add(i)


    # -----------------------------------------------------------------
    # 7. Print selected cells with focused metadata
    # -----------------------------------------------------------------

    print("\nSelected historical code cells:")
    print(sorted(cells_to_print))

    for i in sorted(cells_to_print):

        source = get_cell_source(cells[i])

        print("\n" + "-" * 140)
        print(f"NOTEBOOK CELL INDEX: {i}")
        print("-" * 140)

        execution_count = cells[i].get("execution_count")
        print(f"Execution count: {execution_count}")

        matched_terms = [
            term
            for term in PRIMARY_ANCHORS + SUPPORTING_TERMS
            if term.lower() in source.lower()
        ]

        print(f"Matched terms: {matched_terms}")
        print("\nSOURCE:\n")

        # Avoid catastrophic output if an unrelated giant cell is encountered.
        if len(source) > 12000:
            print(source[:12000])
            print("\n...[CELL TRUNCATED AFTER 12,000 CHARACTERS]...")
        else:
            print(source)


# ---------------------------------------------------------------------
# 8. Completion statement
# ---------------------------------------------------------------------

print("\n" + "=" * 140)
print("NOTEBOOK EXTRACTION COMPLETE")
print("=" * 140)

print(
    "No model training was performed. "
    "This cell only parsed historical notebook source code."
)
```

```python
# Cell 18 — Freeze Neural Training Specifications and Define Reusable Functions:
# Encode the historically verified NAB and BLUED Dense/LSTM Autoencoder settings,
# save them as a permanent experiment specification, and define deterministic
# model-building, callback, scoring, and evaluation utilities for the five-seed runs.

import gc
import json
import random
import time

import numpy as np
import pandas as pd
import tensorflow as tf

from sklearn.metrics import (
    confusion_matrix,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score,
)

from tensorflow.keras import Model, Input
from tensorflow.keras.layers import (
    Dense,
    LSTM,
    RepeatVector,
    TimeDistributed,
)
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import (
    EarlyStopping,
    ReduceLROnPlateau,
)


# ---------------------------------------------------------------------
# 1. Freeze the historically verified neural training specifications
# ---------------------------------------------------------------------

NEURAL_SPECS = {
    "NAB": {
        "Dense Autoencoder": {
            "input_shape": [23],
            "architecture": [23, 16, 8, 4, 8, 16, 23],
            "hidden_activation": "relu",
            "output_activation": "linear",
            "optimizer": "Adam",
            "initial_learning_rate": 0.001,
            "loss": "mse",
            "maximum_epochs": 150,
            "batch_size": 256,
            "shuffle": True,
            "early_stopping": {
                "monitor": "val_loss",
                "patience": 10,
                "min_delta": 1e-5,
                "restore_best_weights": True,
            },
            "reduce_lr": None,
            "threshold_percentile": 95.0,
            "score_reduction_axes": [1],
        },

        "LSTM Autoencoder": {
            "input_shape": [60, 1],
            "encoder_units": [32, 16],
            "decoder_units": [16, 32],
            "activation": "tanh",
            "output_activation": "linear",
            "optimizer": "Adam",
            "initial_learning_rate": 0.001,
            "loss": "mse",
            "maximum_epochs": 100,
            "batch_size": 256,
            "shuffle": True,
            "early_stopping": {
                "monitor": "val_loss",
                "patience": 10,
                "min_delta": 1e-5,
                "restore_best_weights": True,
            },
            "reduce_lr": {
                "monitor": "val_loss",
                "factor": 0.5,
                "patience": 5,
                "min_lr": 1e-6,
                "min_delta": 1e-4,
            },
            "threshold_percentile": 95.0,
            "score_reduction_axes": [1, 2],
        },
    },

    "BLUED": {
        "Dense Autoencoder": {
            "input_shape": [69],
            "architecture": [69, 48, 24, 8, 24, 48, 69],
            "hidden_activation": "relu",
            "output_activation": "linear",
            "optimizer": "Adam",
            "initial_learning_rate": 0.001,
            "loss": "mse",
            "maximum_epochs": 150,
            "batch_size": 256,
            "shuffle": True,
            "early_stopping": {
                "monitor": "val_loss",
                "patience": 10,
                "min_delta": 0.0,
                "restore_best_weights": True,
            },
            "reduce_lr": {
                "monitor": "val_loss",
                "factor": 0.5,
                "patience": 5,
                "min_lr": 1e-6,
                "min_delta": 1e-4,
            },
            "threshold_percentile": 95.0,
            "score_reduction_axes": [1],
        },

        "LSTM Autoencoder": {
            "input_shape": [60, 3],
            "encoder_units": [32, 16],
            "decoder_units": [16, 32],
            "activation": "tanh",
            "output_activation": "linear",
            "optimizer": "Adam",
            "initial_learning_rate": 0.001,
            "loss": "mse",
            "maximum_epochs": 100,
            "batch_size": 256,
            "shuffle": True,
            "early_stopping": {
                "monitor": "val_loss",
                "patience": 10,
                "min_delta": 0.0,
                "restore_best_weights": True,
            },
            "reduce_lr": {
                "monitor": "val_loss",
                "factor": 0.5,
                "patience": 5,
                "min_lr": 1e-6,
                "min_delta": 1e-4,
            },
            "threshold_percentile": 95.0,
            "score_reduction_axes": [1, 2],
        },
    },
}


# ---------------------------------------------------------------------
# 2. Record important protocol note about callback defaults
# ---------------------------------------------------------------------

NEURAL_SPEC_NOTES = {
    "NAB_Dense_AE": (
        "Settings taken from the corrected post-purge metadata. "
        "EarlyStopping min_delta=1e-5 was explicitly preserved."
    ),

    "NAB_LSTM_AE": (
        "Settings taken from the corrected post-purge metadata. "
        "ReduceLROnPlateau factor=0.5, patience=5, min_lr=1e-6. "
        "min_delta was not explicitly stored and therefore uses the "
        "Keras default value of 1e-4."
    ),

    "BLUED_Dense_AE": (
        "Settings recovered from the original historical notebook. "
        "EarlyStopping and ReduceLROnPlateau arguments omitted min_delta, "
        "so the corresponding Keras defaults are preserved."
    ),

    "BLUED_LSTM_AE": (
        "Settings recovered from the original historical notebook. "
        "EarlyStopping and ReduceLROnPlateau arguments omitted min_delta, "
        "so the corresponding Keras defaults are preserved."
    ),

    "seed_interpretation": (
        "Seed 42 is included because it was the historical seed. "
        "Exact numerical reproduction of the historical seed-42 model is "
        "not assumed because backend, hardware, and runtime environment can "
        "affect floating-point optimization. The purpose of this experiment "
        "is controlled estimation of stochastic training variability."
    ),
}


# ---------------------------------------------------------------------
# 3. Save the frozen neural specification before training
# ---------------------------------------------------------------------

NEURAL_SPEC_PATH = (
    FINAL_REVISION_ROOT
    / "FROZEN_NEURAL_TRAINING_SPECIFICATIONS.json"
)

with open(
    NEURAL_SPEC_PATH,
    "w",
    encoding="utf-8",
) as file:
    json.dump(
        {
            "specifications": NEURAL_SPECS,
            "notes": NEURAL_SPEC_NOTES,
            "fixed_seeds": FIXED_SEEDS,
        },
        file,
        indent=2,
    )


# ---------------------------------------------------------------------
# 4. Deterministic seed-reset utility
# ---------------------------------------------------------------------

def reset_experiment_seed(seed):
    """
    Reset Python, NumPy, and TensorFlow stochastic state before
    constructing each independent neural model.
    """

    tf.keras.backend.clear_session()
    gc.collect()

    random.seed(seed)
    np.random.seed(seed)
    tf.keras.utils.set_random_seed(seed)

    try:
        tf.config.experimental.enable_op_determinism()
    except Exception:
        pass


# ---------------------------------------------------------------------
# 5. Dense Autoencoder builders
# ---------------------------------------------------------------------

def build_nab_dense_autoencoder(seed):
    """
    Build the historically verified NAB 23-feature Dense Autoencoder.
    """

    reset_experiment_seed(seed)

    model_input = Input(
        shape=(23,),
        name="feature_input",
    )

    x = Dense(
        16,
        activation="relu",
        name="encoder_16",
    )(model_input)

    x = Dense(
        8,
        activation="relu",
        name="encoder_8",
    )(x)

    x = Dense(
        4,
        activation="relu",
        name="latent_4",
    )(x)

    x = Dense(
        8,
        activation="relu",
        name="decoder_8",
    )(x)

    x = Dense(
        16,
        activation="relu",
        name="decoder_16",
    )(x)

    model_output = Dense(
        23,
        activation="linear",
        name="reconstruction",
    )(x)

    model = Model(
        inputs=model_input,
        outputs=model_output,
        name=f"NAB_Dense_Autoencoder_Seed_{seed}",
    )

    model.compile(
        optimizer=Adam(
            learning_rate=0.001,
        ),
        loss="mse",
    )

    return model


def build_blued_dense_autoencoder(seed):
    """
    Build the historically verified BLUED 69-feature Dense Autoencoder.
    """

    reset_experiment_seed(seed)

    model_input = Input(
        shape=(69,),
        name="feature_input",
    )

    x = Dense(
        48,
        activation="relu",
        name="encoder_48",
    )(model_input)

    x = Dense(
        24,
        activation="relu",
        name="encoder_24",
    )(x)

    x = Dense(
        8,
        activation="relu",
        name="latent_8",
    )(x)

    x = Dense(
        24,
        activation="relu",
        name="decoder_24",
    )(x)

    x = Dense(
        48,
        activation="relu",
        name="decoder_48",
    )(x)

    model_output = Dense(
        69,
        activation="linear",
        name="reconstruction",
    )(x)

    model = Model(
        inputs=model_input,
        outputs=model_output,
        name=f"BLUED_Dense_Autoencoder_Seed_{seed}",
    )

    model.compile(
        optimizer=Adam(
            learning_rate=0.001,
        ),
        loss="mse",
    )

    return model


# ---------------------------------------------------------------------
# 6. LSTM Autoencoder builders
# ---------------------------------------------------------------------

def build_nab_lstm_autoencoder(seed):
    """
    Build the historically verified NAB 60-step one-channel LSTM Autoencoder.
    """

    reset_experiment_seed(seed)

    model_input = Input(
        shape=(60, 1),
        name="sequence_input",
    )

    x = LSTM(
        32,
        return_sequences=True,
        name="encoder_lstm_1",
    )(model_input)

    x = LSTM(
        16,
        return_sequences=False,
        name="latent_lstm",
    )(x)

    x = RepeatVector(
        60,
        name="repeat_latent_vector",
    )(x)

    x = LSTM(
        16,
        return_sequences=True,
        name="decoder_lstm_1",
    )(x)

    x = LSTM(
        32,
        return_sequences=True,
        name="decoder_lstm_2",
    )(x)

    model_output = TimeDistributed(
        Dense(
            1,
            activation="linear",
        ),
        name="reconstructed_sequence",
    )(x)

    model = Model(
        inputs=model_input,
        outputs=model_output,
        name=f"NAB_LSTM_Autoencoder_Seed_{seed}",
    )

    model.compile(
        optimizer=Adam(
            learning_rate=0.001,
        ),
        loss="mse",
    )

    return model


def build_blued_lstm_autoencoder(seed):
    """
    Build the historically verified BLUED 60-step three-channel LSTM Autoencoder.
    """

    reset_experiment_seed(seed)

    model_input = Input(
        shape=(60, 3),
        name="sequence_input",
    )

    x = LSTM(
        32,
        return_sequences=True,
        name="encoder_lstm_32",
    )(model_input)

    x = LSTM(
        16,
        return_sequences=False,
        name="latent_lstm_16",
    )(x)

    x = RepeatVector(
        60,
        name="repeat_latent",
    )(x)

    x = LSTM(
        16,
        return_sequences=True,
        name="decoder_lstm_16",
    )(x)

    x = LSTM(
        32,
        return_sequences=True,
        name="decoder_lstm_32",
    )(x)

    model_output = TimeDistributed(
        Dense(
            3,
            activation="linear",
        ),
        name="sequence_reconstruction",
    )(x)

    model = Model(
        inputs=model_input,
        outputs=model_output,
        name=f"BLUED_LSTM_Autoencoder_Seed_{seed}",
    )

    model.compile(
        optimizer=Adam(
            learning_rate=0.001,
        ),
        loss="mse",
    )

    return model


# ---------------------------------------------------------------------
# 7. Dataset/model-specific callback builders
# ---------------------------------------------------------------------

def build_callbacks(dataset_name, model_name):
    """
    Construct callbacks exactly according to the frozen historical protocol.
    """

    spec = NEURAL_SPECS[
        dataset_name
    ][
        model_name
    ]

    callbacks = []

    early = spec["early_stopping"]

    callbacks.append(
        EarlyStopping(
            monitor=early["monitor"],
            patience=early["patience"],
            min_delta=early["min_delta"],
            restore_best_weights=early[
                "restore_best_weights"
            ],
            verbose=0,
        )
    )

    reduce_lr = spec["reduce_lr"]

    if reduce_lr is not None:
        callbacks.append(
            ReduceLROnPlateau(
                monitor=reduce_lr["monitor"],
                factor=reduce_lr["factor"],
                patience=reduce_lr["patience"],
                min_lr=reduce_lr["min_lr"],
                min_delta=reduce_lr["min_delta"],
                verbose=0,
            )
        )

    return callbacks


# ---------------------------------------------------------------------
# 8. Reconstruction-score utility
# ---------------------------------------------------------------------

def calculate_reconstruction_scores(
    model,
    X,
    batch_size,
    score_axes,
):
    """
    Generate reconstruction MSE anomaly scores.
    Higher score = more unusual.
    """

    reconstructed = model.predict(
        X,
        batch_size=batch_size,
        verbose=0,
    )

    scores = np.mean(
        np.square(
            X - reconstructed
        ),
        axis=tuple(score_axes),
    )

    return scores.astype(
        np.float64
    )


# ---------------------------------------------------------------------
# 9. Binary evaluation utility
# ---------------------------------------------------------------------

def calculate_binary_metrics(
    y_true,
    anomaly_scores,
    predictions,
):
    """
    Compute the thesis technical evaluation metrics.
    """

    y_true = np.asarray(
        y_true,
        dtype=np.int8,
    )

    anomaly_scores = np.asarray(
        anomaly_scores,
        dtype=np.float64,
    )

    predictions = np.asarray(
        predictions,
        dtype=np.int8,
    )

    tn, fp, fn, tp = confusion_matrix(
        y_true,
        predictions,
        labels=[0, 1],
    ).ravel()

    precision = precision_score(
        y_true,
        predictions,
        zero_division=0,
    )

    recall = recall_score(
        y_true,
        predictions,
        zero_division=0,
    )

    f1 = f1_score(
        y_true,
        predictions,
        zero_division=0,
    )

    fpr = (
        fp / (fp + tn)
        if (fp + tn) > 0
        else np.nan
    )

    roc_auc = roc_auc_score(
        y_true,
        anomaly_scores,
    )

    pr_auc = average_precision_score(
        y_true,
        anomaly_scores,
    )

    return {
        "tn": int(tn),
        "fp": int(fp),
        "fn": int(fn),
        "tp": int(tp),

        "precision": float(
            precision
        ),

        "recall": float(
            recall
        ),

        "f1_score": float(
            f1
        ),

        "false_positive_rate": float(
            fpr
        ),

        "roc_auc": float(
            roc_auc
        ),

        "pr_auc": float(
            pr_auc
        ),

        "flagged_windows": int(
            predictions.sum()
        ),

        "flagged_ratio_percent": float(
            100.0 * predictions.mean()
        ),

        "evaluation_windows": int(
            len(y_true)
        ),
    }


# ---------------------------------------------------------------------
# 10. Validation-threshold utility
# ---------------------------------------------------------------------

def derive_validation_threshold(
    validation_scores,
    percentile=95.0,
):
    """
    Derive the label-blind operating threshold from validation scores only.
    """

    return float(
        np.percentile(
            validation_scores,
            percentile,
        )
    )


# ---------------------------------------------------------------------
# 11. Training-history utility
# ---------------------------------------------------------------------

def history_to_dataframe(history):
    """
    Convert one Keras History object to an auditable table.
    """

    history_df = pd.DataFrame(
        history.history
    )

    history_df.insert(
        0,
        "epoch",
        np.arange(
            1,
            len(history_df) + 1,
        ),
    )

    return history_df


# ---------------------------------------------------------------------
# 12. Architecture sanity checks
# ---------------------------------------------------------------------

architecture_checks = []

for dataset_name, model_name, builder, expected_parameters in [
    (
        "NAB",
        "Dense Autoencoder",
        build_nab_dense_autoencoder,
        1131,
    ),
    (
        "NAB",
        "LSTM Autoencoder",
        build_nab_lstm_autoencoder,
        15905,
    ),
    (
        "BLUED",
        "Dense Autoencoder",
        build_blued_dense_autoencoder,
        9533,
    ),
    (
        "BLUED",
        "LSTM Autoencoder",
        build_blued_lstm_autoencoder,
        16227,
    ),
]:
    temporary_model = builder(
        FIXED_SEEDS[0]
    )

    observed_parameters = (
        temporary_model.count_params()
    )

    if observed_parameters != expected_parameters:
        raise ValueError(
            f"{dataset_name} {model_name}: "
            f"parameter-count mismatch. "
            f"Expected {expected_parameters:,}, "
            f"observed {observed_parameters:,}."
        )

    architecture_checks.append({
        "dataset": dataset_name,
        "model": model_name,
        "expected_parameters": expected_parameters,
        "observed_parameters": observed_parameters,
        "status": "PASS",
    })

    del temporary_model
    tf.keras.backend.clear_session()
    gc.collect()


architecture_check_df = pd.DataFrame(
    architecture_checks
)

ARCHITECTURE_AUDIT_PATH = (
    FINAL_REVISION_ROOT
    / "NEURAL_ARCHITECTURE_REPRODUCTION_AUDIT.csv"
)

architecture_check_df.to_csv(
    ARCHITECTURE_AUDIT_PATH,
    index=False,
)


# ---------------------------------------------------------------------
# 13. Final report
# ---------------------------------------------------------------------

print("=" * 120)
print("NEURAL TRAINING SPECIFICATION FROZEN")
print("=" * 120)

print("\nFixed seeds:")
print(FIXED_SEEDS)

print("\nArchitecture reproduction audit:")
print(
    architecture_check_df.to_string(
        index=False
    )
)

print("\nNAB Dense AE:")
print(
    "  23 → 16 → 8 → 4 → 8 → 16 → 23"
)
print(
    "  Adam LR=0.001 | max 150 epochs | batch=256"
)
print(
    "  EarlyStopping patience=10, min_delta=1e-5"
)

print("\nNAB LSTM AE:")
print(
    "  (60,1) | LSTM 32 → 16 → 16 → 32"
)
print(
    "  Adam LR=0.001 | max 100 epochs | batch=256"
)
print(
    "  EarlyStopping patience=10, min_delta=1e-5"
)
print(
    "  ReduceLR factor=0.5, patience=5, min_lr=1e-6"
)

print("\nBLUED Dense AE:")
print(
    "  69 → 48 → 24 → 8 → 24 → 48 → 69"
)
print(
    "  Adam LR=0.001 | max 150 epochs | batch=256"
)
print(
    "  EarlyStopping patience=10"
)
print(
    "  ReduceLR factor=0.5, patience=5, min_lr=1e-6"
)

print("\nBLUED LSTM AE:")
print(
    "  (60,3) | LSTM 32 → 16 → 16 → 32"
)
print(
    "  Adam LR=0.001 | max 100 epochs | batch=256"
)
print(
    "  EarlyStopping patience=10"
)
print(
    "  ReduceLR factor=0.5, patience=5, min_lr=1e-6"
)

print(
    f"\nFrozen neural specification saved to:\n"
    f"{NEURAL_SPEC_PATH}"
)

print(
    f"\nArchitecture audit saved to:\n"
    f"{ARCHITECTURE_AUDIT_PATH}"
)

print(
    "\nNo neural model was trained in Cell 18."
)
```

```python
# Cell 19 — Run Five-Seed NAB Dense Autoencoder Robustness Experiment:
# Retrain the final leakage-controlled NAB Dense Autoencoder using all five
# predetermined seeds while keeping architecture, data, validation protocol,
# threshold rule, and all hyperparameters fixed. Save every run independently
# and aggregate the completed seed results without selecting a best seed.

import gc
import json
import time

import numpy as np
import pandas as pd
import tensorflow as tf


# ---------------------------------------------------------------------
# 1. Define output directory
# ---------------------------------------------------------------------

NAB_DENSE_SEED_ROOT = (
    NAB_SEED_ROOT
    / "DENSE_AUTOENCODER"
)

NAB_DENSE_SEED_ROOT.mkdir(
    parents=True,
    exist_ok=True,
)


# ---------------------------------------------------------------------
# 2. Frozen experiment settings
# ---------------------------------------------------------------------

dataset_name = "NAB"
model_name = "Dense Autoencoder"

spec = NEURAL_SPECS[
    dataset_name
][
    model_name
]

MAX_EPOCHS = int(
    spec["maximum_epochs"]
)

BATCH_SIZE = int(
    spec["batch_size"]
)

THRESHOLD_PERCENTILE = float(
    spec["threshold_percentile"]
)

SCORE_AXES = tuple(
    spec["score_reduction_axes"]
)


# ---------------------------------------------------------------------
# 3. Verify authoritative inputs once more
# ---------------------------------------------------------------------

assert nab_dense_X_train.shape == (62096, 23)
assert nab_dense_X_validation.shape == (20648, 23)
assert nab_dense_X_test.shape == (20668, 23)

assert len(nab_dense_y_validation) == 20648
assert len(nab_dense_y_test) == 20668

assert np.isfinite(
    nab_dense_X_train
).all()

assert np.isfinite(
    nab_dense_X_validation
).all()

assert np.isfinite(
    nab_dense_X_test
).all()


# ---------------------------------------------------------------------
# 4. Run each predetermined seed independently
# ---------------------------------------------------------------------

all_seed_metric_frames = []

for run_number, seed in enumerate(
    FIXED_SEEDS,
    start=1,
):

    seed_dir = (
        NAB_DENSE_SEED_ROOT
        / f"seed_{seed}"
    )

    seed_dir.mkdir(
        parents=True,
        exist_ok=True,
    )

    completion_file = (
        seed_dir
        / "RUN_COMPLETE.json"
    )

    metrics_file = (
        seed_dir
        / "metrics.csv"
    )

    # -------------------------------------------------------------
    # Resume support
    # -------------------------------------------------------------

    if (
        completion_file.exists()
        and metrics_file.exists()
    ):
        print(
            "\n" + "=" * 120
        )
        print(
            f"Seed {seed} already completed — "
            "loading saved metrics."
        )
        print(
            "=" * 120
        )

        saved_metrics = pd.read_csv(
            metrics_file
        )

        all_seed_metric_frames.append(
            saved_metrics
        )

        continue


    print(
        "\n" + "=" * 120
    )
    print(
        f"NAB DENSE AUTOENCODER — "
        f"RUN {run_number}/"
        f"{len(FIXED_SEEDS)} — "
        f"SEED {seed}"
    )
    print(
        "=" * 120
    )


    # -------------------------------------------------------------
    # 4.1 Build model and reset all stochastic state
    # -------------------------------------------------------------

    model = build_nab_dense_autoencoder(
        seed
    )

    callbacks = build_callbacks(
        dataset_name="NAB",
        model_name="Dense Autoencoder",
    )


    # -------------------------------------------------------------
    # 4.2 Train using training data only with validation monitoring
    # -------------------------------------------------------------

    training_start = time.perf_counter()

    history = model.fit(
        nab_dense_X_train,
        nab_dense_X_train,
        validation_data=(
            nab_dense_X_validation,
            nab_dense_X_validation,
        ),
        epochs=MAX_EPOCHS,
        batch_size=BATCH_SIZE,
        shuffle=True,
        callbacks=callbacks,
        verbose=2,
    )

    training_seconds = (
        time.perf_counter()
        - training_start
    )


    # -------------------------------------------------------------
    # 4.3 Convert and inspect training history
    # -------------------------------------------------------------

    history_df = history_to_dataframe(
        history
    )

    epochs_run = int(
        len(history_df)
    )

    best_epoch_position = int(
        history_df[
            "val_loss"
        ].idxmin()
        + 1
    )

    best_validation_loss = float(
        history_df[
            "val_loss"
        ].min()
    )


    # -------------------------------------------------------------
    # 4.4 Generate validation anomaly scores
    # -------------------------------------------------------------

    validation_score_start = (
        time.perf_counter()
    )

    validation_scores = (
        calculate_reconstruction_scores(
            model=model,
            X=nab_dense_X_validation,
            batch_size=BATCH_SIZE,
            score_axes=SCORE_AXES,
        )
    )

    validation_scoring_seconds = (
        time.perf_counter()
        - validation_score_start
    )


    # -------------------------------------------------------------
    # 4.5 Derive this run's threshold from validation scores only
    # -------------------------------------------------------------

    threshold = (
        derive_validation_threshold(
            validation_scores,
            percentile=THRESHOLD_PERCENTILE,
        )
    )

    validation_predictions = (
        validation_scores > threshold
    ).astype(np.int8)


    # Labels are revealed only for evaluation.
    validation_metrics = (
        calculate_binary_metrics(
            y_true=nab_dense_y_validation,
            anomaly_scores=validation_scores,
            predictions=validation_predictions,
        )
    )


    # -------------------------------------------------------------
    # 4.6 Generate test anomaly scores using frozen threshold
    # -------------------------------------------------------------

    test_score_start = (
        time.perf_counter()
    )

    test_scores = (
        calculate_reconstruction_scores(
            model=model,
            X=nab_dense_X_test,
            batch_size=BATCH_SIZE,
            score_axes=SCORE_AXES,
        )
    )

    test_scoring_seconds = (
        time.perf_counter()
        - test_score_start
    )

    test_predictions = (
        test_scores > threshold
    ).astype(np.int8)


    # Labels are revealed only for final evaluation.
    test_metrics = (
        calculate_binary_metrics(
            y_true=nab_dense_y_test,
            anomaly_scores=test_scores,
            predictions=test_predictions,
        )
    )


    # -------------------------------------------------------------
    # 4.7 Build auditable validation/test result rows
    # -------------------------------------------------------------

    common_metadata = {
        "dataset": "NAB",
        "model": "Dense Autoencoder",
        "seed": int(seed),
        "run_number": int(run_number),
        "threshold_percentile":
            THRESHOLD_PERCENTILE,
        "threshold": float(threshold),
        "training_seconds":
            float(training_seconds),
        "epochs_run":
            int(epochs_run),
        "best_epoch_position":
            int(best_epoch_position),
        "best_validation_loss":
            float(best_validation_loss),
        "batch_size":
            int(BATCH_SIZE),
        "maximum_epochs":
            int(MAX_EPOCHS),
    }

    validation_row = {
        **common_metadata,
        "split": "validation",
        "scoring_seconds":
            float(
                validation_scoring_seconds
            ),
        **validation_metrics,
    }

    test_row = {
        **common_metadata,
        "split": "test",
        "scoring_seconds":
            float(
                test_scoring_seconds
            ),
        **test_metrics,
    }

    run_metrics_df = pd.DataFrame(
        [
            validation_row,
            test_row,
        ]
    )


    # -------------------------------------------------------------
    # 4.8 Save complete row-level score artifacts
    # -------------------------------------------------------------

    validation_scores_df = pd.DataFrame({
        "window_id":
            nab_dense_validation_ids,

        "label":
            nab_dense_y_validation,

        "anomaly_score":
            validation_scores,

        "threshold":
            threshold,

        "flagged":
            validation_predictions,
    })

    test_scores_df = pd.DataFrame({
        "window_id":
            nab_dense_test_ids,

        "label":
            nab_dense_y_test,

        "anomaly_score":
            test_scores,

        "threshold":
            threshold,

        "flagged":
            test_predictions,
    })


    # -------------------------------------------------------------
    # 4.9 Save this seed immediately
    # -------------------------------------------------------------

    history_df.to_csv(
        seed_dir
        / "training_history.csv",
        index=False,
    )

    run_metrics_df.to_csv(
        metrics_file,
        index=False,
    )

    validation_scores_df.to_csv(
        seed_dir
        / "validation_scores.csv",
        index=False,
    )

    test_scores_df.to_csv(
        seed_dir
        / "test_scores.csv",
        index=False,
    )

    model.save(
        seed_dir
        / "model.keras"
    )


    # -------------------------------------------------------------
    # 4.10 Save run metadata / completion marker
    # -------------------------------------------------------------

    run_metadata = {
        "dataset": "NAB",
        "model":
            "Dense Autoencoder",

        "seed":
            int(seed),

        "run_number":
            int(run_number),

        "training_rows":
            int(
                len(
                    nab_dense_X_train
                )
            ),

        "validation_rows":
            int(
                len(
                    nab_dense_X_validation
                )
            ),

        "test_rows":
            int(
                len(
                    nab_dense_X_test
                )
            ),

        "threshold_source":
            (
                "95th percentile of "
                "this seed's validation "
                "reconstruction scores"
            ),

        "labels_used_for_training":
            False,

        "labels_used_for_threshold":
            False,

        "labels_used_for_evaluation":
            True,

        "threshold":
            float(threshold),

        "epochs_run":
            int(epochs_run),

        "best_epoch_position":
            int(
                best_epoch_position
            ),

        "best_validation_loss":
            float(
                best_validation_loss
            ),

        "training_seconds":
            float(
                training_seconds
            ),

        "test_f1":
            float(
                test_metrics[
                    "f1_score"
                ]
            ),

        "test_roc_auc":
            float(
                test_metrics[
                    "roc_auc"
                ]
            ),

        "test_pr_auc":
            float(
                test_metrics[
                    "pr_auc"
                ]
            ),

        "status":
            "COMPLETE",
    }

    with open(
        completion_file,
        "w",
        encoding="utf-8",
    ) as file:
        json.dump(
            run_metadata,
            file,
            indent=2,
        )


    # -------------------------------------------------------------
    # 4.11 Add to in-memory aggregate
    # -------------------------------------------------------------

    all_seed_metric_frames.append(
        run_metrics_df
    )


    # -------------------------------------------------------------
    # 4.12 Print concise run result
    # -------------------------------------------------------------

    print(
        f"\nSeed {seed} complete"
    )

    print(
        f"Epochs run: "
        f"{epochs_run}"
    )

    print(
        f"Best validation epoch: "
        f"{best_epoch_position}"
    )

    print(
        f"Validation threshold: "
        f"{threshold:.8f}"
    )

    print(
        f"Test Precision: "
        f"{test_metrics['precision']:.6f}"
    )

    print(
        f"Test Recall: "
        f"{test_metrics['recall']:.6f}"
    )

    print(
        f"Test F1: "
        f"{test_metrics['f1_score']:.6f}"
    )

    print(
        f"Test FPR: "
        f"{test_metrics['false_positive_rate']:.6f}"
    )

    print(
        f"Test ROC-AUC: "
        f"{test_metrics['roc_auc']:.6f}"
    )

    print(
        f"Test PR-AUC: "
        f"{test_metrics['pr_auc']:.6f}"
    )

    print(
        f"Training time: "
        f"{training_seconds:.2f} seconds"
    )


    # -------------------------------------------------------------
    # 4.13 Release GPU memory before next seed
    # -------------------------------------------------------------

    del model
    del history

    tf.keras.backend.clear_session()
    gc.collect()


# ---------------------------------------------------------------------
# 5. Combine all completed seed runs
# ---------------------------------------------------------------------

all_seed_metrics_df = pd.concat(
    all_seed_metric_frames,
    ignore_index=True,
)

all_seed_metrics_df = (
    all_seed_metrics_df
    .sort_values(
        [
            "seed",
            "split",
        ]
    )
    .reset_index(
        drop=True
    )
)


ALL_SEED_METRICS_PATH = (
    NAB_DENSE_SEED_ROOT
    / "nab_dense_autoencoder_all_seed_metrics.csv"
)

all_seed_metrics_df.to_csv(
    ALL_SEED_METRICS_PATH,
    index=False,
)


# ---------------------------------------------------------------------
# 6. Verify that all five seeds completed
# ---------------------------------------------------------------------

test_seed_results = (
    all_seed_metrics_df[
        all_seed_metrics_df[
            "split"
        ] == "test"
    ]
    .copy()
    .sort_values(
        "seed"
    )
    .reset_index(
        drop=True
    )
)

observed_seeds = sorted(
    test_seed_results[
        "seed"
    ]
    .astype(int)
    .tolist()
)

expected_seeds = sorted(
    FIXED_SEEDS
)

if observed_seeds != expected_seeds:
    raise ValueError(
        "Five-seed completion check failed.\n"
        f"Expected: {expected_seeds}\n"
        f"Observed: {observed_seeds}"
    )


# ---------------------------------------------------------------------
# 7. Compute descriptive seed-robustness summary
# ---------------------------------------------------------------------

summary_metrics = [
    "precision",
    "recall",
    "f1_score",
    "false_positive_rate",
    "roc_auc",
    "pr_auc",
    "flagged_ratio_percent",
    "threshold",
    "epochs_run",
    "training_seconds",
]

summary_rows = []

for metric in summary_metrics:

    values = pd.to_numeric(
        test_seed_results[
            metric
        ],
        errors="coerce",
    )

    summary_rows.append({
        "dataset":
            "NAB",

        "model":
            "Dense Autoencoder",

        "split":
            "test",

        "metric":
            metric,

        "n_seeds":
            int(
                values.notna().sum()
            ),

        "mean":
            float(
                values.mean()
            ),

        "sample_std":
            float(
                values.std(
                    ddof=1
                )
            ),

        "minimum":
            float(
                values.min()
            ),

        "maximum":
            float(
                values.max()
            ),
    })


seed_summary_df = pd.DataFrame(
    summary_rows
)

SEED_SUMMARY_PATH = (
    NAB_DENSE_SEED_ROOT
    / "nab_dense_autoencoder_seed_summary.csv"
)

seed_summary_df.to_csv(
    SEED_SUMMARY_PATH,
    index=False,
)


# ---------------------------------------------------------------------
# 8. Save compact five-seed test table
# ---------------------------------------------------------------------

compact_columns = [
    "seed",
    "threshold",
    "epochs_run",
    "precision",
    "recall",
    "f1_score",
    "false_positive_rate",
    "roc_auc",
    "pr_auc",
    "flagged_windows",
    "flagged_ratio_percent",
    "training_seconds",
]

compact_test_df = (
    test_seed_results[
        compact_columns
    ]
    .copy()
)

COMPACT_TEST_PATH = (
    NAB_DENSE_SEED_ROOT
    / "nab_dense_autoencoder_five_seed_test_results.csv"
)

compact_test_df.to_csv(
    COMPACT_TEST_PATH,
    index=False,
)


# ---------------------------------------------------------------------
# 9. Print final five-seed results
# ---------------------------------------------------------------------

print(
    "\n" + "=" * 120
)
print(
    "NAB DENSE AUTOENCODER — "
    "FIVE-SEED ROBUSTNESS COMPLETE"
)
print(
    "=" * 120
)

print(
    "\nIndividual test results:"
)

print(
    compact_test_df.to_string(
        index=False
    )
)

print(
    "\nMean ± sample standard deviation:"
)

for metric in [
    "precision",
    "recall",
    "f1_score",
    "false_positive_rate",
    "roc_auc",
    "pr_auc",
]:

    row = seed_summary_df.loc[
        seed_summary_df[
            "metric"
        ] == metric
    ].iloc[0]

    print(
        f"{metric:<24} "
        f"{row['mean']:.6f} "
        f"± "
        f"{row['sample_std']:.6f}"
    )


print(
    "\nSaved aggregate files:"
)

print(
    f"  {ALL_SEED_METRICS_PATH}"
)

print(
    f"  {SEED_SUMMARY_PATH}"
)

print(
    f"  {COMPACT_TEST_PATH}"
)

print(
    "\nAll five predetermined seeds were retained. "
    "No best-seed selection was performed."
)
```

```python
# Cell 21 — Consolidate NAB Five-Seed Neural Robustness Results:
# Combine the completed Dense AE and LSTM-AE five-seed NAB results, quantify
# stochastic variability, compare mean operating-point and ranking behavior,
# and contextualize the original corrected single-seed thesis results.

import numpy as np
import pandas as pd
from pathlib import Path


# ---------------------------------------------------------------------
# 1. Define saved five-seed result files
# ---------------------------------------------------------------------

NAB_DENSE_TEST_PATH = (
    NAB_SEED_ROOT
    / "DENSE_AUTOENCODER"
    / "nab_dense_autoencoder_five_seed_test_results.csv"
)

NAB_LSTM_TEST_PATH = (
    NAB_SEED_ROOT
    / "LSTM_AUTOENCODER"
    / "nab_lstm_autoencoder_five_seed_test_results.csv"
)

NAB_COMBINED_ROOT = (
    NAB_SEED_ROOT
    / "COMBINED_NEURAL_ROBUSTNESS"
)

NAB_COMBINED_ROOT.mkdir(
    parents=True,
    exist_ok=True,
)


# ---------------------------------------------------------------------
# 2. Load completed five-seed results
# ---------------------------------------------------------------------

dense_df = pd.read_csv(
    NAB_DENSE_TEST_PATH
)

lstm_df = pd.read_csv(
    NAB_LSTM_TEST_PATH
)

dense_df.insert(
    0,
    "model",
    "Dense Autoencoder",
)

lstm_df.insert(
    0,
    "model",
    "LSTM Autoencoder",
)

combined_df = pd.concat(
    [
        dense_df,
        lstm_df,
    ],
    ignore_index=True,
)


# ---------------------------------------------------------------------
# 3. Verify the frozen five-seed design
# ---------------------------------------------------------------------

for model_name, model_df in combined_df.groupby(
    "model"
):

    observed_seeds = sorted(
        model_df["seed"]
        .astype(int)
        .tolist()
    )

    if observed_seeds != sorted(FIXED_SEEDS):

        raise ValueError(
            f"{model_name}: seed mismatch.\n"
            f"Expected: {sorted(FIXED_SEEDS)}\n"
            f"Observed: {observed_seeds}"
        )


# ---------------------------------------------------------------------
# 4. Calculate model-level stochastic-robustness statistics
# ---------------------------------------------------------------------

comparison_metrics = [
    "precision",
    "recall",
    "f1_score",
    "false_positive_rate",
    "roc_auc",
    "pr_auc",
    "threshold",
    "flagged_ratio_percent",
    "epochs_run",
    "training_seconds",
]

summary_rows = []

for model_name, model_df in combined_df.groupby(
    "model"
):

    for metric in comparison_metrics:

        values = pd.to_numeric(
            model_df[metric],
            errors="coerce",
        )

        mean_value = float(
            values.mean()
        )

        std_value = float(
            values.std(ddof=1)
        )

        cv_percent = (
            100.0 * std_value / abs(mean_value)
            if mean_value != 0
            else np.nan
        )

        summary_rows.append({
            "dataset":
                "NAB",

            "model":
                model_name,

            "metric":
                metric,

            "n_seeds":
                int(values.notna().sum()),

            "mean":
                mean_value,

            "sample_std":
                std_value,

            "minimum":
                float(values.min()),

            "maximum":
                float(values.max()),

            "range":
                float(
                    values.max()
                    - values.min()
                ),

            "coefficient_of_variation_percent":
                float(cv_percent),
        })


summary_df = pd.DataFrame(
    summary_rows
)


# ---------------------------------------------------------------------
# 5. Historical corrected single-seed results from the frozen thesis
# ---------------------------------------------------------------------

historical_results = {
    "Dense Autoencoder": {
        "precision": 0.285882,
        "recall": 0.192050,
        "f1_score": 0.229755,
        "false_positive_rate": 0.067248,
        "roc_auc": 0.576805,
        "pr_auc": 0.196589,
        "threshold": 0.336275,
    },

    "LSTM Autoencoder": {
        "precision": 0.239643,
        "recall": 0.116096,
        "f1_score": 0.156416,
        "false_positive_rate": 0.051636,
        "roc_auc": 0.566271,
        "pr_auc": 0.166804,
        "threshold": 0.404100,
    },
}


historical_context_rows = []

for model_name, metric_values in (
    historical_results.items()
):

    for metric, historical_value in (
        metric_values.items()
    ):

        summary_row = summary_df.loc[
            (summary_df["model"] == model_name)
            & (summary_df["metric"] == metric)
        ]

        if len(summary_row) != 1:
            raise ValueError(
                f"Missing robustness summary for "
                f"{model_name} / {metric}"
            )

        summary_row = summary_row.iloc[0]

        historical_context_rows.append({
            "dataset":
                "NAB",

            "model":
                model_name,

            "metric":
                metric,

            "historical_corrected_value":
                historical_value,

            "five_seed_mean":
                float(
                    summary_row["mean"]
                ),

            "five_seed_sample_std":
                float(
                    summary_row[
                        "sample_std"
                    ]
                ),

            "five_seed_minimum":
                float(
                    summary_row[
                        "minimum"
                    ]
                ),

            "five_seed_maximum":
                float(
                    summary_row[
                        "maximum"
                    ]
                ),

            "historical_inside_five_seed_range":
                bool(
                    summary_row["minimum"]
                    <= historical_value
                    <= summary_row["maximum"]
                ),
        })


historical_context_df = pd.DataFrame(
    historical_context_rows
)


# ---------------------------------------------------------------------
# 6. Compare the two models using their five-seed means
# ---------------------------------------------------------------------

primary_metrics = [
    "precision",
    "recall",
    "f1_score",
    "false_positive_rate",
    "roc_auc",
    "pr_auc",
]

mean_comparison_rows = []

for metric in primary_metrics:

    dense_mean = float(
        summary_df.loc[
            (summary_df["model"] == "Dense Autoencoder")
            & (summary_df["metric"] == metric),
            "mean",
        ].iloc[0]
    )

    dense_std = float(
        summary_df.loc[
            (summary_df["model"] == "Dense Autoencoder")
            & (summary_df["metric"] == metric),
            "sample_std",
        ].iloc[0]
    )

    lstm_mean = float(
        summary_df.loc[
            (summary_df["model"] == "LSTM Autoencoder")
            & (summary_df["metric"] == metric),
            "mean",
        ].iloc[0]
    )

    lstm_std = float(
        summary_df.loc[
            (summary_df["model"] == "LSTM Autoencoder")
            & (summary_df["metric"] == metric),
            "sample_std",
        ].iloc[0]
    )

    mean_comparison_rows.append({
        "metric":
            metric,

        "dense_ae_mean":
            dense_mean,

        "dense_ae_sample_std":
            dense_std,

        "lstm_ae_mean":
            lstm_mean,

        "lstm_ae_sample_std":
            lstm_std,

        "dense_minus_lstm_mean":
            dense_mean - lstm_mean,
    })


mean_comparison_df = pd.DataFrame(
    mean_comparison_rows
)


# ---------------------------------------------------------------------
# 7. Save all combined NAB robustness artifacts
# ---------------------------------------------------------------------

COMBINED_RESULTS_PATH = (
    NAB_COMBINED_ROOT
    / "nab_neural_five_seed_combined_test_results.csv"
)

SUMMARY_PATH = (
    NAB_COMBINED_ROOT
    / "nab_neural_five_seed_summary_comparison.csv"
)

HISTORICAL_CONTEXT_PATH = (
    NAB_COMBINED_ROOT
    / "nab_neural_historical_single_seed_context.csv"
)

MEAN_COMPARISON_PATH = (
    NAB_COMBINED_ROOT
    / "nab_dense_vs_lstm_five_seed_mean_comparison.csv"
)


combined_df.to_csv(
    COMBINED_RESULTS_PATH,
    index=False,
)

summary_df.to_csv(
    SUMMARY_PATH,
    index=False,
)

historical_context_df.to_csv(
    HISTORICAL_CONTEXT_PATH,
    index=False,
)

mean_comparison_df.to_csv(
    MEAN_COMPARISON_PATH,
    index=False,
)


# ---------------------------------------------------------------------
# 8. Print thesis-relevant comparison
# ---------------------------------------------------------------------

print("=" * 120)
print("NAB FIVE-SEED NEURAL ROBUSTNESS COMPARISON")
print("=" * 120)

print("\nPrimary metric comparison:")

display_table = (
    mean_comparison_df.copy()
)

for column in [
    "dense_ae_mean",
    "dense_ae_sample_std",
    "lstm_ae_mean",
    "lstm_ae_sample_std",
    "dense_minus_lstm_mean",
]:
    display_table[column] = (
        display_table[column]
        .round(6)
    )

print(
    display_table.to_string(
        index=False
    )
)


# ---------------------------------------------------------------------
# 9. Print F1 stochastic variability explicitly
# ---------------------------------------------------------------------

print("\nF1 stochastic variability:")

for model_name in [
    "Dense Autoencoder",
    "LSTM Autoencoder",
]:

    row = summary_df.loc[
        (summary_df["model"] == model_name)
        & (summary_df["metric"] == "f1_score")
    ].iloc[0]

    print(
        f"{model_name:<20} "
        f"mean={row['mean']:.6f}, "
        f"SD={row['sample_std']:.6f}, "
        f"range=[{row['minimum']:.6f}, "
        f"{row['maximum']:.6f}], "
        f"CV={row['coefficient_of_variation_percent']:.2f}%"
    )


# ---------------------------------------------------------------------
# 10. Historical-context check
# ---------------------------------------------------------------------

print(
    "\nHistorical corrected single-seed values "
    "inside current five-seed ranges:"
)

historical_display = (
    historical_context_df[
        [
            "model",
            "metric",
            "historical_corrected_value",
            "five_seed_minimum",
            "five_seed_maximum",
            "historical_inside_five_seed_range",
        ]
    ]
)

print(
    historical_display.to_string(
        index=False
    )
)


print("\nSaved comparison artifacts:")

print(
    f"  {COMBINED_RESULTS_PATH}"
)

print(
    f"  {SUMMARY_PATH}"
)

print(
    f"  {HISTORICAL_CONTEXT_PATH}"
)

print(
    f"  {MEAN_COMPARISON_PATH}"
)

print(
    "\nNo models were trained in Cell 21."
)
```

```python
# Cell 23 — Run Five-Seed BLUED LSTM Autoencoder Robustness Experiment:
# Retrain the final three-channel BLUED LSTM Autoencoder across all five
# predetermined seeds while keeping the purged chronological split,
# architecture, callbacks, threshold rule, and event-alignment evaluation fixed.

import gc
import json
import time

import numpy as np
import pandas as pd
import tensorflow as tf


# ---------------------------------------------------------------------
# 1. Define output directory
# ---------------------------------------------------------------------

BLUED_LSTM_SEED_ROOT = (
    BLUED_SEED_ROOT
    / "LSTM_AUTOENCODER"
)

BLUED_LSTM_SEED_ROOT.mkdir(
    parents=True,
    exist_ok=True,
)


# ---------------------------------------------------------------------
# 2. Frozen experiment settings
# ---------------------------------------------------------------------

dataset_name = "BLUED"
model_name = "LSTM Autoencoder"

spec = NEURAL_SPECS[
    dataset_name
][
    model_name
]

MAX_EPOCHS = int(
    spec["maximum_epochs"]
)

BATCH_SIZE = int(
    spec["batch_size"]
)

THRESHOLD_PERCENTILE = float(
    spec["threshold_percentile"]
)

SCORE_AXES = tuple(
    spec["score_reduction_axes"]
)


# ---------------------------------------------------------------------
# 3. Verify authoritative BLUED LSTM inputs
# ---------------------------------------------------------------------

assert blued_lstm_X_train.shape == (
    12248,
    60,
    3,
)

assert blued_lstm_X_validation.shape == (
    4079,
    60,
    3,
)

assert blued_lstm_X_test.shape == (
    4079,
    60,
    3,
)

assert len(
    blued_y_validation
) == 4079

assert len(
    blued_y_test
) == 4079

assert int(
    blued_y_test.sum()
) == 450

for array_name, array in {
    "train":
        blued_lstm_X_train,

    "validation":
        blued_lstm_X_validation,

    "test":
        blued_lstm_X_test,
}.items():

    if not np.isfinite(
        array
    ).all():

        raise ValueError(
            f"BLUED LSTM AE {array_name} "
            "contains non-finite values."
        )


# ---------------------------------------------------------------------
# 4. Run each predetermined seed independently
# ---------------------------------------------------------------------

all_seed_metric_frames = []

for run_number, seed in enumerate(
    FIXED_SEEDS,
    start=1,
):

    seed_dir = (
        BLUED_LSTM_SEED_ROOT
        / f"seed_{seed}"
    )

    seed_dir.mkdir(
        parents=True,
        exist_ok=True,
    )

    completion_file = (
        seed_dir
        / "RUN_COMPLETE.json"
    )

    metrics_file = (
        seed_dir
        / "metrics.csv"
    )


    # -------------------------------------------------------------
    # Resume support
    # -------------------------------------------------------------

    if (
        completion_file.exists()
        and metrics_file.exists()
    ):

        print(
            "\n" + "=" * 120
        )

        print(
            f"Seed {seed} already completed — "
            "loading saved metrics."
        )

        print(
            "=" * 120
        )

        saved_metrics = pd.read_csv(
            metrics_file
        )

        all_seed_metric_frames.append(
            saved_metrics
        )

        continue


    print(
        "\n" + "=" * 120
    )

    print(
        f"BLUED LSTM AUTOENCODER — "
        f"RUN {run_number}/"
        f"{len(FIXED_SEEDS)} — "
        f"SEED {seed}"
    )

    print(
        "=" * 120
    )


    # -------------------------------------------------------------
    # 4.1 Build independently initialized model
    # -------------------------------------------------------------

    model = (
        build_blued_lstm_autoencoder(
            seed
        )
    )

    callbacks = build_callbacks(
        dataset_name="BLUED",
        model_name="LSTM Autoencoder",
    )


    # -------------------------------------------------------------
    # 4.2 Train label-blind
    # -------------------------------------------------------------

    training_start = (
        time.perf_counter()
    )

    history = model.fit(
        blued_lstm_X_train,
        blued_lstm_X_train,

        validation_data=(
            blued_lstm_X_validation,
            blued_lstm_X_validation,
        ),

        epochs=MAX_EPOCHS,
        batch_size=BATCH_SIZE,
        shuffle=True,
        callbacks=callbacks,
        verbose=2,
    )

    training_seconds = (
        time.perf_counter()
        - training_start
    )


    # -------------------------------------------------------------
    # 4.3 Training-history diagnostics
    # -------------------------------------------------------------

    history_df = (
        history_to_dataframe(
            history
        )
    )

    epochs_run = int(
        len(history_df)
    )

    best_epoch_position = int(
        history_df[
            "val_loss"
        ].idxmin()
        + 1
    )

    best_validation_loss = float(
        history_df[
            "val_loss"
        ].min()
    )

    if "learning_rate" in history_df.columns:

        final_learning_rate = float(
            history_df[
                "learning_rate"
            ].iloc[-1]
        )

        minimum_learning_rate = float(
            history_df[
                "learning_rate"
            ].min()
        )

    else:

        final_learning_rate = np.nan
        minimum_learning_rate = np.nan


    # -------------------------------------------------------------
    # 4.4 Validation reconstruction scores
    # -------------------------------------------------------------

    validation_score_start = (
        time.perf_counter()
    )

    validation_scores = (
        calculate_reconstruction_scores(
            model=model,
            X=blued_lstm_X_validation,
            batch_size=BATCH_SIZE,
            score_axes=SCORE_AXES,
        )
    )

    validation_scoring_seconds = (
        time.perf_counter()
        - validation_score_start
    )


    # -------------------------------------------------------------
    # 4.5 Derive validation-only q95 threshold
    # -------------------------------------------------------------

    threshold = (
        derive_validation_threshold(
            validation_scores,
            percentile=(
                THRESHOLD_PERCENTILE
            ),
        )
    )

    validation_predictions = (
        validation_scores
        > threshold
    ).astype(np.int8)


    # Event labels are revealed only after threshold selection.
    validation_metrics = (
        calculate_binary_metrics(
            y_true=(
                blued_y_validation
            ),
            anomaly_scores=(
                validation_scores
            ),
            predictions=(
                validation_predictions
            ),
        )
    )


    # -------------------------------------------------------------
    # 4.6 Test reconstruction scores
    # -------------------------------------------------------------

    test_score_start = (
        time.perf_counter()
    )

    test_scores = (
        calculate_reconstruction_scores(
            model=model,
            X=blued_lstm_X_test,
            batch_size=BATCH_SIZE,
            score_axes=SCORE_AXES,
        )
    )

    test_scoring_seconds = (
        time.perf_counter()
        - test_score_start
    )

    test_predictions = (
        test_scores
        > threshold
    ).astype(np.int8)


    # BLUED labels represent switching-event windows,
    # not confirmed anomaly labels.
    test_metrics = (
        calculate_binary_metrics(
            y_true=(
                blued_y_test
            ),
            anomaly_scores=(
                test_scores
            ),
            predictions=(
                test_predictions
            ),
        )
    )


    # -------------------------------------------------------------
    # 4.7 Build auditable result rows
    # -------------------------------------------------------------

    common_metadata = {
        "dataset":
            "BLUED",

        "evaluation_semantics":
            "switching_event_alignment",

        "model":
            "LSTM Autoencoder",

        "seed":
            int(seed),

        "run_number":
            int(run_number),

        "threshold_percentile":
            THRESHOLD_PERCENTILE,

        "threshold":
            float(threshold),

        "training_seconds":
            float(
                training_seconds
            ),

        "epochs_run":
            int(
                epochs_run
            ),

        "best_epoch_position":
            int(
                best_epoch_position
            ),

        "best_validation_loss":
            float(
                best_validation_loss
            ),

        "final_learning_rate":
            float(
                final_learning_rate
            ),

        "minimum_learning_rate":
            float(
                minimum_learning_rate
            ),

        "batch_size":
            int(
                BATCH_SIZE
            ),

        "maximum_epochs":
            int(
                MAX_EPOCHS
            ),
    }


    validation_row = {
        **common_metadata,

        "split":
            "validation",

        "scoring_seconds":
            float(
                validation_scoring_seconds
            ),

        **validation_metrics,
    }


    test_row = {
        **common_metadata,

        "split":
            "test",

        "scoring_seconds":
            float(
                test_scoring_seconds
            ),

        **test_metrics,
    }


    run_metrics_df = pd.DataFrame(
        [
            validation_row,
            test_row,
        ]
    )


    # -------------------------------------------------------------
    # 4.8 Save row-level score artifacts
    # -------------------------------------------------------------

    validation_scores_df = pd.DataFrame({
        "model_index":
            blued_validation_indices,

        "event_window_label":
            blued_y_validation,

        "anomaly_score":
            validation_scores,

        "threshold":
            threshold,

        "flagged":
            validation_predictions,
    })


    test_scores_df = pd.DataFrame({
        "model_index":
            blued_test_indices,

        "event_window_label":
            blued_y_test,

        "anomaly_score":
            test_scores,

        "threshold":
            threshold,

        "flagged":
            test_predictions,
    })


    # -------------------------------------------------------------
    # 4.9 Save each seed immediately
    # -------------------------------------------------------------

    history_df.to_csv(
        seed_dir
        / "training_history.csv",
        index=False,
    )

    run_metrics_df.to_csv(
        metrics_file,
        index=False,
    )

    validation_scores_df.to_csv(
        seed_dir
        / "validation_scores.csv",
        index=False,
    )

    test_scores_df.to_csv(
        seed_dir
        / "test_scores.csv",
        index=False,
    )

    model.save(
        seed_dir
        / "model.keras"
    )


    # -------------------------------------------------------------
    # 4.10 Save completion metadata
    # -------------------------------------------------------------

    run_metadata = {
        "dataset":
            "BLUED",

        "evaluation_role":
            (
                "Event-alignment benchmark. "
                "Positive labels denote appliance "
                "switching-event windows rather than "
                "confirmed anomaly windows."
            ),

        "model":
            "LSTM Autoencoder",

        "seed":
            int(seed),

        "run_number":
            int(run_number),

        "training_rows":
            int(
                len(
                    blued_lstm_X_train
                )
            ),

        "validation_rows":
            int(
                len(
                    blued_lstm_X_validation
                )
            ),

        "test_rows":
            int(
                len(
                    blued_lstm_X_test
                )
            ),

        "test_event_windows":
            int(
                blued_y_test.sum()
            ),

        "threshold_source":
            (
                "95th percentile of "
                "this seed's validation "
                "reconstruction scores"
            ),

        "labels_used_for_training":
            False,

        "labels_used_for_threshold":
            False,

        "labels_used_for_evaluation":
            True,

        "threshold":
            float(
                threshold
            ),

        "epochs_run":
            int(
                epochs_run
            ),

        "best_epoch_position":
            int(
                best_epoch_position
            ),

        "best_validation_loss":
            float(
                best_validation_loss
            ),

        "final_learning_rate":
            (
                None
                if np.isnan(
                    final_learning_rate
                )
                else float(
                    final_learning_rate
                )
            ),

        "training_seconds":
            float(
                training_seconds
            ),

        "test_precision":
            float(
                test_metrics[
                    "precision"
                ]
            ),

        "test_recall":
            float(
                test_metrics[
                    "recall"
                ]
            ),

        "test_f1":
            float(
                test_metrics[
                    "f1_score"
                ]
            ),

        "test_fpr":
            float(
                test_metrics[
                    "false_positive_rate"
                ]
            ),

        "test_roc_auc":
            float(
                test_metrics[
                    "roc_auc"
                ]
            ),

        "test_pr_auc":
            float(
                test_metrics[
                    "pr_auc"
                ]
            ),

        "status":
            "COMPLETE",
    }


    with open(
        completion_file,
        "w",
        encoding="utf-8",
    ) as file:

        json.dump(
            run_metadata,
            file,
            indent=2,
        )


    # -------------------------------------------------------------
    # 4.11 Add to aggregate
    # -------------------------------------------------------------

    all_seed_metric_frames.append(
        run_metrics_df
    )


    # -------------------------------------------------------------
    # 4.12 Concise run report
    # -------------------------------------------------------------

    print(
        f"\nSeed {seed} complete"
    )

    print(
        f"Epochs run: "
        f"{epochs_run}"
    )

    print(
        f"Best validation epoch: "
        f"{best_epoch_position}"
    )

    print(
        f"Best validation loss: "
        f"{best_validation_loss:.8f}"
    )

    print(
        f"Validation threshold: "
        f"{threshold:.8f}"
    )

    if not np.isnan(
        final_learning_rate
    ):

        print(
            f"Final learning rate: "
            f"{final_learning_rate:.8g}"
        )

    print(
        f"Test Event-Alignment Precision: "
        f"{test_metrics['precision']:.6f}"
    )

    print(
        f"Test Event-Alignment Recall: "
        f"{test_metrics['recall']:.6f}"
    )

    print(
        f"Test Event-Alignment F1: "
        f"{test_metrics['f1_score']:.6f}"
    )

    print(
        f"Test FPR: "
        f"{test_metrics['false_positive_rate']:.6f}"
    )

    print(
        f"Test ROC-AUC: "
        f"{test_metrics['roc_auc']:.6f}"
    )

    print(
        f"Test PR-AUC: "
        f"{test_metrics['pr_auc']:.6f}"
    )

    print(
        f"Training time: "
        f"{training_seconds:.2f} seconds"
    )


    # -------------------------------------------------------------
    # 4.13 Release GPU memory
    # -------------------------------------------------------------

    del model
    del history

    tf.keras.backend.clear_session()
    gc.collect()


# ---------------------------------------------------------------------
# 5. Aggregate all five runs
# ---------------------------------------------------------------------

all_seed_metrics_df = pd.concat(
    all_seed_metric_frames,
    ignore_index=True,
)

all_seed_metrics_df = (
    all_seed_metrics_df
    .sort_values(
        [
            "seed",
            "split",
        ]
    )
    .reset_index(
        drop=True
    )
)


ALL_SEED_METRICS_PATH = (
    BLUED_LSTM_SEED_ROOT
    / "blued_lstm_autoencoder_all_seed_metrics.csv"
)

all_seed_metrics_df.to_csv(
    ALL_SEED_METRICS_PATH,
    index=False,
)


# ---------------------------------------------------------------------
# 6. Verify all five seeds completed
# ---------------------------------------------------------------------

test_seed_results = (
    all_seed_metrics_df[
        all_seed_metrics_df[
            "split"
        ] == "test"
    ]
    .copy()
    .sort_values(
        "seed"
    )
    .reset_index(
        drop=True
    )
)

observed_seeds = sorted(
    test_seed_results[
        "seed"
    ]
    .astype(int)
    .tolist()
)

expected_seeds = sorted(
    FIXED_SEEDS
)

if observed_seeds != expected_seeds:

    raise ValueError(
        "Five-seed completion check failed.\n"
        f"Expected: {expected_seeds}\n"
        f"Observed: {observed_seeds}"
    )


# ---------------------------------------------------------------------
# 7. Compute descriptive stochastic-robustness summary
# ---------------------------------------------------------------------

summary_metrics = [
    "precision",
    "recall",
    "f1_score",
    "false_positive_rate",
    "roc_auc",
    "pr_auc",
    "flagged_ratio_percent",
    "threshold",
    "epochs_run",
    "best_validation_loss",
    "training_seconds",
]

summary_rows = []

for metric in summary_metrics:

    values = pd.to_numeric(
        test_seed_results[
            metric
        ],
        errors="coerce",
    )

    summary_rows.append({
        "dataset":
            "BLUED",

        "evaluation_semantics":
            "switching_event_alignment",

        "model":
            "LSTM Autoencoder",

        "split":
            "test",

        "metric":
            metric,

        "n_seeds":
            int(
                values.notna().sum()
            ),

        "mean":
            float(
                values.mean()
            ),

        "sample_std":
            float(
                values.std(
                    ddof=1
                )
            ),

        "minimum":
            float(
                values.min()
            ),

        "maximum":
            float(
                values.max()
            ),
    })


seed_summary_df = pd.DataFrame(
    summary_rows
)


SEED_SUMMARY_PATH = (
    BLUED_LSTM_SEED_ROOT
    / "blued_lstm_autoencoder_seed_summary.csv"
)

seed_summary_df.to_csv(
    SEED_SUMMARY_PATH,
    index=False,
)


# ---------------------------------------------------------------------
# 8. Save compact five-seed test table
# ---------------------------------------------------------------------

compact_columns = [
    "seed",
    "threshold",
    "epochs_run",
    "best_epoch_position",
    "best_validation_loss",
    "final_learning_rate",
    "precision",
    "recall",
    "f1_score",
    "false_positive_rate",
    "roc_auc",
    "pr_auc",
    "flagged_windows",
    "flagged_ratio_percent",
    "training_seconds",
]

compact_test_df = (
    test_seed_results[
        compact_columns
    ]
    .copy()
)


COMPACT_TEST_PATH = (
    BLUED_LSTM_SEED_ROOT
    / "blued_lstm_autoencoder_five_seed_test_results.csv"
)

compact_test_df.to_csv(
    COMPACT_TEST_PATH,
    index=False,
)


# ---------------------------------------------------------------------
# 9. Final report
# ---------------------------------------------------------------------

print(
    "\n" + "=" * 120
)

print(
    "BLUED LSTM AUTOENCODER — "
    "FIVE-SEED ROBUSTNESS COMPLETE"
)

print(
    "=" * 120
)

print(
    "\nIMPORTANT: Metrics below measure alignment with "
    "BLUED switching-event annotations, not confirmed anomalies."
)

print(
    "\nIndividual test results:"
)

print(
    compact_test_df.to_string(
        index=False
    )
)


print(
    "\nMean ± sample standard deviation:"
)


for metric in [
    "precision",
    "recall",
    "f1_score",
    "false_positive_rate",
    "roc_auc",
    "pr_auc",
]:

    row = seed_summary_df.loc[
        seed_summary_df[
            "metric"
        ] == metric
    ].iloc[0]

    print(
        f"{metric:<24} "
        f"{row['mean']:.6f} "
        f"± "
        f"{row['sample_std']:.6f}"
    )


print(
    "\nSaved aggregate files:"
)

print(
    f"  {ALL_SEED_METRICS_PATH}"
)

print(
    f"  {SEED_SUMMARY_PATH}"
)

print(
    f"  {COMPACT_TEST_PATH}"
)

print(
    "\nAll five predetermined seeds were retained. "
    "No best-seed selection was performed."
)
```

```python
# Cell 24 — Consolidate All Repeated-Seed Neural Robustness Evidence:
# Combine the five-seed NAB and BLUED Dense/LSTM Autoencoder results into one
# auditable master summary, quantify F1 stochastic variability, distinguish
# thresholded operating-point performance from continuous ranking performance,
# and contextualize the historical single-seed thesis results without treating
# the new seed distribution as a confidence interval.

import numpy as np
import pandas as pd
from pathlib import Path


# ---------------------------------------------------------------------
# 1. Define the four completed five-seed result files
# ---------------------------------------------------------------------

RESULT_FILES = {
    ("NAB", "Dense Autoencoder"):
        NAB_SEED_ROOT
        / "DENSE_AUTOENCODER"
        / "nab_dense_autoencoder_five_seed_test_results.csv",

    ("NAB", "LSTM Autoencoder"):
        NAB_SEED_ROOT
        / "LSTM_AUTOENCODER"
        / "nab_lstm_autoencoder_five_seed_test_results.csv",

    ("BLUED", "Dense Autoencoder"):
        BLUED_SEED_ROOT
        / "DENSE_AUTOENCODER"
        / "blued_dense_autoencoder_five_seed_test_results.csv",

    ("BLUED", "LSTM Autoencoder"):
        BLUED_SEED_ROOT
        / "LSTM_AUTOENCODER"
        / "blued_lstm_autoencoder_five_seed_test_results.csv",
}


# ---------------------------------------------------------------------
# 2. Create consolidated output directory
# ---------------------------------------------------------------------

MASTER_ROBUSTNESS_ROOT = (
    REPEATED_SEED_ROOT
    / "MASTER_NEURAL_ROBUSTNESS_SUMMARY"
)

MASTER_ROBUSTNESS_ROOT.mkdir(
    parents=True,
    exist_ok=True,
)


# ---------------------------------------------------------------------
# 3. Load and annotate all twenty test runs
# ---------------------------------------------------------------------

all_frames = []

for (
    dataset_name,
    model_name,
), path in RESULT_FILES.items():

    if not path.exists():
        raise FileNotFoundError(
            f"Missing five-seed result file:\n{path}"
        )

    df = pd.read_csv(
        path
    )

    df.insert(
        0,
        "dataset",
        dataset_name,
    )

    df.insert(
        1,
        "model",
        model_name,
    )

    if dataset_name == "NAB":

        semantics = (
            "anomaly_window_evaluation"
        )

    else:

        semantics = (
            "switching_event_alignment"
        )

    df.insert(
        2,
        "evaluation_semantics",
        semantics,
    )

    all_frames.append(
        df
    )


all_runs_df = pd.concat(
    all_frames,
    ignore_index=True,
)


# ---------------------------------------------------------------------
# 4. Verify experimental completeness
# ---------------------------------------------------------------------

expected_seeds = sorted(
    FIXED_SEEDS
)

completion_rows = []

for (
    dataset_name,
    model_name,
), group in all_runs_df.groupby(
    [
        "dataset",
        "model",
    ]
):

    observed_seeds = sorted(
        group[
            "seed"
        ]
        .astype(int)
        .tolist()
    )

    passed = (
        observed_seeds
        == expected_seeds
    )

    completion_rows.append({
        "dataset":
            dataset_name,

        "model":
            model_name,

        "observed_seeds":
            ",".join(
                str(x)
                for x
                in observed_seeds
            ),

        "expected_seeds":
            ",".join(
                str(x)
                for x
                in expected_seeds
            ),

        "n_runs":
            len(group),

        "status":
            (
                "PASS"
                if passed
                else "FAIL"
            ),
    })

    if not passed:

        raise ValueError(
            f"Seed completeness failure for "
            f"{dataset_name} / {model_name}.\n"
            f"Expected: {expected_seeds}\n"
            f"Observed: {observed_seeds}"
        )


completion_df = pd.DataFrame(
    completion_rows
)


# ---------------------------------------------------------------------
# 5. Compute master robustness statistics
# ---------------------------------------------------------------------

METRICS = [
    "precision",
    "recall",
    "f1_score",
    "false_positive_rate",
    "roc_auc",
    "pr_auc",
    "flagged_ratio_percent",
    "threshold",
    "epochs_run",
    "training_seconds",
]

summary_rows = []

for (
    dataset_name,
    model_name,
), group in all_runs_df.groupby(
    [
        "dataset",
        "model",
    ]
):

    semantics = (
        "anomaly_window_evaluation"
        if dataset_name == "NAB"
        else "switching_event_alignment"
    )

    for metric in METRICS:

        values = pd.to_numeric(
            group[
                metric
            ],
            errors="coerce",
        )

        mean_value = float(
            values.mean()
        )

        std_value = float(
            values.std(
                ddof=1
            )
        )

        minimum = float(
            values.min()
        )

        maximum = float(
            values.max()
        )

        cv_percent = (
            100.0
            * std_value
            / abs(mean_value)
            if mean_value != 0
            else np.nan
        )

        summary_rows.append({
            "dataset":
                dataset_name,

            "evaluation_semantics":
                semantics,

            "model":
                model_name,

            "metric":
                metric,

            "n_seeds":
                int(
                    values.notna().sum()
                ),

            "mean":
                mean_value,

            "sample_std":
                std_value,

            "minimum":
                minimum,

            "maximum":
                maximum,

            "range":
                maximum
                - minimum,

            "coefficient_of_variation_percent":
                float(
                    cv_percent
                ),
        })


master_summary_df = pd.DataFrame(
    summary_rows
)


# ---------------------------------------------------------------------
# 6. Build thesis-ready one-row-per-model summary
# ---------------------------------------------------------------------

def get_summary_value(
    dataset_name,
    model_name,
    metric,
    field,
):

    row = master_summary_df.loc[
        (
            master_summary_df[
                "dataset"
            ] == dataset_name
        )
        &
        (
            master_summary_df[
                "model"
            ] == model_name
        )
        &
        (
            master_summary_df[
                "metric"
            ] == metric
        )
    ]

    if len(row) != 1:
        raise ValueError(
            f"Expected one summary row for "
            f"{dataset_name} / "
            f"{model_name} / "
            f"{metric}"
        )

    return float(
        row.iloc[0][field]
    )


compact_rows = []

for dataset_name in [
    "NAB",
    "BLUED",
]:

    for model_name in [
        "Dense Autoencoder",
        "LSTM Autoencoder",
    ]:

        row = {
            "dataset":
                dataset_name,

            "evaluation_semantics":
                (
                    "Anomaly-window evaluation"
                    if dataset_name == "NAB"
                    else
                    "Switching-event alignment"
                ),

            "model":
                model_name,
        }

        for metric in [
            "precision",
            "recall",
            "f1_score",
            "false_positive_rate",
            "roc_auc",
            "pr_auc",
        ]:

            row[
                f"{metric}_mean"
            ] = get_summary_value(
                dataset_name,
                model_name,
                metric,
                "mean",
            )

            row[
                f"{metric}_sample_std"
            ] = get_summary_value(
                dataset_name,
                model_name,
                metric,
                "sample_std",
            )

        row[
            "f1_minimum"
        ] = get_summary_value(
            dataset_name,
            model_name,
            "f1_score",
            "minimum",
        )

        row[
            "f1_maximum"
        ] = get_summary_value(
            dataset_name,
            model_name,
            "f1_score",
            "maximum",
        )

        row[
            "f1_cv_percent"
        ] = get_summary_value(
            dataset_name,
            model_name,
            "f1_score",
            "coefficient_of_variation_percent",
        )

        row[
            "threshold_minimum"
        ] = get_summary_value(
            dataset_name,
            model_name,
            "threshold",
            "minimum",
        )

        row[
            "threshold_maximum"
        ] = get_summary_value(
            dataset_name,
            model_name,
            "threshold",
            "maximum",
        )

        row[
            "epochs_minimum"
        ] = get_summary_value(
            dataset_name,
            model_name,
            "epochs_run",
            "minimum",
        )

        row[
            "epochs_maximum"
        ] = get_summary_value(
            dataset_name,
            model_name,
            "epochs_run",
            "maximum",
        )

        compact_rows.append(
            row
        )


compact_summary_df = pd.DataFrame(
    compact_rows
)


# ---------------------------------------------------------------------
# 7. Historical corrected single-seed context
# ---------------------------------------------------------------------

HISTORICAL_RESULTS = {
    ("NAB", "Dense Autoencoder"): {
        "precision": 0.285882,
        "recall": 0.192050,
        "f1_score": 0.229755,
        "false_positive_rate": 0.067248,
        "roc_auc": 0.576805,
        "pr_auc": 0.196589,
        "threshold": 0.336275,
    },

    ("NAB", "LSTM Autoencoder"): {
        "precision": 0.239643,
        "recall": 0.116096,
        "f1_score": 0.156416,
        "false_positive_rate": 0.051636,
        "roc_auc": 0.566271,
        "pr_auc": 0.166804,
        "threshold": 0.404100,
    },

    ("BLUED", "Dense Autoencoder"): {
        "precision": 0.559055,
        "recall": 0.315556,
        "f1_score": 0.403409,
        "false_positive_rate": 0.030862,
        "roc_auc": 0.731353,
        "pr_auc": 0.369363,
        "threshold": 0.077476,
    },

    ("BLUED", "LSTM Autoencoder"): {
        "precision": 0.593220,
        "recall": 0.311111,
        "f1_score": 0.408163,
        "false_positive_rate": 0.026454,
        "roc_auc": 0.752145,
        "pr_auc": 0.392771,
        "threshold": 0.093733,
    },
}


historical_rows = []

for (
    dataset_name,
    model_name,
), metric_dict in HISTORICAL_RESULTS.items():

    for (
        metric,
        historical_value,
    ) in metric_dict.items():

        current_mean = (
            get_summary_value(
                dataset_name,
                model_name,
                metric,
                "mean",
            )
        )

        current_std = (
            get_summary_value(
                dataset_name,
                model_name,
                metric,
                "sample_std",
            )
        )

        current_min = (
            get_summary_value(
                dataset_name,
                model_name,
                metric,
                "minimum",
            )
        )

        current_max = (
            get_summary_value(
                dataset_name,
                model_name,
                metric,
                "maximum",
            )
        )

        historical_rows.append({
            "dataset":
                dataset_name,

            "model":
                model_name,

            "metric":
                metric,

            "historical_single_seed":
                historical_value,

            "current_five_seed_mean":
                current_mean,

            "current_five_seed_sample_std":
                current_std,

            "current_five_seed_minimum":
                current_min,

            "current_five_seed_maximum":
                current_max,

            "historical_inside_current_seed_range":
                bool(
                    current_min
                    <= historical_value
                    <= current_max
                ),

            "interpretation":
                (
                    "Context only; the current seed range "
                    "is not a confidence interval and the "
                    "historical run used a different runtime context."
                ),
        })


historical_context_df = pd.DataFrame(
    historical_rows
)


# ---------------------------------------------------------------------
# 8. Dense-versus-LSTM comparisons within each dataset
# ---------------------------------------------------------------------

comparison_rows = []

for dataset_name in [
    "NAB",
    "BLUED",
]:

    for metric in [
        "f1_score",
        "roc_auc",
        "pr_auc",
        "false_positive_rate",
    ]:

        dense_mean = (
            get_summary_value(
                dataset_name,
                "Dense Autoencoder",
                metric,
                "mean",
            )
        )

        lstm_mean = (
            get_summary_value(
                dataset_name,
                "LSTM Autoencoder",
                metric,
                "mean",
            )
        )

        comparison_rows.append({
            "dataset":
                dataset_name,

            "metric":
                metric,

            "dense_ae_mean":
                dense_mean,

            "lstm_ae_mean":
                lstm_mean,

            "dense_minus_lstm":
                dense_mean
                - lstm_mean,
        })


dense_vs_lstm_df = pd.DataFrame(
    comparison_rows
)


# ---------------------------------------------------------------------
# 9. Save master artifacts
# ---------------------------------------------------------------------

ALL_RUNS_PATH = (
    MASTER_ROBUSTNESS_ROOT
    / "all_20_neural_seed_test_runs.csv"
)

MASTER_SUMMARY_PATH = (
    MASTER_ROBUSTNESS_ROOT
    / "master_neural_robustness_summary.csv"
)

COMPACT_SUMMARY_PATH = (
    MASTER_ROBUSTNESS_ROOT
    / "thesis_ready_neural_robustness_summary.csv"
)

HISTORICAL_CONTEXT_PATH = (
    MASTER_ROBUSTNESS_ROOT
    / "historical_single_seed_context.csv"
)

DENSE_VS_LSTM_PATH = (
    MASTER_ROBUSTNESS_ROOT
    / "dense_vs_lstm_mean_comparison.csv"
)

COMPLETION_PATH = (
    MASTER_ROBUSTNESS_ROOT
    / "five_seed_completion_audit.csv"
)


all_runs_df.to_csv(
    ALL_RUNS_PATH,
    index=False,
)

master_summary_df.to_csv(
    MASTER_SUMMARY_PATH,
    index=False,
)

compact_summary_df.to_csv(
    COMPACT_SUMMARY_PATH,
    index=False,
)

historical_context_df.to_csv(
    HISTORICAL_CONTEXT_PATH,
    index=False,
)

dense_vs_lstm_df.to_csv(
    DENSE_VS_LSTM_PATH,
    index=False,
)

completion_df.to_csv(
    COMPLETION_PATH,
    index=False,
)


# ---------------------------------------------------------------------
# 10. Print concise thesis-ready table
# ---------------------------------------------------------------------

print(
    "=" * 130
)

print(
    "MASTER FIVE-SEED NEURAL ROBUSTNESS SUMMARY"
)

print(
    "=" * 130
)

print(
    "\nCompletion audit:"
)

print(
    completion_df.to_string(
        index=False
    )
)


print(
    "\nThesis-ready primary metrics:"
)

for _, row in (
    compact_summary_df.iterrows()
):

    print(
        "\n"
        f"{row['dataset']} — "
        f"{row['model']}"
    )

    print(
        f"  Semantics: "
        f"{row['evaluation_semantics']}"
    )

    print(
        f"  Precision: "
        f"{row['precision_mean']:.6f} "
        f"± "
        f"{row['precision_sample_std']:.6f}"
    )

    print(
        f"  Recall:    "
        f"{row['recall_mean']:.6f} "
        f"± "
        f"{row['recall_sample_std']:.6f}"
    )

    print(
        f"  F1:        "
        f"{row['f1_score_mean']:.6f} "
        f"± "
        f"{row['f1_score_sample_std']:.6f}"
    )

    print(
        f"  F1 range:  "
        f"[{row['f1_minimum']:.6f}, "
        f"{row['f1_maximum']:.6f}]"
    )

    print(
        f"  F1 CV:     "
        f"{row['f1_cv_percent']:.2f}%"
    )

    print(
        f"  FPR:       "
        f"{row['false_positive_rate_mean']:.6f} "
        f"± "
        f"{row['false_positive_rate_sample_std']:.6f}"
    )

    print(
        f"  ROC-AUC:   "
        f"{row['roc_auc_mean']:.6f} "
        f"± "
        f"{row['roc_auc_sample_std']:.6f}"
    )

    print(
        f"  PR-AUC:    "
        f"{row['pr_auc_mean']:.6f} "
        f"± "
        f"{row['pr_auc_sample_std']:.6f}"
    )

    print(
        f"  Threshold range: "
        f"[{row['threshold_minimum']:.6f}, "
        f"{row['threshold_maximum']:.6f}]"
    )

    print(
        f"  Epoch range: "
        f"[{int(row['epochs_minimum'])}, "
        f"{int(row['epochs_maximum'])}]"
    )


# ---------------------------------------------------------------------
# 11. Print the central cross-model finding
# ---------------------------------------------------------------------

print(
    "\n" + "=" * 130
)

print(
    "DENSE AE VS LSTM AE — FIVE-SEED MEAN DIFFERENCES"
)

print(
    "=" * 130
)

print(
    dense_vs_lstm_df
    .round(6)
    .to_string(
        index=False
    )
)


print(
    "\nInterpretive boundary:"
)

print(
    "- These standard deviations describe stochastic "
    "training variability across the five predetermined seeds."
)

print(
    "- They are not confidence intervals, statistical "
    "significance tests, temporal uncertainty estimates, "
    "or evidence of real-world accuracy."
)

print(
    "- BLUED metrics quantify switching-event alignment; "
    "they must not be described as anomaly-detection accuracy."
)


print(
    "\nSaved master artifacts:"
)

for path in [
    ALL_RUNS_PATH,
    MASTER_SUMMARY_PATH,
    COMPACT_SUMMARY_PATH,
    HISTORICAL_CONTEXT_PATH,
    DENSE_VS_LSTM_PATH,
    COMPLETION_PATH,
]:

    print(
        f"  {path}"
    )


print(
    "\nNo models were trained in Cell 24."
)
```

```python
# Cell 25 — Freeze Simple Statistical Baseline Specification:
# Predeclare one deterministic maximum-absolute-z-score statistical anomaly baseline
# before any validation threshold or test performance is calculated. Fit feature
# location/scale statistics from the training partitions only and save the complete
# specification and training-statistics audit for NAB and BLUED.

import json
from pathlib import Path

import numpy as np
import pandas as pd


# ---------------------------------------------------------------------
# 1. Create baseline workspace
# ---------------------------------------------------------------------

STATISTICAL_BASELINE_ROOT = (
    BASELINE_ROOT
    / "MAX_ABSOLUTE_Z_SCORE"
)

STATISTICAL_BASELINE_ROOT.mkdir(
    parents=True,
    exist_ok=True,
)


# ---------------------------------------------------------------------
# 2. Freeze the baseline formulation
# ---------------------------------------------------------------------

BASELINE_EPSILON = 1e-12
BASELINE_THRESHOLD_PERCENTILE = 95.0

BASELINE_SPECIFICATION = {
    "baseline_name":
        "Maximum Absolute Z-Score",

    "baseline_family":
        "Simple deterministic statistical anomaly baseline",

    "purpose":
        (
            "Provide a simple non-learned statistical reference "
            "under the same frozen train/validation/test populations "
            "and validation-percentile threshold protocol used by "
            "the label-blind anomaly-detection experiments."
        ),

    "datasets": {
        "NAB": {
            "representation":
                "Final corrected 23-feature window representation",

            "training_rows":
                int(
                    len(
                        nab_dense_X_train
                    )
                ),

            "validation_rows":
                int(
                    len(
                        nab_dense_X_validation
                    )
                ),

            "test_rows":
                int(
                    len(
                        nab_dense_X_test
                    )
                ),

            "evaluation_semantics":
                "Anomaly-window evaluation",
        },

        "BLUED": {
            "representation":
                "Final 69-feature window representation",

            "training_rows":
                int(
                    len(
                        blued_X_train
                    )
                ),

            "validation_rows":
                int(
                    len(
                        blued_X_validation
                    )
                ),

            "test_rows":
                int(
                    len(
                        blued_X_test
                    )
                ),

            "evaluation_semantics":
                (
                    "Switching-event alignment; BLUED event labels "
                    "are not treated as confirmed anomaly labels."
                ),
        },
    },

    "training_statistics": {
        "location":
            "Per-feature arithmetic mean from training partition only",

        "scale":
            (
                "Per-feature population standard deviation "
                "(numpy std with ddof=0) from training partition only"
            ),

        "zero_variance_rule":
            (
                "Features with training standard deviation <= 1e-12 "
                "are excluded from the anomaly-score maximum."
            ),

        "epsilon":
            BASELINE_EPSILON,
    },

    "window_score": {
        "feature_score":
            "absolute((x_j - training_mean_j) / training_std_j)",

        "aggregation":
            (
                "Maximum absolute standardized deviation across all "
                "retained non-zero-variance features"
            ),

        "direction":
            "Higher score indicates greater statistical deviation",
    },

    "threshold_protocol": {
        "source":
            "Validation anomaly scores only",

        "percentile":
            BASELINE_THRESHOLD_PERCENTILE,

        "prediction_operator":
            "score > threshold",

        "validation_labels_used":
            False,

        "test_labels_used_for_selection":
            False,
    },

    "labels": {
        "used_for_training":
            False,

        "used_for_feature_statistics":
            False,

        "used_for_threshold_selection":
            False,

        "used_for_final_evaluation":
            True,
    },

    "model_selection": {
        "alternative_baselines_compared_before_test":
            False,

        "hyperparameter_tuning":
            False,

        "test_set_tuning":
            False,

        "note":
            (
                "This single baseline formulation is frozen before "
                "any baseline validation threshold or test metric is calculated."
            ),
    },

    "comparison_boundary":
        (
            "The baseline operates on engineered feature vectors. "
            "It is therefore representation-matched to the feature-based "
            "pipeline and Dense Autoencoder, but does not use the "
            "60-step sequential representation of the LSTM Autoencoder."
        ),

    "interpretive_boundary":
        (
            "BLUED baseline metrics quantify correspondence with annotated "
            "switching-event windows and must not be described as "
            "confirmed anomaly-detection accuracy."
        ),
}


# ---------------------------------------------------------------------
# 3. Save frozen specification BEFORE fitting statistics
# ---------------------------------------------------------------------

BASELINE_SPEC_PATH = (
    STATISTICAL_BASELINE_ROOT
    / "FROZEN_MAX_ABSOLUTE_Z_SCORE_SPECIFICATION.json"
)

with open(
    BASELINE_SPEC_PATH,
    "w",
    encoding="utf-8",
) as file:

    json.dump(
        BASELINE_SPECIFICATION,
        file,
        indent=2,
    )


# ---------------------------------------------------------------------
# 4. Training-only statistics helper
# ---------------------------------------------------------------------

def fit_statistical_baseline_statistics(
    X_train,
    epsilon=1e-12,
):
    """
    Fit per-feature mean and population standard deviation using
    the training partition only. No labels are used.
    """

    X_train = np.asarray(
        X_train,
        dtype=np.float64,
    )

    feature_mean = np.mean(
        X_train,
        axis=0,
    )

    feature_std = np.std(
        X_train,
        axis=0,
        ddof=0,
    )

    retained_mask = (
        feature_std
        > epsilon
    )

    return (
        feature_mean,
        feature_std,
        retained_mask,
    )


# ---------------------------------------------------------------------
# 5. Fit NAB statistics from TRAINING DATA ONLY
# ---------------------------------------------------------------------

(
    nab_baseline_mean,
    nab_baseline_std,
    nab_baseline_retained_mask,
) = fit_statistical_baseline_statistics(
    nab_dense_X_train,
    epsilon=BASELINE_EPSILON,
)


# ---------------------------------------------------------------------
# 6. Fit BLUED statistics from TRAINING DATA ONLY
# ---------------------------------------------------------------------

(
    blued_baseline_mean,
    blued_baseline_std,
    blued_baseline_retained_mask,
) = fit_statistical_baseline_statistics(
    blued_X_train,
    epsilon=BASELINE_EPSILON,
)


# ---------------------------------------------------------------------
# 7. Sanity checks on training-only statistics
# ---------------------------------------------------------------------

assert nab_baseline_mean.shape == (23,)
assert nab_baseline_std.shape == (23,)
assert nab_baseline_retained_mask.shape == (23,)

assert blued_baseline_mean.shape == (69,)
assert blued_baseline_std.shape == (69,)
assert blued_baseline_retained_mask.shape == (69,)

if not np.isfinite(
    nab_baseline_mean
).all():
    raise ValueError(
        "Non-finite NAB baseline means detected."
    )

if not np.isfinite(
    nab_baseline_std
).all():
    raise ValueError(
        "Non-finite NAB baseline standard deviations detected."
    )

if not np.isfinite(
    blued_baseline_mean
).all():
    raise ValueError(
        "Non-finite BLUED baseline means detected."
    )

if not np.isfinite(
    blued_baseline_std
).all():
    raise ValueError(
        "Non-finite BLUED baseline standard deviations detected."
    )

if int(
    nab_baseline_retained_mask.sum()
) == 0:
    raise ValueError(
        "NAB baseline retained zero usable features."
    )

if int(
    blued_baseline_retained_mask.sum()
) == 0:
    raise ValueError(
        "BLUED baseline retained zero usable features."
    )


# ---------------------------------------------------------------------
# 8. Save exact training statistics
# ---------------------------------------------------------------------

NAB_STATS_PATH = (
    STATISTICAL_BASELINE_ROOT
    / "nab_training_only_zscore_statistics.npz"
)

BLUED_STATS_PATH = (
    STATISTICAL_BASELINE_ROOT
    / "blued_training_only_zscore_statistics.npz"
)


np.savez_compressed(
    NAB_STATS_PATH,

    feature_mean=
        nab_baseline_mean,

    feature_std=
        nab_baseline_std,

    retained_mask=
        nab_baseline_retained_mask,

    epsilon=np.array(
        BASELINE_EPSILON,
        dtype=np.float64,
    ),
)


np.savez_compressed(
    BLUED_STATS_PATH,

    feature_mean=
        blued_baseline_mean,

    feature_std=
        blued_baseline_std,

    retained_mask=
        blued_baseline_retained_mask,

    epsilon=np.array(
        BASELINE_EPSILON,
        dtype=np.float64,
    ),
)


# ---------------------------------------------------------------------
# 9. Create feature-statistics audit tables
# ---------------------------------------------------------------------

nab_stats_df = pd.DataFrame({
    "feature_position":
        np.arange(
            23
        ),

    "training_mean":
        nab_baseline_mean,

    "training_std_ddof0":
        nab_baseline_std,

    "retained":
        nab_baseline_retained_mask,
})


blued_stats_df = pd.DataFrame({
    "feature_position":
        np.arange(
            69
        ),

    "training_mean":
        blued_baseline_mean,

    "training_std_ddof0":
        blued_baseline_std,

    "retained":
        blued_baseline_retained_mask,
})


NAB_AUDIT_PATH = (
    STATISTICAL_BASELINE_ROOT
    / "nab_training_statistics_audit.csv"
)

BLUED_AUDIT_PATH = (
    STATISTICAL_BASELINE_ROOT
    / "blued_training_statistics_audit.csv"
)


nab_stats_df.to_csv(
    NAB_AUDIT_PATH,
    index=False,
)

blued_stats_df.to_csv(
    BLUED_AUDIT_PATH,
    index=False,
)


# ---------------------------------------------------------------------
# 10. Freeze an execution manifest
# ---------------------------------------------------------------------

execution_manifest = {
    "baseline":
        "Maximum Absolute Z-Score",

    "specification_frozen":
        True,

    "test_performance_observed_before_freeze":
        False,

    "validation_performance_observed_before_freeze":
        False,

    "nab_total_features":
        23,

    "nab_retained_features":
        int(
            nab_baseline_retained_mask.sum()
        ),

    "nab_excluded_zero_variance_features":
        int(
            23
            - nab_baseline_retained_mask.sum()
        ),

    "blued_total_features":
        69,

    "blued_retained_features":
        int(
            blued_baseline_retained_mask.sum()
        ),

    "blued_excluded_zero_variance_features":
        int(
            69
            - blued_baseline_retained_mask.sum()
        ),

    "validation_threshold_percentile":
        BASELINE_THRESHOLD_PERCENTILE,

    "prediction_operator":
        ">",

    "ready_for_evaluation":
        True,
}


BASELINE_MANIFEST_PATH = (
    STATISTICAL_BASELINE_ROOT
    / "BASELINE_EXECUTION_MANIFEST.json"
)

with open(
    BASELINE_MANIFEST_PATH,
    "w",
    encoding="utf-8",
) as file:

    json.dump(
        execution_manifest,
        file,
        indent=2,
    )


# ---------------------------------------------------------------------
# 11. Final report
# ---------------------------------------------------------------------

print(
    "=" * 125
)

print(
    "SIMPLE STATISTICAL BASELINE SPECIFICATION FROZEN"
)

print(
    "=" * 125
)


print(
    "\nBaseline:"
)

print(
    "  Maximum Absolute Z-Score"
)


print(
    "\nScore definition:"
)

print(
    "  Fit per-feature mean and population SD on training data only."
)

print(
    "  Standardize each window feature using those frozen statistics."
)

print(
    "  Window score = maximum absolute z-score across retained features."
)


print(
    "\nThreshold protocol:"
)

print(
    "  q95 of validation scores only"
)

print(
    "  Prediction rule: score > threshold"
)

print(
    "  No validation labels used for threshold selection."
)


print(
    "\nNAB feature audit:"
)

print(
    f"  Total features:                 23"
)

print(
    f"  Retained features:              "
    f"{int(nab_baseline_retained_mask.sum())}"
)

print(
    f"  Zero-variance features excluded: "
    f"{int(23 - nab_baseline_retained_mask.sum())}"
)


print(
    "\nBLUED feature audit:"
)

print(
    f"  Total features:                 69"
)

print(
    f"  Retained features:              "
    f"{int(blued_baseline_retained_mask.sum())}"
)

print(
    f"  Zero-variance features excluded: "
    f"{int(69 - blued_baseline_retained_mask.sum())}"
)


print(
    "\nInterpretive boundary:"
)

print(
    "  NAB: anomaly-window evaluation."
)

print(
    "  BLUED: switching-event alignment only; "
    "not confirmed anomaly accuracy."
)


print(
    "\nFrozen artifacts:"
)

for path in [
    BASELINE_SPEC_PATH,
    NAB_STATS_PATH,
    BLUED_STATS_PATH,
    NAB_AUDIT_PATH,
    BLUED_AUDIT_PATH,
    BASELINE_MANIFEST_PATH,
]:

    print(
        f"  {path}"
    )


print(
    "\nNo validation anomaly scores, test anomaly scores, "
    "thresholds, or performance metrics were calculated in Cell 25."
)
```

```python
# Cell 26 — Execute Frozen Statistical Baseline on NAB and BLUED:
# Apply the predeclared maximum-absolute-z-score baseline exactly once using
# training-only feature statistics, derive each dataset's q95 threshold from
# validation scores only, evaluate the frozen test partitions, and save all
# row-level scores and thesis-ready baseline metrics without any tuning.

import json
import numpy as np
import pandas as pd


# ---------------------------------------------------------------------
# 1. Reload frozen specification and training-only statistics
# ---------------------------------------------------------------------

with open(
    BASELINE_SPEC_PATH,
    "r",
    encoding="utf-8",
) as file:
    frozen_baseline_spec = json.load(file)


with np.load(
    NAB_STATS_PATH,
    allow_pickle=False,
) as data:

    nab_baseline_mean = data[
        "feature_mean"
    ].copy()

    nab_baseline_std = data[
        "feature_std"
    ].copy()

    nab_baseline_retained_mask = data[
        "retained_mask"
    ].astype(bool)


with np.load(
    BLUED_STATS_PATH,
    allow_pickle=False,
) as data:

    blued_baseline_mean = data[
        "feature_mean"
    ].copy()

    blued_baseline_std = data[
        "feature_std"
    ].copy()

    blued_baseline_retained_mask = data[
        "retained_mask"
    ].astype(bool)


# ---------------------------------------------------------------------
# 2. Verify the frozen specification before evaluation
# ---------------------------------------------------------------------

assert frozen_baseline_spec[
    "baseline_name"
] == "Maximum Absolute Z-Score"

assert frozen_baseline_spec[
    "threshold_protocol"
][
    "percentile"
] == 95.0

assert frozen_baseline_spec[
    "threshold_protocol"
][
    "prediction_operator"
] == "score > threshold"

assert int(
    nab_baseline_retained_mask.sum()
) == 23

assert int(
    blued_baseline_retained_mask.sum()
) == 66


# ---------------------------------------------------------------------
# 3. Define the frozen anomaly-score calculation
# ---------------------------------------------------------------------

def maximum_absolute_zscore_scores(
    X,
    training_mean,
    training_std,
    retained_mask,
):
    """
    Frozen baseline:
    abs z-score per retained feature,
    followed by maximum across retained features.
    """

    X = np.asarray(
        X,
        dtype=np.float64,
    )

    training_mean = np.asarray(
        training_mean,
        dtype=np.float64,
    )

    training_std = np.asarray(
        training_std,
        dtype=np.float64,
    )

    retained_mask = np.asarray(
        retained_mask,
        dtype=bool,
    )

    retained_X = X[
        :,
        retained_mask,
    ]

    retained_mean = training_mean[
        retained_mask
    ]

    retained_std = training_std[
        retained_mask
    ]

    absolute_z = np.abs(
        (
            retained_X
            - retained_mean
        )
        / retained_std
    )

    scores = np.max(
        absolute_z,
        axis=1,
    )

    # Useful audit field:
    # which retained feature generated the maximum score?
    local_argmax = np.argmax(
        absolute_z,
        axis=1,
    )

    retained_original_positions = np.flatnonzero(
        retained_mask
    )

    maximum_feature_position = (
        retained_original_positions[
            local_argmax
        ]
    )

    return (
        scores.astype(
            np.float64
        ),
        maximum_feature_position.astype(
            np.int32
        ),
    )


# ---------------------------------------------------------------------
# 4. NAB validation scores and frozen q95 threshold
# ---------------------------------------------------------------------

(
    nab_validation_scores,
    nab_validation_max_feature,
) = maximum_absolute_zscore_scores(
    X=nab_dense_X_validation,
    training_mean=nab_baseline_mean,
    training_std=nab_baseline_std,
    retained_mask=nab_baseline_retained_mask,
)


NAB_BASELINE_THRESHOLD = (
    derive_validation_threshold(
        nab_validation_scores,
        percentile=95.0,
    )
)


nab_validation_predictions = (
    nab_validation_scores
    > NAB_BASELINE_THRESHOLD
).astype(np.int8)


nab_validation_metrics = (
    calculate_binary_metrics(
        y_true=nab_dense_y_validation,
        anomaly_scores=nab_validation_scores,
        predictions=nab_validation_predictions,
    )
)


# ---------------------------------------------------------------------
# 5. NAB test evaluation using frozen validation threshold
# ---------------------------------------------------------------------

(
    nab_test_scores,
    nab_test_max_feature,
) = maximum_absolute_zscore_scores(
    X=nab_dense_X_test,
    training_mean=nab_baseline_mean,
    training_std=nab_baseline_std,
    retained_mask=nab_baseline_retained_mask,
)


nab_test_predictions = (
    nab_test_scores
    > NAB_BASELINE_THRESHOLD
).astype(np.int8)


nab_test_metrics = (
    calculate_binary_metrics(
        y_true=nab_dense_y_test,
        anomaly_scores=nab_test_scores,
        predictions=nab_test_predictions,
    )
)


# ---------------------------------------------------------------------
# 6. BLUED validation scores and frozen q95 threshold
# ---------------------------------------------------------------------

(
    blued_validation_scores,
    blued_validation_max_feature,
) = maximum_absolute_zscore_scores(
    X=blued_X_validation,
    training_mean=blued_baseline_mean,
    training_std=blued_baseline_std,
    retained_mask=blued_baseline_retained_mask,
)


BLUED_BASELINE_THRESHOLD = (
    derive_validation_threshold(
        blued_validation_scores,
        percentile=95.0,
    )
)


blued_validation_predictions = (
    blued_validation_scores
    > BLUED_BASELINE_THRESHOLD
).astype(np.int8)


blued_validation_metrics = (
    calculate_binary_metrics(
        y_true=blued_y_validation,
        anomaly_scores=blued_validation_scores,
        predictions=blued_validation_predictions,
    )
)


# ---------------------------------------------------------------------
# 7. BLUED test event-alignment evaluation
# ---------------------------------------------------------------------

(
    blued_test_scores,
    blued_test_max_feature,
) = maximum_absolute_zscore_scores(
    X=blued_X_test,
    training_mean=blued_baseline_mean,
    training_std=blued_baseline_std,
    retained_mask=blued_baseline_retained_mask,
)


blued_test_predictions = (
    blued_test_scores
    > BLUED_BASELINE_THRESHOLD
).astype(np.int8)


blued_test_metrics = (
    calculate_binary_metrics(
        y_true=blued_y_test,
        anomaly_scores=blued_test_scores,
        predictions=blued_test_predictions,
    )
)


# ---------------------------------------------------------------------
# 8. Build summary result table
# ---------------------------------------------------------------------

baseline_results = [
    {
        "dataset":
            "NAB",

        "evaluation_semantics":
            "anomaly_window_evaluation",

        "baseline":
            "Maximum Absolute Z-Score",

        "split":
            "validation",

        "threshold":
            NAB_BASELINE_THRESHOLD,

        "retained_features":
            int(
                nab_baseline_retained_mask.sum()
            ),

        **nab_validation_metrics,
    },

    {
        "dataset":
            "NAB",

        "evaluation_semantics":
            "anomaly_window_evaluation",

        "baseline":
            "Maximum Absolute Z-Score",

        "split":
            "test",

        "threshold":
            NAB_BASELINE_THRESHOLD,

        "retained_features":
            int(
                nab_baseline_retained_mask.sum()
            ),

        **nab_test_metrics,
    },

    {
        "dataset":
            "BLUED",

        "evaluation_semantics":
            "switching_event_alignment",

        "baseline":
            "Maximum Absolute Z-Score",

        "split":
            "validation",

        "threshold":
            BLUED_BASELINE_THRESHOLD,

        "retained_features":
            int(
                blued_baseline_retained_mask.sum()
            ),

        **blued_validation_metrics,
    },

    {
        "dataset":
            "BLUED",

        "evaluation_semantics":
            "switching_event_alignment",

        "baseline":
            "Maximum Absolute Z-Score",

        "split":
            "test",

        "threshold":
            BLUED_BASELINE_THRESHOLD,

        "retained_features":
            int(
                blued_baseline_retained_mask.sum()
            ),

        **blued_test_metrics,
    },
]


baseline_results_df = pd.DataFrame(
    baseline_results
)


# ---------------------------------------------------------------------
# 9. Save row-level NAB score artifacts
# ---------------------------------------------------------------------

nab_validation_score_df = pd.DataFrame({
    "window_id":
        nab_dense_validation_ids,

    "label":
        nab_dense_y_validation,

    "anomaly_score":
        nab_validation_scores,

    "maximum_feature_position":
        nab_validation_max_feature,

    "threshold":
        NAB_BASELINE_THRESHOLD,

    "flagged":
        nab_validation_predictions,
})


nab_test_score_df = pd.DataFrame({
    "window_id":
        nab_dense_test_ids,

    "label":
        nab_dense_y_test,

    "anomaly_score":
        nab_test_scores,

    "maximum_feature_position":
        nab_test_max_feature,

    "threshold":
        NAB_BASELINE_THRESHOLD,

    "flagged":
        nab_test_predictions,
})


# ---------------------------------------------------------------------
# 10. Save row-level BLUED score artifacts
# ---------------------------------------------------------------------

blued_validation_score_df = pd.DataFrame({
    "model_index":
        blued_validation_indices,

    "event_window_label":
        blued_y_validation,

    "anomaly_score":
        blued_validation_scores,

    "maximum_feature_position":
        blued_validation_max_feature,

    "threshold":
        BLUED_BASELINE_THRESHOLD,

    "flagged":
        blued_validation_predictions,
})


blued_test_score_df = pd.DataFrame({
    "model_index":
        blued_test_indices,

    "event_window_label":
        blued_y_test,

    "anomaly_score":
        blued_test_scores,

    "maximum_feature_position":
        blued_test_max_feature,

    "threshold":
        BLUED_BASELINE_THRESHOLD,

    "flagged":
        blued_test_predictions,
})


# ---------------------------------------------------------------------
# 11. Save baseline artifacts
# ---------------------------------------------------------------------

BASELINE_RESULTS_PATH = (
    STATISTICAL_BASELINE_ROOT
    / "max_absolute_zscore_baseline_results.csv"
)

NAB_VALIDATION_SCORES_PATH = (
    STATISTICAL_BASELINE_ROOT
    / "nab_validation_scores.csv"
)

NAB_TEST_SCORES_PATH = (
    STATISTICAL_BASELINE_ROOT
    / "nab_test_scores.csv"
)

BLUED_VALIDATION_SCORES_PATH = (
    STATISTICAL_BASELINE_ROOT
    / "blued_validation_scores.csv"
)

BLUED_TEST_SCORES_PATH = (
    STATISTICAL_BASELINE_ROOT
    / "blued_test_scores.csv"
)


baseline_results_df.to_csv(
    BASELINE_RESULTS_PATH,
    index=False,
)

nab_validation_score_df.to_csv(
    NAB_VALIDATION_SCORES_PATH,
    index=False,
)

nab_test_score_df.to_csv(
    NAB_TEST_SCORES_PATH,
    index=False,
)

blued_validation_score_df.to_csv(
    BLUED_VALIDATION_SCORES_PATH,
    index=False,
)

blued_test_score_df.to_csv(
    BLUED_TEST_SCORES_PATH,
    index=False,
)


# ---------------------------------------------------------------------
# 12. Save execution metadata
# ---------------------------------------------------------------------

execution_metadata = {
    "baseline":
        "Maximum Absolute Z-Score",

    "formulation_frozen_before_evaluation":
        True,

    "alternative_baselines_tested":
        False,

    "hyperparameter_tuning":
        False,

    "test_set_tuning":
        False,

    "nab": {
        "retained_features":
            int(
                nab_baseline_retained_mask.sum()
            ),

        "validation_threshold":
            float(
                NAB_BASELINE_THRESHOLD
            ),

        "test_f1":
            float(
                nab_test_metrics[
                    "f1_score"
                ]
            ),

        "test_roc_auc":
            float(
                nab_test_metrics[
                    "roc_auc"
                ]
            ),

        "test_pr_auc":
            float(
                nab_test_metrics[
                    "pr_auc"
                ]
            ),
    },

    "blued": {
        "retained_features":
            int(
                blued_baseline_retained_mask.sum()
            ),

        "validation_threshold":
            float(
                BLUED_BASELINE_THRESHOLD
            ),

        "test_event_alignment_f1":
            float(
                blued_test_metrics[
                    "f1_score"
                ]
            ),

        "test_roc_auc":
            float(
                blued_test_metrics[
                    "roc_auc"
                ]
            ),

        "test_pr_auc":
            float(
                blued_test_metrics[
                    "pr_auc"
                ]
            ),

        "semantic_warning":
            (
                "BLUED labels denote switching-event "
                "windows, not confirmed anomaly labels."
            ),
    },
}


BASELINE_EXECUTION_PATH = (
    STATISTICAL_BASELINE_ROOT
    / "BASELINE_EXECUTION_RESULTS.json"
)


with open(
    BASELINE_EXECUTION_PATH,
    "w",
    encoding="utf-8",
) as file:

    json.dump(
        execution_metadata,
        file,
        indent=2,
    )


# ---------------------------------------------------------------------
# 13. Print thesis-relevant results
# ---------------------------------------------------------------------

print(
    "=" * 125
)

print(
    "FROZEN SIMPLE STATISTICAL BASELINE — EVALUATION COMPLETE"
)

print(
    "=" * 125
)


print(
    "\nNAB — Maximum Absolute Z-Score"
)

print(
    f"  Validation q95 threshold: "
    f"{NAB_BASELINE_THRESHOLD:.6f}"
)

print(
    f"  Test Precision: "
    f"{nab_test_metrics['precision']:.6f}"
)

print(
    f"  Test Recall: "
    f"{nab_test_metrics['recall']:.6f}"
)

print(
    f"  Test F1: "
    f"{nab_test_metrics['f1_score']:.6f}"
)

print(
    f"  Test FPR: "
    f"{nab_test_metrics['false_positive_rate']:.6f}"
)

print(
    f"  Test ROC-AUC: "
    f"{nab_test_metrics['roc_auc']:.6f}"
)

print(
    f"  Test PR-AUC: "
    f"{nab_test_metrics['pr_auc']:.6f}"
)

print(
    f"  Test flagged ratio: "
    f"{nab_test_metrics['flagged_ratio_percent']:.3f}%"
)


print(
    "\nBLUED — Maximum Absolute Z-Score"
)

print(
    "  IMPORTANT: metrics below measure switching-event "
    "alignment, not confirmed anomaly accuracy."
)

print(
    f"  Validation q95 threshold: "
    f"{BLUED_BASELINE_THRESHOLD:.6f}"
)

print(
    f"  Test Event-Alignment Precision: "
    f"{blued_test_metrics['precision']:.6f}"
)

print(
    f"  Test Event-Alignment Recall: "
    f"{blued_test_metrics['recall']:.6f}"
)

print(
    f"  Test Event-Alignment F1: "
    f"{blued_test_metrics['f1_score']:.6f}"
)

print(
    f"  Test FPR: "
    f"{blued_test_metrics['false_positive_rate']:.6f}"
)

print(
    f"  Test ROC-AUC: "
    f"{blued_test_metrics['roc_auc']:.6f}"
)

print(
    f"  Test PR-AUC: "
    f"{blued_test_metrics['pr_auc']:.6f}"
)

print(
    f"  Test flagged ratio: "
    f"{blued_test_metrics['flagged_ratio_percent']:.3f}%"
)


print(
    "\nSaved artifacts:"
)

for path in [
    BASELINE_RESULTS_PATH,
    NAB_VALIDATION_SCORES_PATH,
    NAB_TEST_SCORES_PATH,
    BLUED_VALIDATION_SCORES_PATH,
    BLUED_TEST_SCORES_PATH,
    BASELINE_EXECUTION_PATH,
]:

    print(
        f"  {path}"
    )


print(
    "\nThe baseline formulation was not modified after "
    "validation or test performance was observed."
)
```

```python
# Cell 28 — Integrate Statistical Baseline with Authoritative Model Results and Seed Robustness:
# Combine the frozen simple statistical baseline with the authoritative corrected NAB
# and final BLUED five-model tables, while preserving the original thesis results as
# the primary model-comparison values and attaching five-seed neural robustness only
# as supplementary stochastic-variability evidence.

import numpy as np
import pandas as pd
from pathlib import Path


# ---------------------------------------------------------------------
# 1. Authoritative source files identified by Cell 27
# ---------------------------------------------------------------------

NAB_AUTHORITATIVE_PATH = (
    CODE_ROOT
    / "experiments"
    / "EXP_SPLIT_BOUNDARY_PURGE_CORRECTION"
    / "NAB"
    / "FIVE_MODEL_CORRECTION"
    / "FINAL_NAB_FIVE_MODEL_PUBLICATION_TABLE_CORRECTED.csv"
)

BLUED_AUTHORITATIVE_PATH = (
    CODE_ROOT
    / "experiments"
    / "EXP_BLUED_FIVE_MODEL_LABEL_BLIND"
    / "final_five_model_comparison"
    / "publication_outputs"
    / "tables"
    / "blued_five_model_complete_results.csv"
)

BASELINE_RESULTS_PATH = (
    STATISTICAL_BASELINE_ROOT
    / "max_absolute_zscore_baseline_results.csv"
)

ROBUSTNESS_SUMMARY_PATH = (
    MASTER_ROBUSTNESS_ROOT
    / "thesis_ready_neural_robustness_summary.csv"
)


# ---------------------------------------------------------------------
# 2. Verify source files exist
# ---------------------------------------------------------------------

for path in [
    NAB_AUTHORITATIVE_PATH,
    BLUED_AUTHORITATIVE_PATH,
    BASELINE_RESULTS_PATH,
    ROBUSTNESS_SUMMARY_PATH,
]:

    if not path.exists():
        raise FileNotFoundError(
            f"Required source file not found:\n{path}"
        )


# ---------------------------------------------------------------------
# 3. Load source tables
# ---------------------------------------------------------------------

nab_authoritative = pd.read_csv(
    NAB_AUTHORITATIVE_PATH
)

blued_authoritative = pd.read_csv(
    BLUED_AUTHORITATIVE_PATH
)

baseline_results = pd.read_csv(
    BASELINE_RESULTS_PATH
)

robustness_summary = pd.read_csv(
    ROBUSTNESS_SUMMARY_PATH
)


# ---------------------------------------------------------------------
# 4. Verify authoritative model sets
# ---------------------------------------------------------------------

EXPECTED_MODELS = {
    "One-Class SVM",
    "Isolation Forest",
    "Dense Autoencoder",
    "Local Outlier Factor",
    "LSTM Autoencoder",
}

observed_nab_models = set(
    nab_authoritative[
        "Model"
    ].tolist()
)

observed_blued_models = set(
    blued_authoritative[
        "model"
    ].tolist()
)

if observed_nab_models != EXPECTED_MODELS:
    raise ValueError(
        "Unexpected NAB authoritative model set.\n"
        f"Observed: {observed_nab_models}"
    )

if observed_blued_models != EXPECTED_MODELS:
    raise ValueError(
        "Unexpected BLUED authoritative model set.\n"
        f"Observed: {observed_blued_models}"
    )


# ---------------------------------------------------------------------
# 5. Representation labels
# ---------------------------------------------------------------------

NAB_REPRESENTATION = {
    "One-Class SVM":
        "23 engineered window features",

    "Isolation Forest":
        "23 engineered window features",

    "Dense Autoencoder":
        "23 engineered window features",

    "Local Outlier Factor":
        "23 engineered window features",

    "LSTM Autoencoder":
        "60-step single-channel temporal sequence",

    "Maximum Absolute Z-Score":
        "23 engineered window features",
}


# ---------------------------------------------------------------------
# 6. Convert authoritative NAB results to common schema
# ---------------------------------------------------------------------

nab_rows = []

for _, row in nab_authoritative.iterrows():

    nab_rows.append({
        "dataset":
            "NAB",

        "evaluation_semantics":
            "Anomaly-window evaluation",

        "method":
            row["Model"],

        "input_representation":
            NAB_REPRESENTATION[
                row["Model"]
            ],

        "result_basis":
            "Authoritative corrected thesis result",

        "precision":
            float(
                row["Precision"]
            ),

        "recall":
            float(
                row["Recall"]
            ),

        "f1_score":
            float(
                row["F1-Score"]
            ),

        "false_positive_rate":
            float(
                row["FPR"]
            ),

        "roc_auc":
            float(
                row["ROC-AUC"]
            ),

        "pr_auc":
            float(
                row["PR-AUC"]
            ),

        "flagged_windows":
            int(
                row["Flagged Windows"]
            ),

        "flagged_ratio_percent":
            float(
                row["Flagged Ratio (%)"]
            ),

        "threshold":
            float(
                row["Threshold"]
            ),
    })


# ---------------------------------------------------------------------
# 7. Convert authoritative BLUED results to common schema
# ---------------------------------------------------------------------

blued_rows = []

for _, row in blued_authoritative.iterrows():

    blued_rows.append({
        "dataset":
            "BLUED",

        "evaluation_semantics":
            "Switching-event alignment",

        "method":
            row["model"],

        "input_representation":
            row[
                "input_representation"
            ],

        "result_basis":
            "Authoritative final thesis result",

        "precision":
            float(
                row["precision"]
            ),

        "recall":
            float(
                row["recall"]
            ),

        "f1_score":
            float(
                row["f1_score"]
            ),

        "false_positive_rate":
            float(
                row["false_positive_rate"]
            ),

        "roc_auc":
            float(
                row["roc_auc"]
            ),

        "pr_auc":
            float(
                row["pr_auc"]
            ),

        "flagged_windows":
            int(
                row["flagged_windows"]
            ),

        "flagged_ratio_percent":
            float(
                row["flagged_ratio_percent"]
            ),

        "threshold":
            np.nan,
    })


# ---------------------------------------------------------------------
# 8. Add the frozen deterministic statistical baseline
# ---------------------------------------------------------------------

baseline_test = (
    baseline_results[
        baseline_results[
            "split"
        ] == "test"
    ]
    .copy()
)


if set(
    baseline_test["dataset"]
) != {
    "NAB",
    "BLUED",
}:

    raise ValueError(
        "Expected exactly one NAB and one BLUED "
        "test baseline result."
    )


for _, row in baseline_test.iterrows():

    dataset_name = row[
        "dataset"
    ]

    if dataset_name == "NAB":

        representation = (
            "23 engineered window features"
        )

        semantics = (
            "Anomaly-window evaluation"
        )

    elif dataset_name == "BLUED":

        representation = (
            "69 engineered window features "
            "(66 non-zero-variance features retained)"
        )

        semantics = (
            "Switching-event alignment"
        )

    else:

        raise ValueError(
            f"Unexpected baseline dataset: "
            f"{dataset_name}"
        )


    new_row = {
        "dataset":
            dataset_name,

        "evaluation_semantics":
            semantics,

        "method":
            "Maximum Absolute Z-Score",

        "input_representation":
            representation,

        "result_basis":
            (
                "Predeclared deterministic "
                "statistical baseline"
            ),

        "precision":
            float(
                row["precision"]
            ),

        "recall":
            float(
                row["recall"]
            ),

        "f1_score":
            float(
                row["f1_score"]
            ),

        "false_positive_rate":
            float(
                row[
                    "false_positive_rate"
                ]
            ),

        "roc_auc":
            float(
                row["roc_auc"]
            ),

        "pr_auc":
            float(
                row["pr_auc"]
            ),

        "flagged_windows":
            int(
                row["flagged_windows"]
            ),

        "flagged_ratio_percent":
            float(
                row[
                    "flagged_ratio_percent"
                ]
            ),

        "threshold":
            float(
                row["threshold"]
            ),
    }


    if dataset_name == "NAB":

        nab_rows.append(
            new_row
        )

    else:

        blued_rows.append(
            new_row
        )


# ---------------------------------------------------------------------
# 9. Create integrated six-method tables
# ---------------------------------------------------------------------

nab_integrated_df = pd.DataFrame(
    nab_rows
)

blued_integrated_df = pd.DataFrame(
    blued_rows
)


nab_integrated_df = (
    nab_integrated_df
    .sort_values(
        "f1_score",
        ascending=False,
    )
    .reset_index(
        drop=True
    )
)

blued_integrated_df = (
    blued_integrated_df
    .sort_values(
        "f1_score",
        ascending=False,
    )
    .reset_index(
        drop=True
    )
)


nab_integrated_df.insert(
    2,
    "f1_rank",
    np.arange(
        1,
        len(
            nab_integrated_df
        ) + 1,
    ),
)

blued_integrated_df.insert(
    2,
    "f1_rank",
    np.arange(
        1,
        len(
            blued_integrated_df
        ) + 1,
    ),
)


# ---------------------------------------------------------------------
# 10. Add five-seed robustness context for neural methods only
# ---------------------------------------------------------------------

ROBUSTNESS_COLUMNS = {
    "precision_mean":
        "seed_precision_mean",

    "precision_sample_std":
        "seed_precision_sample_std",

    "recall_mean":
        "seed_recall_mean",

    "recall_sample_std":
        "seed_recall_sample_std",

    "f1_score_mean":
        "seed_f1_mean",

    "f1_score_sample_std":
        "seed_f1_sample_std",

    "f1_minimum":
        "seed_f1_minimum",

    "f1_maximum":
        "seed_f1_maximum",

    "f1_cv_percent":
        "seed_f1_cv_percent",

    "false_positive_rate_mean":
        "seed_fpr_mean",

    "false_positive_rate_sample_std":
        "seed_fpr_sample_std",

    "roc_auc_mean":
        "seed_roc_auc_mean",

    "roc_auc_sample_std":
        "seed_roc_auc_sample_std",

    "pr_auc_mean":
        "seed_pr_auc_mean",

    "pr_auc_sample_std":
        "seed_pr_auc_sample_std",
}


def attach_robustness_context(
    integrated_df,
    dataset_name,
):

    output = (
        integrated_df.copy()
    )

    # Create empty supplementary columns.
    output[
        "seed_runs"
    ] = np.nan

    for destination in (
        ROBUSTNESS_COLUMNS.values()
    ):
        output[
            destination
        ] = np.nan


    for model_name in [
        "Dense Autoencoder",
        "LSTM Autoencoder",
    ]:

        robustness_row = (
            robustness_summary.loc[
                (
                    robustness_summary[
                        "dataset"
                    ] == dataset_name
                )
                &
                (
                    robustness_summary[
                        "model"
                    ] == model_name
                )
            ]
        )

        if len(
            robustness_row
        ) != 1:

            raise ValueError(
                f"Expected one robustness row for "
                f"{dataset_name} / "
                f"{model_name}"
            )

        robustness_row = (
            robustness_row.iloc[0]
        )

        target_mask = (
            output[
                "method"
            ] == model_name
        )

        if int(
            target_mask.sum()
        ) != 1:

            raise ValueError(
                f"Expected one integrated result "
                f"for {dataset_name} / "
                f"{model_name}"
            )

        output.loc[
            target_mask,
            "seed_runs",
        ] = 5

        for (
            source,
            destination,
        ) in ROBUSTNESS_COLUMNS.items():

            output.loc[
                target_mask,
                destination,
            ] = float(
                robustness_row[
                    source
                ]
            )

    return output


nab_integrated_df = (
    attach_robustness_context(
        nab_integrated_df,
        "NAB",
    )
)

blued_integrated_df = (
    attach_robustness_context(
        blued_integrated_df,
        "BLUED",
    )
)


# ---------------------------------------------------------------------
# 11. Create a thesis-ready neural robustness supplement
# ---------------------------------------------------------------------

neural_supplement_rows = []

for dataset_name, table in [
    (
        "NAB",
        nab_integrated_df,
    ),
    (
        "BLUED",
        blued_integrated_df,
    ),
]:

    for model_name in [
        "Dense Autoencoder",
        "LSTM Autoencoder",
    ]:

        row = table.loc[
            table[
                "method"
            ] == model_name
        ].iloc[0]

        neural_supplement_rows.append({
            "dataset":
                dataset_name,

            "evaluation_semantics":
                row[
                    "evaluation_semantics"
                ],

            "model":
                model_name,

            "authoritative_single_run_f1":
                float(
                    row[
                        "f1_score"
                    ]
                ),

            "five_seed_f1_mean":
                float(
                    row[
                        "seed_f1_mean"
                    ]
                ),

            "five_seed_f1_sample_std":
                float(
                    row[
                        "seed_f1_sample_std"
                    ]
                ),

            "five_seed_f1_minimum":
                float(
                    row[
                        "seed_f1_minimum"
                    ]
                ),

            "five_seed_f1_maximum":
                float(
                    row[
                        "seed_f1_maximum"
                    ]
                ),

            "five_seed_f1_cv_percent":
                float(
                    row[
                        "seed_f1_cv_percent"
                    ]
                ),

            "authoritative_single_run_roc_auc":
                float(
                    row[
                        "roc_auc"
                    ]
                ),

            "five_seed_roc_auc_mean":
                float(
                    row[
                        "seed_roc_auc_mean"
                    ]
                ),

            "five_seed_roc_auc_sample_std":
                float(
                    row[
                        "seed_roc_auc_sample_std"
                    ]
                ),

            "authoritative_single_run_pr_auc":
                float(
                    row[
                        "pr_auc"
                    ]
                ),

            "five_seed_pr_auc_mean":
                float(
                    row[
                        "seed_pr_auc_mean"
                    ]
                ),

            "five_seed_pr_auc_sample_std":
                float(
                    row[
                        "seed_pr_auc_sample_std"
                    ]
                ),

            "interpretation":
                (
                    "Five-seed statistics quantify "
                    "stochastic training variability "
                    "under the frozen current runtime; "
                    "they do not replace the authoritative "
                    "single-run thesis result."
                ),
        })


neural_supplement_df = pd.DataFrame(
    neural_supplement_rows
)


# ---------------------------------------------------------------------
# 12. Create baseline-vs-model comparison rows
# ---------------------------------------------------------------------

baseline_comparison_rows = []

for dataset_name, table in [
    (
        "NAB",
        nab_integrated_df,
    ),
    (
        "BLUED",
        blued_integrated_df,
    ),
]:

    baseline_row = table.loc[
        table[
            "method"
        ] == "Maximum Absolute Z-Score"
    ].iloc[0]

    for _, model_row in (
        table.loc[
            table[
                "method"
            ]
            != "Maximum Absolute Z-Score"
        ]
        .iterrows()
    ):

        baseline_comparison_rows.append({
            "dataset":
                dataset_name,

            "comparison_method":
                model_row[
                    "method"
                ],

            "baseline_f1":
                float(
                    baseline_row[
                        "f1_score"
                    ]
                ),

            "model_f1":
                float(
                    model_row[
                        "f1_score"
                    ]
                ),

            "baseline_minus_model_f1":
                float(
                    baseline_row[
                        "f1_score"
                    ]
                    -
                    model_row[
                        "f1_score"
                    ]
                ),

            "baseline_roc_auc":
                float(
                    baseline_row[
                        "roc_auc"
                    ]
                ),

            "model_roc_auc":
                float(
                    model_row[
                        "roc_auc"
                    ]
                ),

            "baseline_minus_model_roc_auc":
                float(
                    baseline_row[
                        "roc_auc"
                    ]
                    -
                    model_row[
                        "roc_auc"
                    ]
                ),

            "baseline_pr_auc":
                float(
                    baseline_row[
                        "pr_auc"
                    ]
                ),

            "model_pr_auc":
                float(
                    model_row[
                        "pr_auc"
                    ]
                ),

            "baseline_minus_model_pr_auc":
                float(
                    baseline_row[
                        "pr_auc"
                    ]
                    -
                    model_row[
                        "pr_auc"
                    ]
                ),
        })


baseline_comparison_df = pd.DataFrame(
    baseline_comparison_rows
)


# ---------------------------------------------------------------------
# 13. Save final integrated evidence package
# ---------------------------------------------------------------------

FINAL_COMPARISON_ROOT = (
    FINAL_REVISION_ROOT
    / "FINAL_COMPARATIVE_EVIDENCE"
)

FINAL_COMPARISON_ROOT.mkdir(
    parents=True,
    exist_ok=True,
)


NAB_INTEGRATED_PATH = (
    FINAL_COMPARISON_ROOT
    / "nab_six_method_integrated_comparison.csv"
)

BLUED_INTEGRATED_PATH = (
    FINAL_COMPARISON_ROOT
    / "blued_six_method_integrated_comparison.csv"
)

NEURAL_SUPPLEMENT_PATH = (
    FINAL_COMPARISON_ROOT
    / "neural_five_seed_robustness_supplement.csv"
)

BASELINE_COMPARISON_PATH = (
    FINAL_COMPARISON_ROOT
    / "statistical_baseline_vs_existing_models.csv"
)


nab_integrated_df.to_csv(
    NAB_INTEGRATED_PATH,
    index=False,
)

blued_integrated_df.to_csv(
    BLUED_INTEGRATED_PATH,
    index=False,
)

neural_supplement_df.to_csv(
    NEURAL_SUPPLEMENT_PATH,
    index=False,
)

baseline_comparison_df.to_csv(
    BASELINE_COMPARISON_PATH,
    index=False,
)


# ---------------------------------------------------------------------
# 14. Print concise authoritative comparison tables
# ---------------------------------------------------------------------

DISPLAY_COLUMNS = [
    "f1_rank",
    "method",
    "precision",
    "recall",
    "f1_score",
    "false_positive_rate",
    "roc_auc",
    "pr_auc",
    "flagged_ratio_percent",
]


print(
    "=" * 130
)

print(
    "FINAL INTEGRATED COMPARATIVE EVIDENCE"
)

print(
    "=" * 130
)


print(
    "\nNAB — Authoritative results plus frozen statistical baseline"
)

print(
    nab_integrated_df[
        DISPLAY_COLUMNS
    ]
    .round(6)
    .to_string(
        index=False
    )
)


print(
    "\nBLUED — Authoritative event-alignment results "
    "plus frozen statistical baseline"
)

print(
    blued_integrated_df[
        DISPLAY_COLUMNS
    ]
    .round(6)
    .to_string(
        index=False
    )
)


# ---------------------------------------------------------------------
# 15. Print neural robustness supplement
# ---------------------------------------------------------------------

print(
    "\n" + "=" * 130
)

print(
    "NEURAL STOCHASTIC-ROBUSTNESS SUPPLEMENT"
)

print(
    "=" * 130
)

for _, row in (
    neural_supplement_df.iterrows()
):

    print(
        "\n"
        f"{row['dataset']} — "
        f"{row['model']}"
    )

    print(
        f"  Authoritative F1: "
        f"{row['authoritative_single_run_f1']:.6f}"
    )

    print(
        f"  Five-seed F1: "
        f"{row['five_seed_f1_mean']:.6f} "
        f"± "
        f"{row['five_seed_f1_sample_std']:.6f}"
    )

    print(
        f"  Five-seed F1 range: "
        f"[{row['five_seed_f1_minimum']:.6f}, "
        f"{row['five_seed_f1_maximum']:.6f}]"
    )

    print(
        f"  F1 CV: "
        f"{row['five_seed_f1_cv_percent']:.2f}%"
    )

    print(
        f"  Authoritative ROC-AUC: "
        f"{row['authoritative_single_run_roc_auc']:.6f}"
    )

    print(
        f"  Five-seed ROC-AUC: "
        f"{row['five_seed_roc_auc_mean']:.6f} "
        f"± "
        f"{row['five_seed_roc_auc_sample_std']:.6f}"
    )


# ---------------------------------------------------------------------
# 16. Explicit interpretation guardrails
# ---------------------------------------------------------------------

print(
    "\n" + "=" * 130
)

print(
    "INTERPRETIVE GUARDRAILS"
)

print(
    "=" * 130
)

print(
    "- The original corrected/final five-model values remain "
    "the authoritative thesis comparison results."
)

print(
    "- Five-seed Dense/LSTM statistics are supplementary "
    "stochastic-robustness evidence and do not replace those values."
)

print(
    "- The statistical baseline was frozen before any validation "
    "or test performance was observed."
)

print(
    "- No alternative statistical baseline was selected after "
    "viewing the test results."
)

print(
    "- BLUED values measure switching-event alignment, "
    "not confirmed anomaly-detection accuracy."
)

print(
    "- Differences between methods are descriptive; "
    "no statistical significance is claimed."
)


print(
    "\nSaved final comparative-evidence artifacts:"
)

for path in [
    NAB_INTEGRATED_PATH,
    BLUED_INTEGRATED_PATH,
    NEURAL_SUPPLEMENT_PATH,
    BASELINE_COMPARISON_PATH,
]:

    print(
        f"  {path}"
    )


print(
    "\nNo models were trained in Cell 28."
)
```
