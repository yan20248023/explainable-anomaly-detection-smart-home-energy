```python
# Cell 10 - Load final boundary-corrected EQE windows and corrected cross-model results

import numpy as np
import pandas as pd
from pathlib import Path

BOUNDARY_CORRECTION_DIR = (
    CODE_DIR
    / "experiments"
    / "EXP_SPLIT_BOUNDARY_PURGE_CORRECTION"
)

CORRECTED_FULL_SCORE_FILE = (
    BOUNDARY_CORRECTION_DIR
    / "AMPDS2"
    / "ampds2_if_scored_boundary_corrected.csv"
)

CORRECTED_CROSS_MODEL_FILE = (
    BOUNDARY_CORRECTION_DIR
    / "CONTROLLED_RESIDENTIAL"
    / "ampds2"
    / "ampds2_controlled_three_model_scored_corrected.csv"
)

SUPERVISOR_FOLLOWUP_DIR = (
    CODE_DIR
    / "experiments"
    / "EXP_FINAL_SUPERVISOR_FOLLOWUP"
    / "AMPDS2_EQE_DIAGNOSTIC"
)

SUPERVISOR_FOLLOWUP_DIR.mkdir(parents=True, exist_ok=True)

RAW_FEATURES = [
    "energy",
    "first_last_diff",
    "iqr",
    "kurtosis",
    "mad",
    "max",
    "max_abs_diff",
    "mean",
    "mean_abs_diff",
    "median",
    "min",
    "n_points",
    "peak_count",
    "q25",
    "q75",
    "range",
    "rms",
    "skew",
    "slope",
    "std",
    "std_diff",
    "sum",
    "var",
]

STANDARDIZED_FEATURES = [f"{feature}_z" for feature in RAW_FEATURES]

FULL_USE_COLUMNS = [
    "dataset",
    "channel_code",
    "channel_name",
    "primary_category",
    "window_start",
    "window_end",
    "series_key",
    "split",
    *RAW_FEATURES,
    *STANDARDIZED_FEATURES,
    "if_decision_function",
    "anomaly_score",
    "predicted_anomaly",
    "if_threshold",
    "predicted_anomaly_corrected",
    "if_threshold_corrected",
]

assert CORRECTED_FULL_SCORE_FILE.exists(), (
    f"Corrected score file not found: {CORRECTED_FULL_SCORE_FILE}"
)

print(f"Reading corrected EQE rows from:\n{CORRECTED_FULL_SCORE_FILE}\n")

eqe_chunks = []
chunk_size = 100_000

for chunk_number, chunk in enumerate(
    pd.read_csv(
        CORRECTED_FULL_SCORE_FILE,
        usecols=FULL_USE_COLUMNS,
        chunksize=chunk_size,
        low_memory=False,
    ),
    start=1,
):
    channel_codes = chunk["channel_code"].astype(str).str.strip().str.upper()
    eqe_chunk = chunk.loc[channel_codes.eq("EQE")].copy()

    if not eqe_chunk.empty:
        eqe_chunks.append(eqe_chunk)

    if chunk_number == 1 or chunk_number % 2 == 0:
        collected_rows = sum(len(part) for part in eqe_chunks)
        print(
            f"Processed chunk {chunk_number}; "
            f"EQE rows collected so far: {collected_rows:,}"
        )

assert eqe_chunks, "No EQE rows were found in the corrected scored dataset."

eqe_full = pd.concat(eqe_chunks, ignore_index=True)

eqe_full["window_start"] = pd.to_datetime(
    eqe_full["window_start"],
    utc=True,
    errors="coerce",
)

eqe_full["window_end"] = pd.to_datetime(
    eqe_full["window_end"],
    utc=True,
    errors="coerce",
)

eqe_full = eqe_full.sort_values("window_start").reset_index(drop=True)

for flag_column in ["predicted_anomaly", "predicted_anomaly_corrected"]:
    eqe_full[flag_column] = pd.to_numeric(
        eqe_full[flag_column],
        errors="coerce",
    ).fillna(0).astype(int)


assert CORRECTED_CROSS_MODEL_FILE.exists(), (
    f"Corrected cross-model file not found: {CORRECTED_CROSS_MODEL_FILE}"
)

controlled_all = pd.read_csv(
    CORRECTED_CROSS_MODEL_FILE,
    low_memory=False,
)

eqe_controlled = controlled_all.loc[
    controlled_all["channel_code"]
    .astype(str)
    .str.strip()
    .str.upper()
    .eq("EQE")
].copy()

eqe_controlled["window_start"] = pd.to_datetime(
    eqe_controlled["window_start"],
    utc=True,
    errors="coerce",
)

eqe_controlled["window_end"] = pd.to_datetime(
    eqe_controlled["window_end"],
    utc=True,
    errors="coerce",
)


print("\n" + "=" * 100)
print("FINAL CORRECTED EQE DATASET VALIDATION")
print("=" * 100)

print(f"Full EQE rows: {len(eqe_full):,}")
print(
    "Full EQE date range: "
    f"{eqe_full['window_start'].min()} to "
    f"{eqe_full['window_end'].max()}"
)

print("\nRows by split:")
print(eqe_full["split"].value_counts().sort_index().to_string())

print("\nCorrected Isolation Forest flagging by split:")
full_flag_summary = (
    eqe_full.groupby("split")
    .agg(
        windows=("predicted_anomaly_corrected", "size"),
        flagged_windows=("predicted_anomaly_corrected", "sum"),
        flagged_ratio=("predicted_anomaly_corrected", "mean"),
        mean_score=("anomaly_score", "mean"),
        median_score=("anomaly_score", "median"),
        minimum_score=("anomaly_score", "min"),
        maximum_score=("anomaly_score", "max"),
    )
    .reset_index()
)

full_flag_summary["flagged_ratio_percent"] = (
    100 * full_flag_summary["flagged_ratio"]
)

print(full_flag_summary.to_string(index=False))


print("\n" + "=" * 100)
print("CORRECTED CONTROLLED CROSS-MODEL EQE VALIDATION")
print("=" * 100)

print(f"Controlled EQE rows: {len(eqe_controlled):,}")

corrected_model_flags = {
    "Isolation Forest": "if_balanced_predicted_anomaly_corrected",
    "LOF": "lof_predicted_anomaly_corrected",
    "Dense Autoencoder": "ae_predicted_anomaly_corrected",
}

for model_name, flag_column in corrected_model_flags.items():
    assert flag_column in eqe_controlled.columns, (
        f"Missing corrected flag column: {flag_column}"
    )

    eqe_controlled[flag_column] = pd.to_numeric(
        eqe_controlled[flag_column],
        errors="coerce",
    ).fillna(0).astype(int)

    print(f"\n{model_name}:")
    model_summary = (
        eqe_controlled.groupby("split")[flag_column]
        .agg(["size", "sum", "mean"])
        .rename(
            columns={
                "size": "windows",
                "sum": "flagged_windows",
                "mean": "flagged_ratio",
            }
        )
    )

    model_summary["flagged_ratio_percent"] = (
        100 * model_summary["flagged_ratio"]
    )

    print(model_summary.to_string())


print(f"\nDiagnostic output directory:\n{SUPERVISOR_FOLLOWUP_DIR}")
```

```python
# Cell 11 - Rank EQE feature shifts between training, validation, and testing

import numpy as np
import pandas as pd
from IPython.display import display

available_raw_features = [
    feature for feature in RAW_FEATURES
    if feature in eqe_full.columns
]

records = []

for feature in available_raw_features:
    z_feature = f"{feature}_z"

    row = {"feature": feature}

    for split_name in ["train", "val", "test"]:
        split_data = eqe_full.loc[
            eqe_full["split"].eq(split_name), feature
        ]
        split_data = pd.to_numeric(split_data, errors="coerce").dropna()

        row[f"{split_name}_raw_mean"] = split_data.mean()
        row[f"{split_name}_raw_median"] = split_data.median()
        row[f"{split_name}_raw_std"] = split_data.std()
        row[f"{split_name}_raw_q05"] = split_data.quantile(0.05)
        row[f"{split_name}_raw_q95"] = split_data.quantile(0.95)

        if z_feature in eqe_full.columns:
            z_data = eqe_full.loc[
                eqe_full["split"].eq(split_name), z_feature
            ]
            z_data = pd.to_numeric(z_data, errors="coerce").dropna()

            row[f"{split_name}_z_mean"] = z_data.mean()
            row[f"{split_name}_z_median"] = z_data.median()
        else:
            row[f"{split_name}_z_mean"] = np.nan
            row[f"{split_name}_z_median"] = np.nan

    row["test_minus_val_raw_median"] = (
        row["test_raw_median"] - row["val_raw_median"]
    )
    row["test_minus_train_raw_median"] = (
        row["test_raw_median"] - row["train_raw_median"]
    )
    row["test_minus_val_z_median"] = (
        row["test_z_median"] - row["val_z_median"]
    )
    row["test_minus_train_z_median"] = (
        row["test_z_median"] - row["train_z_median"]
    )
    row["absolute_test_val_shift"] = abs(
        row["test_minus_val_z_median"]
    )

    records.append(row)

eqe_feature_shift = pd.DataFrame(records).sort_values(
    "absolute_test_val_shift",
    ascending=False
).reset_index(drop=True)

output_columns = [
    "feature",
    "train_raw_median",
    "val_raw_median",
    "test_raw_median",
    "train_z_median",
    "val_z_median",
    "test_z_median",
    "test_minus_val_z_median",
    "test_minus_train_z_median",
]

print("=" * 100)
print("EQE FEATURES RANKED BY TEST–VALIDATION DISTRIBUTION SHIFT")
print("=" * 100)

display(
    eqe_feature_shift[output_columns]
    .head(15)
    .style.format(precision=4)
)

shift_output_file = (
    SUPERVISOR_FOLLOWUP_DIR /
    "ampds2_eqe_feature_shift_summary.csv"
)

eqe_feature_shift.to_csv(shift_output_file, index=False)

print("\nSaved feature-shift summary to:")
print(shift_output_file)
```

```python
# Cell 12 - Locate the EQE regime shift and quantify its effect on the overall AMPds2 test rate

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from IPython.display import display

# ------------------------------------------------------------------
# Part A: Monthly EQE behaviour
# ------------------------------------------------------------------

eqe_monthly_source = eqe_full.copy()

eqe_monthly_source["month"] = (
    eqe_monthly_source["window_start"]
    .dt.tz_localize(None)
    .dt.to_period("M")
    .astype(str)
)

eqe_monthly = (
    eqe_monthly_source
    .groupby(["month", "split"], as_index=False)
    .agg(
        windows=("anomaly_score", "size"),
        raw_mean_median=("mean", "median"),
        raw_rms_median=("rms", "median"),
        raw_q25_median=("q25", "median"),
        raw_q75_median=("q75", "median"),
        score_mean=("anomaly_score", "mean"),
        score_median=("anomaly_score", "median"),
        flagged_windows=(
            "predicted_anomaly_corrected",
            "sum"
        ),
    )
)

eqe_monthly["flagged_ratio_percent"] = (
    100 * eqe_monthly["flagged_windows"] /
    eqe_monthly["windows"]
)

eqe_monthly["month_date"] = pd.to_datetime(
    eqe_monthly["month"] + "-01"
)

print("=" * 100)
print("MONTHLY EQE POWER LEVEL AND CORRECTED ISOLATION FOREST FLAGGING")
print("=" * 100)

display(
    eqe_monthly[
        [
            "month",
            "split",
            "windows",
            "raw_mean_median",
            "raw_rms_median",
            "score_median",
            "flagged_windows",
            "flagged_ratio_percent",
        ]
    ].style.format(
        {
            "raw_mean_median": "{:.3f}",
            "raw_rms_median": "{:.3f}",
            "score_median": "{:.4f}",
            "flagged_ratio_percent": "{:.2f}",
        }
    )
)

test_months = eqe_monthly[
    eqe_monthly["split"].eq("test")
].copy()

high_shift_months = test_months[
    test_months["flagged_ratio_percent"] >= 50
]

if not high_shift_months.empty:
    first_high_shift_month = high_shift_months.iloc[0]

    print("\nFirst test month with at least 50% flagged windows:")
    print(
        f"{first_high_shift_month['month']} — "
        f"{first_high_shift_month['flagged_ratio_percent']:.2f}% flagged; "
        f"median mean power = "
        f"{first_high_shift_month['raw_mean_median']:.3f} W"
    )
else:
    print("\nNo test month reached a 50% flagged-window ratio.")

# ------------------------------------------------------------------
# Part B: Plot monthly power and anomaly flagging
# ------------------------------------------------------------------

fig, axis_power = plt.subplots(figsize=(13, 6))

split_colours = {
    "train": "#2E86AB",
    "val": "#F39C12",
    "test": "#C0392B",
}

for split_name, split_data in eqe_monthly.groupby("split"):
    axis_power.plot(
        split_data["month_date"],
        split_data["raw_mean_median"],
        marker="o",
        linewidth=2,
        label=f"{split_name.capitalize()} median power",
        color=split_colours.get(split_name),
    )

axis_power.set_xlabel("Month")
axis_power.set_ylabel("Median window mean power (W)")
axis_power.grid(alpha=0.25)

axis_flags = axis_power.twinx()
axis_flags.plot(
    eqe_monthly["month_date"],
    eqe_monthly["flagged_ratio_percent"],
    color="black",
    linestyle="--",
    marker="s",
    linewidth=1.8,
    label="Corrected IF flagged ratio",
)
axis_flags.set_ylabel("Flagged windows (%)")
axis_flags.set_ylim(bottom=0)

handles_1, labels_1 = axis_power.get_legend_handles_labels()
handles_2, labels_2 = axis_flags.get_legend_handles_labels()

axis_power.legend(
    handles_1 + handles_2,
    labels_1 + labels_2,
    loc="best",
)

plt.title(
    "AMPds2 EQE Monthly Power Level and Isolation Forest Flagging"
)
plt.tight_layout()

monthly_figure_file = (
    SUPERVISOR_FOLLOWUP_DIR /
    "ampds2_eqe_monthly_regime_shift.png"
)

plt.savefig(
    monthly_figure_file,
    dpi=300,
    bbox_inches="tight"
)
plt.show()

monthly_table_file = (
    SUPERVISOR_FOLLOWUP_DIR /
    "ampds2_eqe_monthly_regime_shift.csv"
)
eqe_monthly.to_csv(monthly_table_file, index=False)

# ------------------------------------------------------------------
# Part C: Contribution of EQE to the overall AMPds2 test increase
# ------------------------------------------------------------------

corrected_channel_summary_file = (
    BOUNDARY_CORRECTION_DIR /
    "AMPDS2" /
    "ampds2_if_corrected_test_channel_summary.csv"
)

channel_summary = pd.read_csv(
    corrected_channel_summary_file
)

total_test_windows = channel_summary["test_windows"].sum()
total_test_flags = channel_summary["flagged_windows"].sum()

eqe_summary = channel_summary[
    channel_summary["channel_code"].eq("EQE")
].iloc[0]

without_eqe_windows = (
    total_test_windows - eqe_summary["test_windows"]
)
without_eqe_flags = (
    total_test_flags - eqe_summary["flagged_windows"]
)

overall_test_ratio = (
    100 * total_test_flags / total_test_windows
)
without_eqe_ratio = (
    100 * without_eqe_flags / without_eqe_windows
)

validation_reference_rate = 5.0

overall_excess_flags = (
    total_test_flags -
    validation_reference_rate / 100 * total_test_windows
)

eqe_excess_flags = (
    eqe_summary["flagged_windows"] -
    validation_reference_rate / 100 *
    eqe_summary["test_windows"]
)

eqe_share_of_excess = (
    100 * eqe_excess_flags / overall_excess_flags
)

impact_summary = pd.DataFrame(
    {
        "scenario": [
            "All AMPds2 channels",
            "AMPds2 excluding EQE",
            "EQE only",
        ],
        "test_windows": [
            total_test_windows,
            without_eqe_windows,
            eqe_summary["test_windows"],
        ],
        "flagged_windows": [
            total_test_flags,
            without_eqe_flags,
            eqe_summary["flagged_windows"],
        ],
        "flagged_ratio_percent": [
            overall_test_ratio,
            without_eqe_ratio,
            eqe_summary["flagged_ratio_percent"],
        ],
    }
)

print("\n" + "=" * 100)
print("EQE CONTRIBUTION TO THE OVERALL AMPDS2 TEST FLAGGED RATE")
print("=" * 100)

display(
    impact_summary.style.format(
        {"flagged_ratio_percent": "{:.3f}"}
    )
)

print(
    f"\nEQE accounts for approximately "
    f"{eqe_share_of_excess:.2f}% of the test flags above "
    f"the 5% validation-reference rate."
)

print("\nSaved outputs:")
print(monthly_table_file)
print(monthly_figure_file)
```

```python
# Cell 13 - Validate the raw one-minute EQE signal and identify the sustained change onset

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from IPython.display import display

raw_eqe_file = (
    AMPDS2_RAW_DIR /
    "Electricity_EQE.csv"
)

print("Reading raw EQE data from:")
print(raw_eqe_file)

raw_eqe = pd.read_csv(raw_eqe_file)

raw_eqe["timestamp"] = pd.to_datetime(
    raw_eqe["unix_ts"],
    unit="s",
    utc=True,
    errors="coerce"
)

raw_eqe = (
    raw_eqe
    .dropna(subset=["timestamp"])
    .sort_values("timestamp")
    .reset_index(drop=True)
)

measurement_columns = [
    column for column in
    ["V", "I", "f", "DPF", "APF", "P", "Pt", "Q", "Qt", "S", "St"]
    if column in raw_eqe.columns
]

for column in measurement_columns:
    raw_eqe[column] = pd.to_numeric(
        raw_eqe[column],
        errors="coerce"
    )

# ------------------------------------------------------------------
# Part A: Timestamp and missing-data quality checks
# ------------------------------------------------------------------

timestamp_intervals = (
    raw_eqe["timestamp"]
    .drop_duplicates()
    .sort_values()
    .diff()
    .dt.total_seconds()
    .dropna()
)

expected_rows = int(
    (
        raw_eqe["timestamp"].max() -
        raw_eqe["timestamp"].min()
    ).total_seconds() / 60
) + 1

unique_timestamps = raw_eqe["timestamp"].nunique()

quality_summary = pd.DataFrame(
    {
        "check": [
            "Raw rows",
            "Unique timestamps",
            "Duplicate timestamps",
            "Expected one-minute timestamps",
            "Missing one-minute timestamps",
            "Non-60-second intervals",
            "Largest timestamp interval (seconds)",
            "Missing P values",
            "Negative P values",
        ],
        "value": [
            len(raw_eqe),
            unique_timestamps,
            raw_eqe["timestamp"].duplicated().sum(),
            expected_rows,
            max(expected_rows - unique_timestamps, 0),
            int((timestamp_intervals != 60).sum()),
            timestamp_intervals.max(),
            raw_eqe["P"].isna().sum(),
            int((raw_eqe["P"] < 0).sum()),
        ],
    }
)

print("\n" + "=" * 100)
print("RAW EQE DATA-QUALITY CHECK")
print("=" * 100)

display(quality_summary)

# ------------------------------------------------------------------
# Part B: Compare the raw validation and test periods
# ------------------------------------------------------------------

validation_start = eqe_full.loc[
    eqe_full["split"].eq("val"),
    "window_start"
].min()

test_start = eqe_full.loc[
    eqe_full["split"].eq("test"),
    "window_start"
].min()

test_end = eqe_full.loc[
    eqe_full["split"].eq("test"),
    "window_end"
].max()

raw_validation = raw_eqe[
    raw_eqe["timestamp"].between(
        validation_start,
        test_start,
        inclusive="left"
    )
].copy()

raw_test = raw_eqe[
    raw_eqe["timestamp"].between(
        test_start,
        test_end,
        inclusive="both"
    )
].copy()

comparison_records = []

for column in measurement_columns:
    validation_values = raw_validation[column].dropna()
    test_values = raw_test[column].dropna()

    comparison_records.append(
        {
            "measurement": column,
            "validation_mean": validation_values.mean(),
            "validation_median": validation_values.median(),
            "validation_std": validation_values.std(),
            "test_mean": test_values.mean(),
            "test_median": test_values.median(),
            "test_std": test_values.std(),
            "test_minus_validation_median":
                test_values.median() -
                validation_values.median(),
            "test_to_validation_mean_ratio":
                test_values.mean() /
                validation_values.mean()
                if validation_values.mean() != 0
                else np.nan,
        }
    )

raw_period_comparison = pd.DataFrame(comparison_records)

print("\n" + "=" * 100)
print("RAW ELECTRICAL MEASUREMENTS: VALIDATION VERSUS TEST")
print("=" * 100)

display(
    raw_period_comparison.style.format(precision=4)
)

print("\nChronological boundaries:")
print("Validation begins:", validation_start)
print("Test begins:      ", test_start)
print("Test ends:        ", test_end)

# ------------------------------------------------------------------
# Part C: Daily signal and sustained-change identification
# ------------------------------------------------------------------

raw_eqe["date"] = (
    raw_eqe["timestamp"]
    .dt.floor("D")
)

daily_eqe = (
    raw_eqe
    .groupby("date", as_index=False)
    .agg(
        observations=("P", "count"),
        p_mean=("P", "mean"),
        p_median=("P", "median"),
        p_std=("P", "std"),
        voltage_median=("V", "median"),
        current_median=("I", "median"),
        apf_median=("APF", "median"),
    )
)

daily_eqe["p_7day_median"] = (
    daily_eqe["p_median"]
    .rolling(window=7, min_periods=7)
    .median()
)

validation_p_median = raw_validation["P"].median()
test_p_median = raw_test["P"].median()

midpoint_threshold = (
    validation_p_median + test_p_median
) / 2

post_test_daily = daily_eqe[
    daily_eqe["date"] >= test_start.floor("D")
].copy()

first_sustained_change = post_test_daily[
    post_test_daily["p_7day_median"] <
    midpoint_threshold
].head(1)

print("\n" + "=" * 100)
print("RAW EQE CHANGE-POINT INDICATOR")
print("=" * 100)

print(
    f"Validation-period raw P median: "
    f"{validation_p_median:.3f} W"
)
print(
    f"Test-period raw P median: "
    f"{test_p_median:.3f} W"
)
print(
    f"Data-derived midpoint threshold: "
    f"{midpoint_threshold:.3f} W"
)

if not first_sustained_change.empty:
    change_row = first_sustained_change.iloc[0]

    print(
        "First post-test date whose trailing seven-day "
        "median falls below the midpoint:"
    )
    print(
        f"{change_row['date']} — "
        f"{change_row['p_7day_median']:.3f} W"
    )
else:
    print(
        "No sustained seven-day reduction was detected."
    )

# ------------------------------------------------------------------
# Part D: Focused raw-signal plot
# ------------------------------------------------------------------

plot_start = pd.Timestamp(
    "2013-09-01",
    tz="UTC"
)

focused_daily = daily_eqe[
    daily_eqe["date"] >= plot_start
].copy()

fig, axis = plt.subplots(figsize=(13, 5.5))

axis.plot(
    focused_daily["date"],
    focused_daily["p_median"],
    color="#AAB7B8",
    linewidth=1,
    alpha=0.8,
    label="Daily median active power",
)

axis.plot(
    focused_daily["date"],
    focused_daily["p_7day_median"],
    color="#C0392B",
    linewidth=2.2,
    label="Seven-day rolling median",
)

axis.axvline(
    test_start,
    color="#2C3E50",
    linestyle="--",
    linewidth=1.8,
    label="Test-period start",
)

axis.axhline(
    midpoint_threshold,
    color="#2874A6",
    linestyle=":",
    linewidth=1.8,
    label="Validation–test midpoint",
)

axis.set_title(
    "AMPds2 EQE Raw Active-Power Regime Shift"
)
axis.set_xlabel("Date")
axis.set_ylabel("Active power (W)")
axis.grid(alpha=0.25)
axis.legend()
plt.tight_layout()

raw_shift_figure_file = (
    SUPERVISOR_FOLLOWUP_DIR /
    "ampds2_eqe_raw_power_regime_shift.png"
)

plt.savefig(
    raw_shift_figure_file,
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ------------------------------------------------------------------
# Part E: Save diagnostic evidence
# ------------------------------------------------------------------

quality_output_file = (
    SUPERVISOR_FOLLOWUP_DIR /
    "ampds2_eqe_raw_data_quality_summary.csv"
)

comparison_output_file = (
    SUPERVISOR_FOLLOWUP_DIR /
    "ampds2_eqe_raw_validation_test_comparison.csv"
)

daily_output_file = (
    SUPERVISOR_FOLLOWUP_DIR /
    "ampds2_eqe_raw_daily_summary.csv"
)

quality_summary.to_csv(
    quality_output_file,
    index=False
)

raw_period_comparison.to_csv(
    comparison_output_file,
    index=False
)

daily_eqe.to_csv(
    daily_output_file,
    index=False
)

print("\nSaved outputs:")
print(quality_output_file)
print(comparison_output_file)
print(daily_output_file)
print(raw_shift_figure_file)
```

```python
# Cell 14 - Test whether redundant magnitude features amplify the EQE regime shift

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

from IPython.display import display
from sklearn.ensemble import IsolationForest
from sklearn.preprocessing import StandardScaler

# This is a diagnostic EQE-only ablation.
# It does not replace the thesis production model.

feature_sets = {
    "Full 23-feature representation": [
        feature for feature in RAW_FEATURES
        if feature in eqe_full.columns
    ],

    "Compact mixed representation": [
        "mean",
        "std",
        "range",
        "first_last_diff",
        "max_abs_diff",
        "mean_abs_diff",
        "slope",
        "skew",
        "kurtosis",
        "peak_count",
    ],

    "Level only": [
        "mean",
    ],

    "Shape and variability only": [
        "iqr",
        "mad",
        "range",
        "std",
        "std_diff",
        "max_abs_diff",
        "mean_abs_diff",
        "first_last_diff",
        "slope",
        "skew",
        "kurtosis",
        "peak_count",
    ],
}

feature_sets = {
    name: [
        feature for feature in features
        if feature in eqe_full.columns
    ]
    for name, features in feature_sets.items()
}

random_seeds = [42, 123, 2026]
ablation_records = []

for representation_name, features in feature_sets.items():

    modelling_data = eqe_full[
        ["split"] + features
    ].copy()

    modelling_data[features] = (
        modelling_data[features]
        .replace([np.inf, -np.inf], np.nan)
    )

    train_medians = (
        modelling_data.loc[
            modelling_data["split"].eq("train"),
            features
        ]
        .median()
    )

    modelling_data[features] = (
        modelling_data[features]
        .fillna(train_medians)
        .fillna(0)
    )

    split_matrices = {
        split_name: modelling_data.loc[
            modelling_data["split"].eq(split_name),
            features
        ].to_numpy(dtype=float)
        for split_name in ["train", "val", "test"]
    }

    scaler = StandardScaler()

    scaled_matrices = {
        "train": scaler.fit_transform(
            split_matrices["train"]
        ),
        "val": scaler.transform(
            split_matrices["val"]
        ),
        "test": scaler.transform(
            split_matrices["test"]
        ),
    }

    for seed in random_seeds:

        diagnostic_model = IsolationForest(
            n_estimators=200,
            max_samples="auto",
            contamination="auto",
            random_state=seed,
            n_jobs=-1,
        )

        diagnostic_model.fit(
            scaled_matrices["train"]
        )

        anomaly_scores = {
            split_name: -diagnostic_model.decision_function(
                split_matrix
            )
            for split_name, split_matrix
            in scaled_matrices.items()
        }

        validation_threshold = np.quantile(
            anomaly_scores["val"],
            0.95
        )

        for split_name in ["train", "val", "test"]:
            scores = anomaly_scores[split_name]

            predicted_flags = (
                scores >= validation_threshold
            )

            ablation_records.append(
                {
                    "representation":
                        representation_name,
                    "number_of_features":
                        len(features),
                    "random_seed":
                        seed,
                    "split":
                        split_name,
                    "windows":
                        len(scores),
                    "flagged_windows":
                        int(predicted_flags.sum()),
                    "flagged_ratio_percent":
                        100 * predicted_flags.mean(),
                    "score_median":
                        np.median(scores),
                    "validation_threshold":
                        validation_threshold,
                }
            )

ablation_results = pd.DataFrame(
    ablation_records
)

ablation_summary = (
    ablation_results
    .groupby(
        [
            "representation",
            "number_of_features",
            "split",
        ],
        as_index=False
    )
    .agg(
        mean_flagged_ratio_percent=(
            "flagged_ratio_percent",
            "mean"
        ),
        minimum_flagged_ratio_percent=(
            "flagged_ratio_percent",
            "min"
        ),
        maximum_flagged_ratio_percent=(
            "flagged_ratio_percent",
            "max"
        ),
        standard_deviation_percent=(
            "flagged_ratio_percent",
            "std"
        ),
    )
)

test_ablation_summary = (
    ablation_summary[
        ablation_summary["split"].eq("test")
    ]
    .sort_values(
        "mean_flagged_ratio_percent",
        ascending=False
    )
    .reset_index(drop=True)
)

production_test_rate = (
    100 *
    eqe_full.loc[
        eqe_full["split"].eq("test"),
        "predicted_anomaly_corrected"
    ].mean()
)

production_row = pd.DataFrame(
    {
        "representation": [
            "Production corrected IF"
        ],
        "number_of_features": [
            len(feature_sets[
                "Full 23-feature representation"
            ])
        ],
        "split": ["test"],
        "mean_flagged_ratio_percent": [
            production_test_rate
        ],
        "minimum_flagged_ratio_percent": [
            production_test_rate
        ],
        "maximum_flagged_ratio_percent": [
            production_test_rate
        ],
        "standard_deviation_percent": [
            0.0
        ],
    }
)

comparison_table = pd.concat(
    [
        production_row,
        test_ablation_summary,
    ],
    ignore_index=True
)

print("=" * 100)
print("EQE REPRESENTATION-SENSITIVITY RESULTS")
print("=" * 100)

print(
    "All diagnostic models were trained only on EQE "
    "training windows and calibrated at the validation "
    "95th percentile."
)

display(
    comparison_table.style.format(
        {
            "mean_flagged_ratio_percent": "{:.3f}",
            "minimum_flagged_ratio_percent": "{:.3f}",
            "maximum_flagged_ratio_percent": "{:.3f}",
            "standard_deviation_percent": "{:.4f}",
        }
    )
)

# ------------------------------------------------------------------
# Plot the diagnostic test rates
# ------------------------------------------------------------------

plot_data = comparison_table.copy()

error_bars = (
    plot_data["maximum_flagged_ratio_percent"] -
    plot_data["minimum_flagged_ratio_percent"]
) / 2

fig, axis = plt.subplots(figsize=(11, 6))

colours = [
    "#34495E",
    "#C0392B",
    "#D35400",
    "#2874A6",
    "#239B56",
][:len(plot_data)]

axis.barh(
    plot_data["representation"],
    plot_data["mean_flagged_ratio_percent"],
    xerr=error_bars,
    color=colours,
    alpha=0.9,
    capsize=4,
)

axis.axvline(
    5,
    color="black",
    linestyle="--",
    linewidth=1.5,
    label="Validation calibration rate (5%)",
)

axis.set_xlabel("Test windows flagged (%)")
axis.set_ylabel("Feature representation")
axis.set_title(
    "AMPds2 EQE Isolation Forest Representation Sensitivity"
)
axis.grid(
    axis="x",
    alpha=0.25
)
axis.legend()

plt.tight_layout()

ablation_figure_file = (
    SUPERVISOR_FOLLOWUP_DIR /
    "ampds2_eqe_representation_sensitivity.png"
)

plt.savefig(
    ablation_figure_file,
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ------------------------------------------------------------------
# Save all results
# ------------------------------------------------------------------

ablation_seed_file = (
    SUPERVISOR_FOLLOWUP_DIR /
    "ampds2_eqe_ablation_seed_level_results.csv"
)

ablation_summary_file = (
    SUPERVISOR_FOLLOWUP_DIR /
    "ampds2_eqe_ablation_summary.csv"
)

ablation_results.to_csv(
    ablation_seed_file,
    index=False
)

comparison_table.to_csv(
    ablation_summary_file,
    index=False
)

print("\nSaved outputs:")
print(ablation_seed_file)
print(ablation_summary_file)
print(ablation_figure_file)
```

```python
# Cell 16 - Reproduce saved LOF scores using the embedded reference matrices and exact feature order

import numpy as np
import pandas as pd

from IPython.display import display

canonical_lof_features = [
    feature for feature in STANDARDIZED_FEATURES
]

dataset_artifacts = {
    "AMPds2": {
        "reference_file":
            LOF_TABLE_DIR /
            "ampds2_lof_reference_sample.csv",
        "scored_file":
            LOF_TABLE_DIR /
            "ampds2_lof_scored_evaluation.csv",
    },
    "UK-DALE": {
        "reference_file":
            LOF_TABLE_DIR /
            "ukdale_lof_reference_sample.csv",
        "scored_file":
            LOF_TABLE_DIR /
            "ukdale_lof_scored_evaluation.csv",
    },
}

reference_audit_records = []
score_audit_records = []

LOF_REPRODUCTION_DETAILS = {}

for dataset_name, paths in dataset_artifacts.items():

    print("\n" + "=" * 100)
    print(f"{dataset_name}: LOF REPRODUCIBILITY AUDIT")
    print("=" * 100)

    loaded_object = LOADED_LOF_MODELS[
        dataset_name
    ]

    if hasattr(loaded_object, "steps"):
        estimator = loaded_object.steps[-1][1]
    else:
        estimator = loaded_object

    stored_fit_matrix = getattr(
        estimator,
        "_fit_X",
        None
    )

    if hasattr(stored_fit_matrix, "toarray"):
        stored_fit_matrix = (
            stored_fit_matrix.toarray()
        )
    else:
        stored_fit_matrix = np.asarray(
            stored_fit_matrix
        )

    reference_data = pd.read_csv(
        paths["reference_file"]
    )

    scored_data = pd.read_csv(
        paths["scored_file"]
    )

    missing_reference_features = [
        feature
        for feature in canonical_lof_features
        if feature not in reference_data.columns
    ]

    missing_scored_features = [
        feature
        for feature in canonical_lof_features
        if feature not in scored_data.columns
    ]

    if (
        missing_reference_features
        or missing_scored_features
    ):
        print(
            "Missing reference features:",
            missing_reference_features
        )
        print(
            "Missing scored features:",
            missing_scored_features
        )
        continue

    reference_matrix = (
        reference_data[
            canonical_lof_features
        ]
        .replace(
            [np.inf, -np.inf],
            np.nan
        )
        .fillna(0)
        .to_numpy(
            dtype=stored_fit_matrix.dtype
        )
    )

    same_reference_shape = (
        reference_matrix.shape ==
        stored_fit_matrix.shape
    )

    if same_reference_shape:
        reference_absolute_difference = np.abs(
            reference_matrix -
            stored_fit_matrix
        )

        reference_maximum_difference = (
            reference_absolute_difference.max()
        )

        reference_mean_difference = (
            reference_absolute_difference.mean()
        )

        reference_element_close_ratio = (
            np.isclose(
                reference_matrix,
                stored_fit_matrix,
                rtol=1e-6,
                atol=1e-7,
            )
            .mean()
        )

        reference_row_close_ratio = (
            np.isclose(
                reference_matrix,
                stored_fit_matrix,
                rtol=1e-6,
                atol=1e-7,
            )
            .all(axis=1)
            .mean()
        )
    else:
        reference_maximum_difference = np.nan
        reference_mean_difference = np.nan
        reference_element_close_ratio = 0.0
        reference_row_close_ratio = 0.0

    reference_audit_records.append(
        {
            "dataset": dataset_name,
            "saved_fit_shape":
                str(stored_fit_matrix.shape),
            "reference_csv_shape":
                str(reference_matrix.shape),
            "same_shape":
                same_reference_shape,
            "maximum_absolute_difference":
                reference_maximum_difference,
            "mean_absolute_difference":
                reference_mean_difference,
            "close_element_ratio":
                reference_element_close_ratio,
            "close_row_ratio":
                reference_row_close_ratio,
            "canonical_feature_count":
                len(canonical_lof_features),
        }
    )

    # --------------------------------------------------------------
    # Score a deterministic sample spanning the whole evaluation file
    # --------------------------------------------------------------

    sample_size = min(
        1000,
        len(scored_data)
    )

    sample_positions = np.unique(
        np.linspace(
            0,
            len(scored_data) - 1,
            sample_size,
            dtype=int,
        )
    )

    score_sample = (
        scored_data
        .iloc[sample_positions]
        .copy()
    )

    score_matrix = (
        score_sample[
            canonical_lof_features
        ]
        .replace(
            [np.inf, -np.inf],
            np.nan
        )
        .fillna(0)
        .to_numpy(
            dtype=stored_fit_matrix.dtype
        )
    )

    computed_negative_score_samples = (
        -estimator.score_samples(
            score_matrix
        )
    )

    computed_negative_decision_function = (
        -estimator.decision_function(
            score_matrix
        )
    )

    computed_log1p_score = np.log1p(
        np.maximum(
            computed_negative_score_samples,
            0
        )
    )

    stored_anomaly_score = pd.to_numeric(
        score_sample["lof_anomaly_score"],
        errors="coerce",
    ).to_numpy()

    formula_candidates = {
        "-score_samples":
            computed_negative_score_samples,

        "-decision_function":
            computed_negative_decision_function,

        "log1p(-score_samples)":
            computed_log1p_score,
    }

    formula_errors = {}

    for formula_name, formula_values in (
        formula_candidates.items()
    ):
        valid_mask = (
            np.isfinite(stored_anomaly_score)
            &
            np.isfinite(formula_values)
        )

        formula_errors[formula_name] = (
            np.mean(
                np.abs(
                    stored_anomaly_score[valid_mask]
                    -
                    formula_values[valid_mask]
                )
            )
            if valid_mask.any()
            else np.nan
        )

    best_formula = min(
        formula_errors,
        key=lambda key: (
            formula_errors[key]
            if np.isfinite(
                formula_errors[key]
            )
            else np.inf
        ),
    )

    raw_absolute_difference = np.abs(
        stored_anomaly_score -
        computed_negative_score_samples
    )

    raw_close_ratio = np.isclose(
        stored_anomaly_score,
        computed_negative_score_samples,
        rtol=1e-6,
        atol=1e-7,
        equal_nan=True,
    ).mean()

    if "lof_log1p_score" in score_sample.columns:
        stored_log1p_score = pd.to_numeric(
            score_sample[
                "lof_log1p_score"
            ],
            errors="coerce",
        ).to_numpy()

        log1p_absolute_difference = np.abs(
            stored_log1p_score -
            computed_log1p_score
        )

        log1p_maximum_difference = (
            np.nanmax(
                log1p_absolute_difference
            )
        )

        log1p_mean_difference = (
            np.nanmean(
                log1p_absolute_difference
            )
        )

        log1p_close_ratio = np.isclose(
            stored_log1p_score,
            computed_log1p_score,
            rtol=1e-6,
            atol=1e-7,
            equal_nan=True,
        ).mean()
    else:
        stored_log1p_score = np.full(
            len(score_sample),
            np.nan
        )

        log1p_maximum_difference = np.nan
        log1p_mean_difference = np.nan
        log1p_close_ratio = np.nan

    if (
        "lof_threshold" in score_sample.columns
        and
        "lof_predicted_anomaly"
        in score_sample.columns
    ):
        stored_threshold = pd.to_numeric(
            score_sample["lof_threshold"],
            errors="coerce",
        ).to_numpy()

        recomputed_flags = (
            computed_negative_score_samples
            >= stored_threshold
        ).astype(int)

        stored_flags = pd.to_numeric(
            score_sample[
                "lof_predicted_anomaly"
            ],
            errors="coerce",
        ).fillna(-1).astype(int).to_numpy()

        flag_agreement_ratio = (
            recomputed_flags ==
            stored_flags
        ).mean()
    else:
        stored_threshold = np.full(
            len(score_sample),
            np.nan
        )

        recomputed_flags = np.full(
            len(score_sample),
            -1
        )

        stored_flags = np.full(
            len(score_sample),
            -1
        )

        flag_agreement_ratio = np.nan

    score_audit_records.append(
        {
            "dataset": dataset_name,
            "evaluation_rows":
                len(scored_data),
            "sampled_rows":
                len(score_sample),
            "best_matching_formula":
                best_formula,
            "mae_for_negative_score_samples":
                formula_errors[
                    "-score_samples"
                ],
            "mae_for_negative_decision_function":
                formula_errors[
                    "-decision_function"
                ],
            "mae_for_log1p_score":
                formula_errors[
                    "log1p(-score_samples)"
                ],
            "raw_score_maximum_difference":
                np.nanmax(
                    raw_absolute_difference
                ),
            "raw_score_mean_difference":
                np.nanmean(
                    raw_absolute_difference
                ),
            "raw_score_close_ratio":
                raw_close_ratio,
            "log1p_score_maximum_difference":
                log1p_maximum_difference,
            "log1p_score_mean_difference":
                log1p_mean_difference,
            "log1p_score_close_ratio":
                log1p_close_ratio,
            "flag_agreement_ratio":
                flag_agreement_ratio,
        }
    )

    detailed_comparison = score_sample.copy()

    detailed_comparison[
        "reloaded_negative_score_samples"
    ] = computed_negative_score_samples

    detailed_comparison[
        "reloaded_negative_decision_function"
    ] = computed_negative_decision_function

    detailed_comparison[
        "reloaded_log1p_score"
    ] = computed_log1p_score

    detailed_comparison[
        "raw_score_absolute_difference"
    ] = raw_absolute_difference

    detailed_comparison[
        "recomputed_lof_flag"
    ] = recomputed_flags

    LOF_REPRODUCTION_DETAILS[
        dataset_name
    ] = detailed_comparison

    detailed_output_file = (
        LOF_FOLLOWUP_DIR /
        (
            dataset_name
            .lower()
            .replace("-", "")
            .replace(" ", "_")
            +
            "_lof_sample_score_reproduction.csv"
        )
    )

    detailed_comparison.to_csv(
        detailed_output_file,
        index=False
    )

    print(
        f"Reference matrix close-row ratio: "
        f"{reference_row_close_ratio:.6f}"
    )

    print(
        f"Reloaded raw-score close ratio: "
        f"{raw_close_ratio:.6f}"
    )

    print(
        f"Recomputed-flag agreement ratio: "
        f"{flag_agreement_ratio:.6f}"
    )

reference_audit = pd.DataFrame(
    reference_audit_records
)

score_reproduction_audit = pd.DataFrame(
    score_audit_records
)

print("\n" + "=" * 100)
print("REFERENCE-MATRIX REPRODUCTION SUMMARY")
print("=" * 100)

display(
    reference_audit.style.format(
        {
            "maximum_absolute_difference":
                "{:.10f}",
            "mean_absolute_difference":
                "{:.10f}",
            "close_element_ratio":
                "{:.6f}",
            "close_row_ratio":
                "{:.6f}",
        }
    )
)

print("\n" + "=" * 100)
print("LOF SCORE AND FLAG REPRODUCTION SUMMARY")
print("=" * 100)

display(
    score_reproduction_audit.style.format(
        {
            "mae_for_negative_score_samples":
                "{:.10f}",
            "mae_for_negative_decision_function":
                "{:.10f}",
            "mae_for_log1p_score":
                "{:.10f}",
            "raw_score_maximum_difference":
                "{:.10f}",
            "raw_score_mean_difference":
                "{:.10f}",
            "raw_score_close_ratio":
                "{:.6f}",
            "log1p_score_maximum_difference":
                "{:.10f}",
            "log1p_score_mean_difference":
                "{:.10f}",
            "log1p_score_close_ratio":
                "{:.6f}",
            "flag_agreement_ratio":
                "{:.6f}",
        }
    )
)

reference_audit_file = (
    LOF_FOLLOWUP_DIR /
    "lof_reference_matrix_reproduction_summary.csv"
)

score_audit_file = (
    LOF_FOLLOWUP_DIR /
    "lof_score_reproduction_summary.csv"
)

reference_audit.to_csv(
    reference_audit_file,
    index=False
)

score_reproduction_audit.to_csv(
    score_audit_file,
    index=False
)

print("\nSaved summaries:")
print(reference_audit_file)
print(score_audit_file)
```

```python
# Cell 17 - Perform strict full LOF reproduction and isolate numerical dtype effects

import numpy as np
import pandas as pd
import sklearn

from IPython.display import display

def score_lof_in_batches(
    estimator,
    feature_matrix,
    batch_size=1000
):
    score_parts = []

    for batch_start in range(
        0,
        len(feature_matrix),
        batch_size
    ):
        batch_end = min(
            batch_start + batch_size,
            len(feature_matrix)
        )

        batch_scores = -estimator.score_samples(
            feature_matrix[
                batch_start:batch_end
            ]
        )

        score_parts.append(batch_scores)

    return np.concatenate(score_parts)

dtype_selection_records = []
strict_reproduction_records = []

print("Current scikit-learn version:")
print(sklearn.__version__)

for dataset_name, paths in dataset_artifacts.items():

    print("\n" + "=" * 100)
    print(
        f"{dataset_name}: STRICT FULL LOF "
        f"REPRODUCTION"
    )
    print("=" * 100)

    loaded_object = LOADED_LOF_MODELS[
        dataset_name
    ]

    if hasattr(loaded_object, "steps"):
        estimator = loaded_object.steps[-1][1]
    else:
        estimator = loaded_object

    scored_data = pd.read_csv(
        paths["scored_file"]
    )

    base_feature_matrix = (
        scored_data[
            canonical_lof_features
        ]
        .replace(
            [np.inf, -np.inf],
            np.nan
        )
        .fillna(0)
        .to_numpy()
    )

    stored_scores = pd.to_numeric(
        scored_data["lof_anomaly_score"],
        errors="coerce",
    ).to_numpy()

    sample_size = min(
        1000,
        len(scored_data)
    )

    sample_positions = np.unique(
        np.linspace(
            0,
            len(scored_data) - 1,
            sample_size,
            dtype=int,
        )
    )

    dtype_candidates = {
        "float64": np.float64,
        "float32": np.float32,
    }

    candidate_errors = {}

    for dtype_name, dtype_value in (
        dtype_candidates.items()
    ):
        sample_matrix = (
            base_feature_matrix[
                sample_positions
            ]
            .astype(
                dtype_value,
                copy=False
            )
        )

        candidate_scores = (
            -estimator.score_samples(
                sample_matrix
            )
        )

        sampled_stored_scores = (
            stored_scores[
                sample_positions
            ]
        )

        sampled_absolute_difference = np.abs(
            candidate_scores -
            sampled_stored_scores
        )

        sampled_relative_difference = (
            sampled_absolute_difference /
            np.maximum(
                np.abs(
                    sampled_stored_scores
                ),
                1.0
            )
        )

        candidate_mae = np.nanmean(
            sampled_absolute_difference
        )

        candidate_errors[
            dtype_name
        ] = candidate_mae

        dtype_selection_records.append(
            {
                "dataset": dataset_name,
                "query_dtype": dtype_name,
                "sampled_rows":
                    len(sample_positions),
                "mean_absolute_difference":
                    candidate_mae,
                "maximum_absolute_difference":
                    np.nanmax(
                        sampled_absolute_difference
                    ),
                "median_relative_difference":
                    np.nanmedian(
                        sampled_relative_difference
                    ),
                "maximum_relative_difference":
                    np.nanmax(
                        sampled_relative_difference
                    ),
                "exact_score_ratio":
                    np.mean(
                        candidate_scores ==
                        sampled_stored_scores
                    ),
            }
        )

    best_dtype_name = min(
        candidate_errors,
        key=candidate_errors.get
    )

    best_dtype = dtype_candidates[
        best_dtype_name
    ]

    print(
        f"Best matching query dtype: "
        f"{best_dtype_name}"
    )

    print(
        f"Sample MAE using float64: "
        f"{candidate_errors['float64']:.12f}"
    )

    print(
        f"Sample MAE using float32: "
        f"{candidate_errors['float32']:.12f}"
    )

    full_feature_matrix = (
        base_feature_matrix.astype(
            best_dtype,
            copy=False
        )
    )

    reproduced_scores = score_lof_in_batches(
        estimator,
        full_feature_matrix,
        batch_size=1000,
    )

    absolute_difference = np.abs(
        reproduced_scores -
        stored_scores
    )

    relative_difference = (
        absolute_difference /
        np.maximum(
            np.abs(stored_scores),
            1.0
        )
    )

    exact_score_ratio = np.mean(
        reproduced_scores ==
        stored_scores
    )

    strict_absolute_close_ratio = (
        np.isclose(
            reproduced_scores,
            stored_scores,
            rtol=0,
            atol=1e-7,
            equal_nan=True,
        )
        .mean()
    )

    strict_relative_close_ratio = (
        np.isclose(
            reproduced_scores,
            stored_scores,
            rtol=1e-9,
            atol=1e-9,
            equal_nan=True,
        )
        .mean()
    )

    stored_thresholds = pd.to_numeric(
        scored_data["lof_threshold"],
        errors="coerce",
    ).to_numpy()

    stored_flags = pd.to_numeric(
        scored_data["lof_predicted_anomaly"],
        errors="coerce",
    ).fillna(-1).astype(int).to_numpy()

    reproduced_flags = (
        reproduced_scores >=
        stored_thresholds
    ).astype(int)

    flag_mismatch_mask = (
        reproduced_flags !=
        stored_flags
    )

    flag_mismatch_count = int(
        flag_mismatch_mask.sum()
    )

    flag_agreement_ratio = np.mean(
        ~flag_mismatch_mask
    )

    # Inspect duplicate density in the reference sample
    duplicate_summary_file = (
        LOF_TABLE_DIR /
        (
            "ampds2_lof_reference_"
            "duplicate_summary.csv"
            if dataset_name == "AMPds2"
            else
            "ukdale_lof_reference_"
            "duplicate_summary.csv"
        )
    )

    duplicate_summary = pd.read_csv(
        duplicate_summary_file
    )

    total_reference_rows = (
        duplicate_summary[
            "reference_rows"
        ].sum()
    )

    total_duplicate_rows = (
        duplicate_summary[
            "exact_duplicate_rows"
        ].sum()
    )

    weighted_duplicate_ratio = (
        total_duplicate_rows /
        total_reference_rows
        if total_reference_rows > 0
        else np.nan
    )

    strict_reproduction_records.append(
        {
            "dataset": dataset_name,
            "evaluation_rows":
                len(scored_data),
            "model_fit_dtype":
                str(
                    estimator._fit_X.dtype
                ),
            "best_query_dtype":
                best_dtype_name,
            "stored_score_minimum":
                np.nanmin(stored_scores),
            "stored_score_median":
                np.nanmedian(stored_scores),
            "stored_score_q95":
                np.nanquantile(
                    stored_scores,
                    0.95
                ),
            "stored_score_maximum":
                np.nanmax(stored_scores),
            "scores_above_one_million":
                int(
                    (
                        stored_scores >
                        1_000_000
                    ).sum()
                ),
            "mean_absolute_difference":
                np.nanmean(
                    absolute_difference
                ),
            "median_absolute_difference":
                np.nanmedian(
                    absolute_difference
                ),
            "maximum_absolute_difference":
                np.nanmax(
                    absolute_difference
                ),
            "median_relative_difference":
                np.nanmedian(
                    relative_difference
                ),
            "maximum_relative_difference":
                np.nanmax(
                    relative_difference
                ),
            "exact_score_ratio":
                exact_score_ratio,
            "absolute_close_ratio_at_1e_7":
                strict_absolute_close_ratio,
            "relative_close_ratio_at_1e_9":
                strict_relative_close_ratio,
            "flag_mismatch_count":
                flag_mismatch_count,
            "flag_agreement_ratio":
                flag_agreement_ratio,
            "weighted_reference_duplicate_ratio":
                weighted_duplicate_ratio,
        }
    )

    discrepancy_columns = [
        column for column in [
            "dataset",
            "building",
            "meter",
            "channel_code",
            "series_key",
            "split",
            "window_start",
            "window_end",
            "lof_anomaly_score",
            "lof_threshold",
            "lof_predicted_anomaly",
        ]
        if column in scored_data.columns
    ]

    discrepancy_details = (
        scored_data[
            discrepancy_columns
        ]
        .copy()
    )

    discrepancy_details[
        "reproduced_lof_score"
    ] = reproduced_scores

    discrepancy_details[
        "absolute_score_difference"
    ] = absolute_difference

    discrepancy_details[
        "relative_score_difference"
    ] = relative_difference

    discrepancy_details[
        "reproduced_lof_flag"
    ] = reproduced_flags

    discrepancy_details[
        "flag_mismatch"
    ] = flag_mismatch_mask

    dataset_file_label = (
        dataset_name
        .lower()
        .replace("-", "")
        .replace(" ", "_")
    )

    discrepancy_output_file = (
        LOF_FOLLOWUP_DIR /
        (
            dataset_file_label +
            "_lof_full_strict_reproduction.csv"
        )
    )

    discrepancy_details.to_csv(
        discrepancy_output_file,
        index=False
    )

    print(
        f"Full score exact-match ratio: "
        f"{exact_score_ratio:.6f}"
    )

    print(
        f"Maximum relative score difference: "
        f"{np.nanmax(relative_difference):.12f}"
    )

    print(
        f"Flag mismatches: "
        f"{flag_mismatch_count:,}"
    )

dtype_selection_summary = pd.DataFrame(
    dtype_selection_records
)

strict_reproduction_summary = pd.DataFrame(
    strict_reproduction_records
)

print("\n" + "=" * 100)
print("QUERY-DTYPE COMPARISON")
print("=" * 100)

display(
    dtype_selection_summary.style.format(
        {
            "mean_absolute_difference":
                "{:.12f}",
            "maximum_absolute_difference":
                "{:.12f}",
            "median_relative_difference":
                "{:.12e}",
            "maximum_relative_difference":
                "{:.12e}",
            "exact_score_ratio":
                "{:.6f}",
        }
    )
)

print("\n" + "=" * 100)
print("STRICT FULL-DATASET REPRODUCTION SUMMARY")
print("=" * 100)

display(
    strict_reproduction_summary.style.format(
        {
            "stored_score_minimum":
                "{:.6f}",
            "stored_score_median":
                "{:.6f}",
            "stored_score_q95":
                "{:.6f}",
            "stored_score_maximum":
                "{:.6f}",
            "mean_absolute_difference":
                "{:.12f}",
            "median_absolute_difference":
                "{:.12f}",
            "maximum_absolute_difference":
                "{:.12f}",
            "median_relative_difference":
                "{:.12e}",
            "maximum_relative_difference":
                "{:.12e}",
            "exact_score_ratio":
                "{:.6f}",
            "absolute_close_ratio_at_1e_7":
                "{:.6f}",
            "relative_close_ratio_at_1e_9":
                "{:.6f}",
            "flag_agreement_ratio":
                "{:.6f}",
            "weighted_reference_duplicate_ratio":
                "{:.6f}",
        }
    )
)

dtype_output_file = (
    LOF_FOLLOWUP_DIR /
    "lof_query_dtype_comparison.csv"
)

strict_output_file = (
    LOF_FOLLOWUP_DIR /
    "lof_strict_full_reproduction_summary.csv"
)

dtype_selection_summary.to_csv(
    dtype_output_file,
    index=False
)

strict_reproduction_summary.to_csv(
    strict_output_file,
    index=False
)

print("\nSaved final LOF audit files:")
print(dtype_output_file)
print(strict_output_file)
```
